# 🔤 정규표현식(Regular Expression)

문자열에서 특정 패턴을 찾고, 추출하고, 바꾸는 강력한 도구입니다.
웹 스크래핑에서 HTML 태그 사이의 데이터, 이메일, 전화번호 등을
정확하게 추출할 때 필수적입니다.

파이썬의 `re` 모듈은 정규표현식을 다루는 내장 라이브러리입니다.
별도 설치 없이 `import re`로 바로 사용할 수 있습니다.

| 항목 | 내용 |
|------|------|
| **라이브러리** | re (내장) |
| **용도** | 패턴 매칭, 텍스트 검색/치환/분할 |
| **핵심 개념** | 패턴 문자열 → 매칭 → 결과 추출 |

---
## 1. 기본 메타문자

정규표현식은 일반 문자와 **메타문자**(특별한 의미를 가진 문자)로 구성됩니다.

| 메타문자 | 의미 | 예시 | 매칭 예 |
|---------|------|------|--------|
| `.` | 임의의 한 문자 | `a.c` | abc, a1c, a@c |
| `^` | 문자열 시작 | `^Hello` | Hello로 시작 |
| `$` | 문자열 끝 | `end$` | end로 끝남 |
| `*` | 0회 이상 반복 | `ab*c` | ac, abc, abbc |
| `+` | 1회 이상 반복 | `ab+c` | abc, abbc (ac X) |
| `?` | 0회 또는 1회 | `colou?r` | color, colour |
| `{n}` | 정확히 n회 | `a{3}` | aaa |
| `{n,m}` | n~m회 | `a{2,4}` | aa, aaa, aaaa |
| `\|` | OR | `cat\|dog` | cat 또는 dog |

```python
import re

# . (임의의 한 문자)
print('--- . 임의의 한 문자 ---')
pattern = r'a.c'
for text in ['abc', 'a1c', 'ac', 'abbc']:
    result = re.search(pattern, text)
    print(f'  "{text}" → {"매칭" if result else "불일치"}: {result.group() if result else "-"}')

# * (0회 이상) vs + (1회 이상)
print('\n--- * vs + ---')
for pat, desc in [(r'ab*c', '* (0회 이상)'), (r'ab+c', '+ (1회 이상)')]:
    for text in ['ac', 'abc', 'abbc']:
        m = re.search(pat, text)
        print(f'  {desc} "{text}" → {m.group() if m else "불일치"}')

# ? (0회 또는 1회)
print('\n--- ? 0회 또는 1회 ---')
for text in ['color', 'colour']:
    m = re.search(r'colou?r', text)
    print(f'  "{text}" → {m.group() if m else "불일치"}')
```

## 2. 문자 클래스 — [ ] 와 \d, \w, \s

특정 문자 집합 중 하나를 매칭합니다.

| 패턴 | 의미 | 매칭 범위 |
|------|------|----------|
| `[abc]` | a, b, c 중 하나 | 지정된 문자 |
| `[a-z]` | 소문자 a~z | 소문자 범위 |
| `[0-9]` | 숫자 0~9 | 숫자 범위 |
| `[^0-9]` | 숫자가 아닌 것 | 부정(NOT) |
| `\d` | 숫자 | `[0-9]`와 동일 |
| `\D` | 숫자가 아닌 것 | `[^0-9]`와 동일 |
| `\w` | 단어 문자 | `[a-zA-Z0-9_]` |
| `\W` | 단어 문자가 아닌 것 | `[^a-zA-Z0-9_]` |
| `\s` | 공백 문자 | 스페이스, 탭, 줄바꿈 |
| `\S` | 공백이 아닌 문자 | `[^\s]`와 동일 |

```python
import re

# [ ] 문자 집합
print('--- [ ] 문자 집합 ---')
for text in ['cat', 'bat', 'rat', 'sat', 'Cat']:
    m = re.search(r'[cbr]at', text)
    print(f'  "{text}" → {m.group() if m else "불일치"}')

# \d, \w, \s 단축 클래스
print('\n--- 단축 클래스 ---')
text = '이름: 김철수, 나이: 25, 이메일: kim@mail.com'
print(f'  숫자만: {re.findall(r"\\d+", text)}')
print(f'  단어문자: {re.findall(r"\\w+", text)}')

# [^] 부정 클래스
print('\n--- [^] 부정 클래스 ---')
text = 'ABC123def456'
print(f'  숫자가 아닌 것: {re.findall(r"[^0-9]+", text)}')
print(f'  대문자가 아닌 것: {re.findall(r"[^A-Z]+", text)}')

# 한글 패턴
text = 'Hello 안녕하세요 123 파이썬'
hangul = re.findall(r'[가-힣]+', text)
print(f'  한글만: {hangul}')
```

## 3. 그룹핑과 캡처 — ( )

괄호 `()`로 패턴의 일부를 그룹으로 묶으면,
매칭 결과에서 해당 부분만 따로 추출할 수 있습니다.

- `match.group()` → 전체 매칭 결과
- `match.group(1)` → 첫 번째 괄호 그룹
- `match.group(2)` → 두 번째 괄호 그룹
- `match.groups()` → 모든 그룹의 튜플

```python
import re

# 그룹 캡처 기본
text = '2024-01-15'
m = re.search(r'(\d{4})-(\d{2})-(\d{2})', text)
print(f'전체 매칭: {m.group()}')
print(f'연도: {m.group(1)}')
print(f'월: {m.group(2)}')
print(f'일: {m.group(3)}')
print(f'모든 그룹: {m.groups()}')

# 명명된 그룹 (?P<name>pattern)
text = '홍길동 <hong@example.com>'
m = re.search(r'(?P<name>\S+)\s+<(?P<email>\S+)>', text)
print(f'이름: {m.group("name")}')
print(f'이메일: {m.group("email")}')
print(f'딕셔너리: {m.groupdict()}')

# 비캡처 그룹 (?:pattern)
text = 'abcabcabc'
m = re.search(r'(?:abc)+', text)
print(f'비캡처 그룹: {m.group()}')
```

## 4. 주요 함수 — search, match, findall, finditer, sub, split

| 함수 | 용도 | 반환값 |
|------|------|--------|
| `re.search()` | 문자열 어디서든 첫 매칭 | Match 객체 또는 None |
| `re.match()` | 문자열 **시작**에서만 매칭 | Match 객체 또는 None |
| `re.findall()` | 모든 매칭을 문자열 리스트로 | 리스트 |
| `re.finditer()` | 모든 매칭을 Match 객체로 | 이터레이터 |
| `re.sub()` | 패턴 매칭 부분 치환 | 치환된 문자열 |
| `re.split()` | 패턴을 기준으로 분할 | 리스트 |

```python
import re

text = 'apple: 100원, banana: 200원, cherry: 150원'

# re.search — 첫 번째 매칭
m = re.search(r'(\w+):\s*(\d+)원', text)
print(f'search: {m.groups()}')

# re.findall — 모든 매칭
results = re.findall(r'(\w+):\s*(\d+)원', text)
print(f'findall: {results}')

# re.finditer — Match 객체 이터레이터
print('finditer:')
for m in re.finditer(r'(\w+):\s*(\d+)원', text):
    print(f'  {m.group(1)} = {m.group(2)}원 (위치: {m.start()}-{m.end()})')

# re.sub — 치환
result = re.sub(r'(\d+)원', r'\1개', text)
print(f'sub: {result}')

# re.split — 분할
parts = re.split(r'[,\s]+', 'apple, banana  cherry')
print(f'split: {parts}')

# re.match vs re.search
print(f'match("world"): {re.match(r"world", "hello world")}')  # None
print(f'search("world"): {re.search(r"world", "hello world").group()}')
```

## 5. 자주 쓰는 패턴 연습

웹 스크래핑과 데이터 처리에서 자주 사용하는 정규표현식 패턴들입니다.

```python
import re

# 이메일 패턴
email_text = '연락처: hong@test.com, admin@site.co.kr 입니다'
emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.-]+', email_text)
print(f'이메일: {emails}')

# 전화번호 패턴
phone_text = '전화: 010-1234-5678, 02-987-6543, 010.5555.6666'
phones = re.findall(r'\d{2,3}[-.]\d{3,4}[-.]\d{4}', phone_text)
print(f'전화번호: {phones}')

# URL 패턴
url_text = '방문: https://example.com/page?id=1 및 http://test.org/path'
urls = re.findall(r'https?://[\w./?=&-]+', url_text)
print(f'URL: {urls}')

# HTML 태그 제거
html = '<p>안녕하세요</p><br/><strong>파이썬</strong>'
clean = re.sub(r'<[^>]+>', '', html)
print(f'HTML 제거: "{clean}"')

# HTML 속성값 추출
tag = '<a href="https://example.com" class="link" id="main">'
href = re.search(r'href="([^"]+)"', tag)
print(f'href 속성: {href.group(1) if href else "없음"}')

# 날짜 형식 통일
dates = '2024/01/15, 2024-02-28, 2024.03.10'
unified = re.sub(r'(\d{4})[/\-.](\d{2})[/\-.](\d{2})', r'\1-\2-\3', dates)
print(f'날짜 통일: {unified}')
```

## 6. 탐욕적 vs 비탐욕적 매칭

`*`과 `+`는 기본적으로 **탐욕적(greedy)**으로 가능한 한 많이 매칭합니다.
`*?`와 `+?`를 사용하면 **비탐욕적(lazy)**으로 가능한 한 적게 매칭합니다.

HTML 태그 추출 등에서 이 차이가 중요합니다.

```python
import re

html = '<div><p>안녕</p><p>반갑</p></div>'

# 탐욕적 매칭 — 가장 긴 매칭
greedy = re.search(r'<p>.*</p>', html)
print(f'탐욕적: {greedy.group()}')

# 비탐욕적 매칭 — 가장 짧은 매칭
lazy = re.findall(r'<p>.*?</p>', html)
print(f'비탐욕적: {lazy}')

# 실전: 링크 텍스트 추출
page = '<a href="p1.html">첫 번째</a> 텍스트 <a href="p2.html">두 번째</a>'
links = re.findall(r'<a[^>]*>(.*?)</a>', page)
print(f'링크 텍스트: {links}')

# 실전: 큰따옴표 안의 내용 추출
text = 'name="김철수" age="25" city="서울"'
values = re.findall(r'"([^"]+)"', text)
print(f'속성값: {values}')
```

## 7. 전방/후방 탐색과 컴파일

- `(?=pattern)` — 전방 탐색: 뒤에 pattern이 오는 위치
- `(?!pattern)` — 부정 전방 탐색: 뒤에 pattern이 오지 않는 위치
- `(?<=pattern)` — 후방 탐색: 앞에 pattern이 오는 위치

같은 패턴을 반복 사용할 때 `re.compile()`로 미리 컴파일하면 성능이 향상됩니다.

```python
import re

# 전방 탐색 — '원' 앞의 숫자만 추출
text = '사과 3000원, 배 5000원, 감 1500원'
prices = re.findall(r'\d+(?=원)', text)
print(f'가격만: {prices}')

# 후방 탐색 — '가격:' 뒤의 숫자
text = '총 가격: 25000원, 할인: 3000원'
amounts = re.findall(r'(?<=: )\d+', text)
print(f'금액: {amounts}')

# URL에서 도메인만 추출
urls = ['https://www.example.com/page', 'http://blog.test.org/post/1']
for url in urls:
    m = re.search(r'(?<=https?://)([\w.-]+)', url)
    print(f'  {url} → {m.group(1) if m else "-"}')

# re.compile — 패턴 미리 컴파일
email_pat = re.compile(r'[\w.+-]+@[\w-]+\.[\w.-]+')
texts = ['문의: info@example.com', '연락처: admin@site.co.kr']
for t in texts:
    m = email_pat.search(t)
    print(f'  {t} → {m.group() if m else "없음"}')

# 플래그: re.IGNORECASE, re.MULTILINE, re.DOTALL
text = 'Python PYTHON python'
print(f'대소문자 무시: {re.findall(r"python", text, re.IGNORECASE)}')

multi = '첫 줄\n두 번째 줄\n세 번째 줄'
print(f'줄 시작: {re.findall(r"^[가-힣]+", multi, re.MULTILINE)}')

html = '<div>첫 줄\n둘째 줄</div>'
print(f'DOTALL: {re.search(r"<div>.*</div>", html, re.DOTALL).group()}')
```

## 🎯 정리

| 기능 | 패턴 | 설명 |
|------|-------|------|
| 임의 문자 | `.` | 줄바꿈 제외 모든 문자 |
| 반복 | `*`, `+`, `?`, `{n,m}` | 0회이상, 1회이상, 0~1회, n~m회 |
| 문자 클래스 | `[a-z]`, `\d`, `\w`, `\s` | 문자 집합 |
| 그룹 캡처 | `(pattern)` | 매칭 결과에서 부분 추출 |
| 명명 그룹 | `(?P<name>pattern)` | 이름으로 그룹 참조 |
| 비캡처 | `(?:pattern)` | 그룹으로 묶되 캡처하지 않음 |
| 전방 탐색 | `(?=pattern)` | 뒤에 패턴이 오는 위치 |
| 후방 탐색 | `(?<=pattern)` | 앞에 패턴이 오는 위치 |
| 탐욕/비탐욕 | `.*` vs `.*?` | 최대/최소 매칭 |
| 한글 | `[가-힣]` | 한글 음절 범위 |
| 컴파일 | `re.compile()` | 반복 사용 시 성능 향상 |
| 플래그 | `re.I`, `re.M`, `re.S` | 대소문자무시, 여러줄, DOTALL |
