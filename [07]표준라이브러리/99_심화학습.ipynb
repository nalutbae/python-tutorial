## 문제 3: datetime — 시간대 변환

서울 시간(KST, UTC+9) `2025-01-15 14:30:00`을 뉴욕 시간(EST, UTC-5)으로 변환하세요.
`zoneinfo.ZoneInfo`를 사용하세요.

````python
# 문제 3: 답을 작성하세요
from datetime import datetime
from zoneinfo import ZoneInfo
# 서울 → 뉴욕 시간 변환
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from datetime import datetime
from zoneinfo import ZoneInfo
seoul_time = datetime(2025, 1, 15, 14, 30, tzinfo=ZoneInfo('Asia/Seoul'))
ny_time = seoul_time.astimezone(ZoneInfo('America/New_York'))
print(f'서울: {seoul_time}')  # 2025-01-15 14:30:00+09:00
print(f'뉴욕: {ny_time}')    # 2025-01-15 00:30:00-05:00
```

## 문제 4: 타입 힌트 — 제네릭과 Callable

다음 함수에 적절한 타입 힌트를 추가하세요.
- `apply`: 함수 `f`와 값 `x`를 받아 `f(x)`를 반환
- `f`는 `int`를 받아 `str`을 반환하는 함수

```python
def apply(f, x):
    return f(x)
```

````python
# 문제 4: 답을 작성하세요
from typing import Callable
# apply 함수에 타입 힌트 추가
def apply(f, x):
    return f(x)

# 테스트
print(apply(str, 42))  # '42'
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from typing import Callable
def apply(f: Callable[[int], str], x: int) -> str:
    return f(x)
print(apply(str, 42))  # '42'
```

</details>

## 문제 5: hashlib — 파일 무결성 검사

문자열의 SHA-256 해시값을 계산하고, 처음 16자리만 출력하세요.
또한 두 문자열의 해시값이 다르면 `'위변조 감지'`를 출력하세요.

````python
# 문제 5: 답을 작성하세요
import hashlib
original = 'Hello, World!'
modified = 'Hello, World?'
# SHA-256 해시 계산 및 비교
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import hashlib
original = 'Hello, World!'
modified = 'Hello, World?'
h1 = hashlib.sha256(original.encode()).hexdigest()
h2 = hashlib.sha256(modified.encode()).hexdigest()
print(f'원본 해시: {h1[:16]}...')
print(f'수정 해시: {h2[:16]}...')
if h1 != h2:
    print('위변조 감지')
```

</details>

## 문제 6: pickle — 객체 직렬화와 복원

딕셔너리를 `pickle`로 직렬화하여 바이트로 만들고, 다시 복원하세요.
또한 원본과 복원 객체가 같은지(`==`)와 같은 객체인지(`is`)를 확인하세요.

````python
# 문제 6: 답을 작성하세요
import pickle
data = {'name': 'Kim', 'scores': [90, 85, 92], 'active': True}
# pickle 직렬화 + 복원 + 비교
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>


## 문제 9: contextlib — suppress로 예외 무시

`contextlib.suppress`를 사용해 `KeyError`와 `ValueError`를 안전하게 무시하는 코드를 작성하세요.
딕셔너리에서 없는 키를 삭제하려 할 때 발생하는 `KeyError`를 무시하세요.

````python
# 문제 9: 답을 작성하세요
from contextlib import suppress
d = {'a': 1, 'b': 2}
# suppress로 KeyError 무시하며 존재하지 않는 키 삭제
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from contextlib import suppress
d = {'a': 1, 'b': 2}
with suppress(KeyError):
    del d['c']  # 키가 없어도 에러 없이 통과
with suppress(KeyError, ValueError):
    del d['z']  # 여러 예외도 가능
print(d)  # {'a': 1, 'b': 2}
```

</details>

`zipfile` 모듈을 사용해 ZIP 파일을 메모리에서 생성하고, 파일 목록과 내용을 출력하세요.
1. `ZipFile`에 `test.txt`와 `data.txt`를 추가
2. 파일 목록 출력
3. `test.txt`의 내용 읽기

````python
# 문제 10: 답을 작성하세요
import zipfile, io
# 메모리에 ZIP 생성 + 파일 목록 + 내용 읽기
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import zipfile, io
buf = io.BytesIO()
with zipfile.ZipFile(buf, 'w', zipfile.ZIP_DEFLATED) as zf:
    zf.writestr('test.txt', 'Hello from test')
    zf.writestr('data.txt', 'Hello from data')
# 읽기
with zipfile.ZipFile(buf, 'r') as zf:
    print('파일 목록:', zf.namelist())
    print('test.txt 내용:', zf.read('test.txt').decode())
```

</details>
