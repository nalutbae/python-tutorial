# 📘 JSON과 Pickle

**JSON**은 데이터 교환 포맷, **Pickle**은 파이썬 객체 직렬화에 사용합니다.

**학습 목표:**
- json: 데이터 직렬화/역직렬화
- pickle: 파이썬 객체 직렬화
- 커스텀 직렬화와 안전성

## 1. JSON — 데이터 직렬화

JSON은 웹 API, 설정 파일 등에서 널리 사용되는 데이터 포맷입니다.

```python
import json

# ┌─────────────────────────────────────────┐
# │  json 주요 함수                            │
# │  json.dumps()  → 파이썬 객체 → JSON 문자열  │
# │  json.loads()  → JSON 문자열 → 파이썬 객체  │
# │  json.dump()   → 파이썬 객체 → 파일         │
# │  json.load()   → 파일 → 파이썬 객체         │
# └─────────────────────────────────────────┘

# 파이썬 객체 → JSON 문자열
data = {
    "이름": "김파이",
    "나이": 25,
    "기술": ["Python", "SQL", "Docker"],
    "활성": True,
    "점수": None
}

json_str = json.dumps(data, ensure_ascii=False, indent=2)
print("JSON 문자열:")
print(json_str)

# JSON 문자열 → 파이썬 객체
parsed = json.loads(json_str)
print(f"\n이름: {parsed['이름']}")
print(f"기술: {parsed['기술']}")
print(f"활성: {parsed['활성']}")
```

```python
import json
import tempfile
from pathlib import Path

# JSON 파일 쓰기/읽기
tmp = Path(tempfile.mkdtemp(prefix="json_demo_"))
data_file = tmp / "data.json"

# 파일에 쓰기
with open(data_file, 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
print(f"파일 저장: {data_file}")

# 파일에서 읽기
with open(data_file, 'r', encoding='utf-8') as f:
    loaded = json.load(f)
print(f"불러온 데이터: {loaded}")

# 커스텀 직렬화
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def to_dict(self):
        return {"name": self.name, "age": self.age}

person = Person("이코딩", 30)
json_str = json.dumps(person.to_dict(), ensure_ascii=False)
print(f"\n커스텀 직렬화: {json_str}")

# 정리
import shutil
shutil.rmtree(tmp, ignore_errors=True)
```

## 2. Pickle — 파이썬 객체 직렬화

Pickle은 파이썬 객체를 바이너리로 직렬화합니다.
JSON보다 더 많은 타입을 지원하지만 파이썬 전용입니다.

```python
import pickle
import tempfile
from pathlib import Path

# ┌─────────────────────────────────────────┐
# │  pickle vs json                           │
# │  pickle: 바이너리, 파이썬 전용, 모든 객체   │
# │  json:    텍스트, 언어 독립, 기본 타입만     │
# │  ⚠️ pickle은 신뢰할 수 없는 출처에서 로드 금지 │
# └─────────────────────────────────────────┘

# 파이썬 객체 직렬화
class Config:
    def __init__(self, name, settings):
        self.name = name
        self.settings = settings

    def __repr__(self):
        return f"Config('{self.name}', {self.settings})"

config = Config("myapp", {"debug": True, "port": 8080})

# 바이트로 직렬화
pickled = pickle.dumps(config)
print(f"직렬화 크기: {len(pickled)} 바이트")

# 역직렬화
restored = pickle.loads(pickled)
print(f"복원된 객체: {restored}")
print(f"이름: {restored.name}, 설정: {restored.settings}")

# 파일에 저장/불러오기
tmp = Path(tempfile.mkdtemp(prefix="pickle_demo_"))
pkl_file = tmp / "config.pkl"

with open(pkl_file, 'wb') as f:
    pickle.dump(config, f)

with open(pkl_file, 'rb') as f:
    loaded = pickle.load(f)
print(f"\n파일에서 복원: {loaded}")

# 정리
import shutil
shutil.rmtree(tmp, ignore_errors=True)
```
