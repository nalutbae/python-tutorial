# 📘 URL과 웹

**urllib**은 URL 처리와 HTTP 요청을, **webbrowser**는 웹 브라우저 제어를
제공하는 파이썬 표준 라이브러리 모듈입니다.

**학습 목표:**
- urllib: URL 파싱, 인코딩, HTTP 요청
- webbrowser: 브라우저 제어
- URL 빌더 패턴

## 1. urllib — URL 파싱과 인코딩

`urllib.parse`는 URL을 분해하고 조립합니다.
`urllib.request`는 HTTP 요청을 보냅니다.

```python
from urllib.parse import urlparse, urlunparse, urlencode, parse_qs, quote, unquote

# ┌─────────────────────────────────────────┐
# │  urllib.parse 주요 함수                    │
# │  urlparse()   → URL 분해                   │
# │  urlunparse() → URL 조립                   │
# │  urlencode()  → 쿼리 문자열 인코딩           │
# │  parse_qs()   → 쿼리 문자열 파싱            │
# │  quote()      → URL 인코딩                  │
# └─────────────────────────────────────────┘

# URL 파싱
url = "https://example.com:8080/path/to/page?name=김파이&age=25#section"
parsed = urlparse(url)
print(f"스킴: {parsed.scheme}")
print(f"호스트: {parsed.hostname}")
print(f"포트: {parsed.port}")
print(f"경로: {parsed.path}")
print(f"쿼리: {parsed.query}")
print(f"프래그먼트: {parsed.fragment}")

# 쿼리 문자열 파싱
params = parse_qs(parsed.query)
print(f"\n파싱된 쿼리: {params}")

# URL 인코딩 (한글 등)
encoded = quote("김파이")
print(f"\n인코딩: 김파이 → {encoded}")
print(f"디코딩: {encoded} → {unquote(encoded)}")
```

```python
from urllib.parse import urlencode, urlunparse, urlparse

# ┌─────────────────────────────────────────┐
# │  URL 빌더 패턴                             │
# │  쿼리 파라미터를 dict로 구성                │
# │  urlencode()로 인코딩 후 URL 조립           │
# └─────────────────────────────────────────┘

# 쿼리 문자열 생성
params = {
    "q": "파이썬 튜토리얼",
    "lang": "ko",
    "page": 1
}
query_string = urlencode(params)
print(f"쿼리 문자열: {query_string}")

# URL 조립
base_url = urlunparse((
    "https",           # scheme
    "search.example.com", # netloc
    "/search",         # path
    "",                # params
    query_string,      # query
    ""                 # fragment
))
print(f"완성된 URL: {base_url}")

# URL 조립 (f-string 방식)
api_url = f"https://api.example.com/users?{urlencode({'id': 42, 'format': 'json'})}"
print(f"API URL: {api_url}")
```

## 2. webbrowser — 브라우저 제어

`webbrowser`는 기본 웹 브라우저를 열거나 URL을 엽니다.

```python
import webbrowser

# ┌─────────────────────────────────────────┐
# │  webbrowser 주요 함수                     │
# │  webbrowser.open(url)       → 기본 브라우저  │
# │  webbrowser.open_new(url)   → 새 창         │
# │  webbrowser.open_new_tab(url) → 새 탭      │
# │  webbrowser.get(name)       → 특정 브라우저│
# └─────────────────────────────────────────┘

# 브라우저 정보
print(f"기본 브라우저 타입: {webbrowser.get().__class__.__name__}")

# URL 열기 (실제로 열리므로 주석 처리)
# webbrowser.open("https://docs.python.org/3/")
# webbrowser.open_new_tab("https://www.google.com")

# 검색 URL 생성 함수
def search_url(query, engine="google"):
    from urllib.parse import urlencode, quote
    engines = {
        "google": "https://www.google.com/search",
        "duckduckgo": "https://duckduckgo.com/",
        "naver": "https://search.naver.com/search.naver"
    }
    base = engines.get(engine, engines["google"])
    return f"{base}?{urlencode({'q': query})}"

# URL 생성 (열지 않음)
url = search_url("파이썬 튜토리얼", "google")
print(f"\n검색 URL: {url}")

url2 = search_url("Python tutorial", "duckduckgo")
print(f"DuckDuckGo URL: {url2}")
```
