# 📘 시간과 날짜

파이썬은 **time**, **datetime** 모듈로 시간과 날짜를 처리합니다.

**학습 목표:**
- time: 타임스탬프, 시간 측정, 포매팅
- datetime: date, time, datetime, timedelta
- 시간대 처리와 실용 패턴

## 1. time 모듈 — 타임스탬프와 측정

`time` 모듈은 유닉스 타임스탬프와 시간 측정을 제공합니다.

```python
import time

# ┌─────────────────────────────────────────┐
# │  time 모듈 주요 기능                      │
# │  time.time()       → 유닉스 타임스탬프      │
# │  time.sleep()      → 대기                   │
# │  time.strftime()   → 시간 포맷팅            │
# │  time.strptime()   → 문자열 파싱            │
# │  time.perf_counter() → 고해상도 타이머    │
# └─────────────────────────────────────────┘

# 현재 타임스탬프
ts = time.time()
print(f"유닉스 타임스탬프: {ts}")
print(f"정수 부분: {int(ts)}")

# struct_time (로컬 시간)
local = time.localtime(ts)
print(f"\nstruct_time:")
print(f"  연: {local.tm_year}")
print(f"  월: {local.tm_mon}")
print(f"  일: {local.tm_mday}")
print(f"  시: {local.tm_hour}")
print(f"  분: {local.tm_min}")
print(f"  초: {local.tm_sec}")

# 포매팅
formatted = time.strftime("%Y-%m-%d %H:%M:%S", local)
print(f"\n포맷된 시간: {formatted}")

# 시간 측정
start = time.perf_counter()
total = sum(range(1000000))
elapsed = time.perf_counter() - start
print(f"\n합계: {total}, 소요: {elapsed:.6f}초")
```

## 2. datetime 모듈 — 날짜와 시간

`datetime`은 날짜와 시간을 다루는 풍부한 기능을 제공합니다.

```python
from datetime import date, time, datetime, timedelta

# ┌─────────────────────────────────────────┐
# │  datetime 주요 클래스                     │
# │  date(y,m,d)     → 날짜                    │
# │  time(h,m,s)     → 시간                    │
# │  datetime(y,m,d,h,m,s) → 날짜+시간        │
# │  timedelta(days=1) → 기간                  │
# └─────────────────────────────────────────┘

# date 객체
today = date.today()
print(f"오늘: {today}")
print(f"년: {today.year}, 월: {today.month}, 일: {today.day}")
print(f"요일: {today.strftime('%A')} (0=월)")

# 특정 날짜
birthday = date(2025, 12, 25)
print(f"\n크리스마스: {birthday}")

# datetime 객체
now = datetime.now()
print(f"\n현재: {now}")
print(f"포맷: {now.strftime('%Y-%m-%d %H:%M:%S')}")

# 문자열 → datetime
parsed = datetime.strptime("2025-07-31 14:30", "%Y-%m-%d %H:%M")
print(f"파싱: {parsed}")

# f-string 포매팅
print(f"\nf-string: {now:%Y년 %m월 %d일 %H시 %M분}")
```

```python
from datetime import date, datetime, timedelta

# ┌─────────────────────────────────────────┐
# │  timedelta — 날짜/시간 연산                │
# │  날짜 간 차이 계산, 날짜 이동               │
# └─────────────────────────────────────────┘

today = date.today()
print(f"오늘: {today}")

# 날짜 이동
tomorrow = today + timedelta(days=1)
next_week = today + timedelta(weeks=1)
next_month = today + timedelta(days=30)
print(f"내일: {tomorrow}")
print(f"다음 주: {next_week}")
print(f"30일 후: {next_month}")

# 날짜 차이
birthday = date(2025, 12, 25)
diff = birthday - today
print(f"\n크리스마스까지: {diff.days}일")

# 날짜 범위 생성
print("\n다음 7일:")
for i in range(7):
    day = today + timedelta(days=i)
    print(f"  {day:%A, %m/%d}")

# 시간 차이
dt1 = datetime(2025, 1, 1, 9, 0)
dt2 = datetime(2025, 1, 1, 17, 30)
hours = (dt2 - dt1).total_seconds() / 3600
print(f"\n근무 시간: {hours}시간")
```
