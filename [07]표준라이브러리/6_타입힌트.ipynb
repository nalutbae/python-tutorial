# 📘 타입 힌트 (Type Hints)

**타입 힌트**는 함수와 변수의 예상 타입을 명시하는 기능입니다.
실행에는 영향 없이 코드 가독성과 도구 지원을 향상시킵니다.

**학습 목표:**
- 기본 타입 힌트: int, str, List, Dict
- 함수 시그니처 타입 힌트
- 선택형 타입: Optional, Union
- 제네릭: TypeVar, Generic

## 1. 기본 타입 힌트

변수와 함수에 타입을 명시합니다. `typing` 모듈과 내장 타입을 사용합니다.

```python
from typing import List, Dict, Tuple, Set, Optional, Union

# ┌─────────────────────────────────────────┐
# │  기본 타입 힌트                            │
# │  변수: name: str = "kim"                   │
# │  함수: def func(x: int) -> str:             │
# │  컬렉션: List[int], Dict[str, float]         │
# │  선택형: Optional[int] = int | None          │
# └─────────────────────────────────────────┘

# 변수 타입 힌트
name: str = "김파이"
age: int = 25
scores: List[float] = [85.5, 92.3, 78.8]
user_info: Dict[str, Union[str, int]] = {"name": "김파이", "age": 25}

print(f"이름: {name}")
print(f"나이: {age}")
print(f"점수: {scores}")
print(f"정보: {user_info}")

# 함수 타입 힌트
def greet(name: str, age: int) -> str:
    return f"{name}님, {age}세 안녕하세요!"

result = greet("김파이", 25)
print(f"\n{result}")

# 선택형 타입
def find_user(user_id: int) -> Optional[str]:
    users = {1: "김파이", 2: "이코딩"}
    return users.get(user_id)

print(f"사용자 1: {find_user(1)}")
print(f"사용자 3: {find_user(3)}")  # None
```

## 2. 컬렉션과 콜러블 타입

리스트, 딕셔너리, 튜플 등의 타입을 명시합니다.

```python
from typing import List, Dict, Tuple, Callable, Iterable

# ┌─────────────────────────────────────────┐
# │  컬렉션 타입 힌트                          │
# │  List[int]          → 정수 리스트             │
# │  Dict[str, int]     → 문자열→정수 딕셔너리   │
# │  Tuple[str, int]    → (문자열, 정수) 튜플   │
# │  Callable[[int], str] → 함수 타입          │
# └─────────────────────────────────────────┘

# 리스트와 딕셔너리
def count_words(words: List[str]) -> Dict[str, int]:
    result: Dict[str, int] = {}
    for word in words:
        result[word] = result.get(word, 0) + 1
    return result

counts = count_words(["apple", "banana", "apple", "cherry"])
print(f"단어 수: {counts}")

# 튜플
def get_user() -> Tuple[str, int, bool]:
    return ("김파이", 25, True)

name, age, active = get_user()
print(f"\n이름: {name}, 나이: {age}, 활성: {active}")

# 콜러블 (함수) 타입
def apply(func: Callable[[int, int], int], a: int, b: int) -> int:
    return func(a, b)

result = apply(lambda x, y: x + y, 3, 5)
print(f"\napply 결과: {result}")

# 이터러블
def sum_positive(numbers: Iterable[int]) -> int:
    return sum(n for n in numbers if n > 0)

print(f"양수 합: {sum_positive([-1, 2, -3, 4, 5])}")
```

## 3. TypeVar과 Generic

`TypeVar`로 제네릭 타입을, `Generic`로 제네릭 클래스를 만듭니다.

```python
from typing import TypeVar, Generic, List, Optional

# ┌─────────────────────────────────────────┐
# │  TypeVar과 Generic                         │
# │  TypeVar → 제네릭 타입 변수                 │
# │  Generic → 제네릭 클래스                    │
# │  타입 안전성을 유지하면서 다양한 타입 지원  │
# └─────────────────────────────────────────┘

# TypeVar: 제네릭 함수
T = TypeVar('T')

def first(items: List[T]) -> Optional[T]:
    """리스트의 첫 번째 요소 반환"""
    return items[0] if items else None

print(f"정수 리스트 첫 요소: {first([1, 2, 3])}")
print(f"문자열 리스트 첫 요소: {first(['a', 'b', 'c'])}")
print(f"빈 리스트: {first([])}")

# Generic 클래스
class Stack(Generic[T]):
    def __init__(self) -> None:
        self.items: List[T] = []

    def push(self, item: T) -> None:
        self.items.append(item)

    def pop(self) -> Optional[T]:
        return self.items.pop() if self.items else None

    def peek(self) -> Optional[T]:
        return self.items[-1] if self.items else None

    def __len__(self) -> int:
        return len(self.items)

# 정수 스택
int_stack: Stack[int] = Stack()
int_stack.push(1)
int_stack.push(2)
print(f"\n정수 스택 pop: {int_stack.pop()}")

# 문자열 스택
str_stack: Stack[str] = Stack()
str_stack.push("hello")
str_stack.push("world")
print(f"문자열 스택 pop: {str_stack.pop()}")
```
