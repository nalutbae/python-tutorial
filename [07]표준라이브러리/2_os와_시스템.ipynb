# 📘 os와 시스템

**os** 모듈은 운영체제 기능을, **sys** 모듈은 파이썬 런타임 정보를 제공합니다.

**학습 목표:**
- os: 환경변수, 프로세스, 파일 시스템
- sys: argv, path, 버전 정보
- 명령행 인수 처리 기초

## 1. os 모듈 — 환경과 파일 시스템

`os` 모듈은 환경변수, 디렉토리, 파일 정보 등을 제공합니다.

```python
import os

# ┌─────────────────────────────────────────┐
# │  os 모듈 주요 기능                        │
# │  os.environ     → 환경변수                 │
# │  os.getcwd()    → 현재 작업 디렉토리          │
# │  os.listdir()   → 디렉토리 내용              │
# │  os.path.join() → 경로 결합                  │
# │  os.stat()      → 파일 정보                  │
# └─────────────────────────────────────────┘

# 환경변수
home = os.getenv('HOME') or os.getenv('USERPROFILE')
print(f"홈 디렉토리: {home}")
print(f"경로 구분자: {os.sep}")
print(f"PATH (처음 3개):")
for p in os.getenv('PATH', '').split(os.pathsep)[:3]:
    print(f"  {p}")

# 디렉토리
print(f"\n현재 디렉토리: {os.getcwd()}")
print(f"파일 수: {len(os.listdir('.'))}")

# 파일 정보
stat = os.stat('.')
print(f"\n디렉토리 정보:")
print(f"  크기: {stat.st_size}")
print(f"  수정 시간: {stat.st_mtime}")
```

## 2. sys 모듈 — 파이썬 런타임

`sys` 모듈은 인터프리터 정보와 제어 기능을 제공합니다.

```python
import sys

# ┌─────────────────────────────────────────┐
# │  sys 모듈 주요 기능                       │
# │  sys.argv      → 명령행 인수               │
# │  sys.path      → 모듈 검색 경로             │
# │  sys.version    → 파이썬 버전               │
# │  sys.platform   → 운영체제                 │
# │  sys.exit()    → 프로그램 종료              │
# └─────────────────────────────────────────┘

print(f"파이썬 버전: {sys.version}")
print(f"버전 정보: {sys.version_info.major}.{sys.version_info.minor}")
print(f"플랫폼: {sys.platform}")
print(f"기본 인코딩: {sys.getdefaultencoding()}")

# 명령행 인수
print(f"\nargv: {sys.argv}")
print(f"스크립트: {sys.argv[0]}")

# 모듈 검색 경로
print(f"\n모듈 검색 경로 (처음 3개):")
for p in sys.path[:3]:
    print(f"  {p}")

# stdout 인코딩
print(f"\nstdout 인코딩: {sys.stdout.encoding}")
```

## 3. 명령행 인수 처리 기초

`sys.argv`로 명령행 인수를, `argparse`로 고급 인수 처리를 할 수 있습니다.

```python
import argparse

# ┌─────────────────────────────────────────┐
# │  argparse 기본 사용법                     │
# │  ArgumentParser() → 파서 생성               │
# │  .add_argument() → 인수 추가               │
# │  .parse_args()   → 인수 파싱               │
# └─────────────────────────────────────────┘

# 기본 파서
parser = argparse.ArgumentParser(description='튜토리얼 도구')
parser.add_argument('name', help='사용자 이름')
parser.add_argument('-a', '--age', type=int, default=0, help='나이')
parser.add_argument('-v', '--verbose', action='store_true', help='상세 출력')

# Jupyter에서는 인수를 직접 전달
args = parser.parse_args(['김파이', '-a', '25', '-v'])
print(f"이름: {args.name}")
print(f"나이: {args.age}")
print(f"상세: {args.verbose}")

# 선택 인수와 기본값
parser2 = argparse.ArgumentParser()
parser2.add_argument('--input', '-i', default='data.csv', help='입력 파일')
parser2.add_argument('--output', '-o', default='result.json', help='출력 파일')
parser2.add_argument('--count', '-n', type=int, default=10, help='처리 건수')

args2 = parser2.parse_args([])
print(f"\n기본값:")
print(f"  입력: {args2.input}")
print(f"  출력: {args2.output}")
print(f"  건수: {args2.count}")
```
