```python
original = [[1, 2], [3, 4]]
```

````python
# 문제 6: 답을 작성하세요
import copy
original = [[1, 2], [3, 4]]
# 얕은 복사와 깊은 복사 차이 확인
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import copy
original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)
original[0][0] = 99
print(f'원본: {original}')   # [[99, 2], [3, 4]]
print(f'얕은복사: {shallow}')  # [[99, 2], [3, 4]] — 영향 받음
print(f'깊은복사: {deep}')    # [[1, 2], [3, 4]] — 영향 없음
```

## 문제 8: contextlib — with문 활용

`contextlib.suppress`를 사용해 딕셔너리에서 존재하지 않는 키를 삭제할 때
발생하는 `KeyError`를 무시하세요.

````python
# 문제 8: 답을 작성하세요
from contextlib import suppress
d = {'a': 1, 'b': 2}
# suppress로 KeyError 무시하며 삭제
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from contextlib import suppress
d = {'a': 1, 'b': 2}
with suppress(KeyError):
    del d['c']  # 키가 없어도 에러 없이 통과
with suppress(KeyError):
    del d['a']  # 존재하는 키는 정상 삭제
print(d)  # {'b': 2}
```

</details>

(메모리에서 `io.BytesIO`를 사용해 실제 파일을 만들지 마세요.)

````python
# 문제 10: 답을 작성하세요
import zipfile, io
# 메모리 ZIP 생성 + 파일 추가 + 목록 출력
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import zipfile, io
buf = io.BytesIO()
with zipfile.ZipFile(buf, 'w') as zf:
    zf.writestr('hello.txt', 'Hello, World!')
    zf.writestr('data.txt', 'Python ZipFile')
with zipfile.ZipFile(buf, 'r') as zf:
    print('파일 목록:', zf.namelist())
    print('hello.txt:', zf.read('hello.txt').decode())
```
