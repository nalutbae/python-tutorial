
````python
# 문제 10: 답을 작성하세요
import random
target = random.Random(42).randint(1, 100)
guesses = [50, 75, 62, 68, 71, 69, 70]
# while 루프로 숫자 맞추기
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import random
target = random.Random(42).randint(1, 100)
guesses = [50, 75, 62, 68, 71, 69, 70]
idx = 0
attempts = 0
while idx < len(guesses):
    guess = guesses[idx]
    attempts += 1
    if guess == target:
        print(f'정답! {target} ({attempts}번 만에 맞췄습니다)')
        break
    elif guess > target:
        print(f'{guess}: 작다')
    else:
        print(f'{guess}: 크다')
    idx += 1
```

</details>
