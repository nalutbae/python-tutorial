# 📘 for 반복문

`for`문은 시퀀스(리스트, 문자열, range 등)의 요소를 순회하며 반복 실행합니다.

**학습 목표:**
- for문 기본 구조와 `range()`
- `enumerate()`, `zip()`으로 다중 반복
- 딕셔너리 순회
- 중첩 for문

## 1. for문 기본과 range()

`for 변수 in 시퀀스:` 형태로 작성합니다.
`range()`는 정수 시퀀스를 생성합니다.

```python
# 리스트 순회
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
# apple
# banana
# cherry
# ┌───────────────────────────────────────────┐
# │  range() — 정수 시퀀스 생성                   │
# │  range(stop):           0 ~ stop-1          │
# │  range(start, stop):    start ~ stop-1       │
# │  range(start, stop, step): start ~ stop-1     │
# │  step간격으로                                   │
# └───────────────────────────────────────────┘

# range(stop): 0부터 stop-1까지
for i in range(5):
    print(i, end=" ")
# 0 1 2 3 4
print()
```

```python
# range(start, stop, step): 시작/끝/간격 지정
for i in range(2, 10, 3):
    print(i, end=" ")
# 2 5 8
print()
```

```python
# 역방향 range
for i in range(10, 0, -2):
    print(i, end=" ")
# 10 8 6 4 2
print()
```

## 2. enumerate()와 zip()

`enumerate()`는 인덱스와 값을 함께, `zip()`은 여러 시퀀스를 병렬로 순회합니다.

```python
# enumerate(): 인덱스와 값을 동시에 얻기
names = ["Alice", "Bob", "Charlie"]

for index, name in enumerate(names):
    print(f"{index}: {name}")
# 0: Alice
# 1: Bob
# 2: Charlie

# 시작 번호 지정
for index, name in enumerate(names, start=1):
    print(f"{index}번: {name}")
```

```python
# 1번: Alice
# 2번: Bob
# 3번: Charlie

# ┌───────────────────────────────────────────┐
# │  zip(): 여러 시퀀스를 병렬로 순회              │
# │  가장 짧은 시퀀스 길이에 맞춰서 순회           │
# │  길이가 다르면 나머지는 무시됨                  │
# └───────────────────────────────────────────┘

letters = ["a", "b", "c"]
numbers = [1, 2, 3]

for letter, number in zip(letters, numbers):
    print(f"{letter}: {number}")
```

```python
# a: 1
# b: 2
# c: 3

# 길이가 다른 경우
short = [10, 20]
long = [1, 2, 3, 4, 5]

for s, l in zip(short, long):
    print(s, l)
```

```python
# 10 1
# 20 2

# zip으로 딕셔너리 만들기
keys = ["name", "age", "city"]
values = ["Alice", 30, "Seoul"]
person = dict(zip(keys, values))
print(f"zip으로 딕셔너리: {person}")
```

## 3. 딕셔너리 순회와 중첩 for문

딕셔너리는 `keys()`, `values()`, `items()`로 순회합니다.
중첩 for문으로 2차원 데이터를 처리할 수 있습니다.

```python
# 딕셔너리 순회
person = {"name": "Alice", "age": 30, "city": "Seoul"}
```

```python
# 키만 순회
print("=== 키 순회 ===")
for key in person:
    print(f"  {key}: {person[key]}")
```

```python
# items()로 키-값 쌍 순회 (권장)
print("\n=== items() 순회 ===")
for key, value in person.items():
    print(f"  {key} = {value}")
```

```python
# 중첩 for문 — 구구단
print("\n=== 구구단 (2~4단) ===")
for dan in range(2, 5):           # 2단 ~ 4단
    for i in range(1, 10):        # 1 ~ 9
        result = dan * i
        print(f"{dan} x {i} = {result}")
    print("---")
```

```python
# ┌─────────────────────────────────────────┐
# │  중첩 for문 주의사항                        │
# │  너무 많은 중첩은 가독성을 떨어뜨립니다        │
# │  3단계 이상 중첩 시 함수로 분리 권장          │
# └─────────────────────────────────────────┘

# 2차원 리스트 순회
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("\n2차원 리스트:")
for row in matrix:
    for val in row:
        print(f"{val:3d}", end="")
    print()  # 줄바꿈
```

## 4. for-else와 반복 패턴

`for-else`의 `else`는 **break 없이 반복이 정상 종료**되었을 때 실행됩니다.
이것은 다른 언어에는 없는 파이썬 특유의 기능입니다.

```python
# for-else: break 없이 정상 종료되면 else 실행
# ┌──────────────────────────────────────────────┐
# │  for-else 구조                                │
# │  for 변수 in 시퀀스:                           │
# │      if 조건: break  → break되면 else 생략     │
# │  else:                                        │
# │      break 없이 반복 완료 시 실행               │
# └──────────────────────────────────────────────┘

# break 없이 완료 → else 실행
for num in range(3):
    print(num, end=" ")
else:
    print("| 정상 완료")    # ← 실행됨
```

```python
# break로 중단 → else 생략
for num in range(10):
    if num == 3:
        print(f"| {num}에서 break!")
        break
else:
    print("이 줄은 실행되지 않음")
```

```python
# 실용 예시: 리스트에서 조건 검색
def find_even(numbers):
    for n in numbers:
        if n % 2 == 0:
            print(f"짝수 발견: {n}")
            break
    else:
        print("짝수가 없습니다")

find_even([1, 3, 4, 6])    # 짝수 발견: 4
find_even([1, 3, 5])        # 짝수가 없습니다
```

```python
# 실용 예시: 소수 판별
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    else:
        return True

for num in [2, 7, 10, 13, 15]:
    print(f"{num}: {'소수' if is_prime(num) else '합성수'}")
```
