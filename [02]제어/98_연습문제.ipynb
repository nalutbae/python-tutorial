## 정답 확인

아래 셀을 실행하여 모든 문제의 정답을 확인하세요.

````python
# 정답

# ---- 문제 1 정답 ----
def get_grade(score):
    """점수(0~100)를 받아 학점을 반환한다."""
    if score < 0 or score > 100:
        return '잘못된 점수'
    elif score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

# ---- 문제 2 정답 ----
def make_ranking(names, scores):
    """이름 리스트와 점수 리스트를 받아 등수 문자열 리스트를 반환한다."""
    result = []
    for i, (name, score) in enumerate(zip(names, scores), start=1):
        result.append(f"{i}위: {name} - {score}점")
    return result

# ---- 문제 3 정답 ----
def collatz_steps(n):
    """양의 정수 n이 1이 될 때까지 콜라츠 규칙을 적용하고 단계 수를 반환한다."""
    steps = 0
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = n * 3 + 1
        steps += 1
    return steps

# ---- 문제 4 정답 ----
def find_first_multiple_of_7(nums):
    """정수 리스트에서 음수를 건너뛰고 첫 번째 7의 배수의 인덱스를 반환한다."""
    for i, num in enumerate(nums):
        if num < 0:
            continue
        if num % 7 == 0:
            return i
    return -1

# ---- 문제 5 정답 ----
def filter_words(words):
    """길이 3 이상이고 모음을 포함하는 단어를 대문자 리스트로 반환한다."""
    vowels = set('aeiouAEIOU')
    return [w.upper() for w in words if len(w) >= 3 and any(c in vowels for c in w)]

# ---- 문제 6 정답 ----
def row_maxes(matrix):
    """2차원 리스트에서 각 행의 최댓값을 리스트로 반환한다. 빈 행은 무시."""
    result = []
    for row in matrix:
        if not row:
            continue
        max_val = row[0]
        for val in row:
            if val > max_val:
                max_val = val
        result.append(max_val)
    return result

# ---- 문제 7 정답 ----
def http_status(code):
    """HTTP 상태 코드를 한국어 설명으로 변환한다."""
    match code:
        case 200:
            return '성공'
        case 301:
            return '리다이렉트'
        case 403:
            return '접근 거부'
        case 404:
            return '찾을 수 없음'
        case 500:
            return '서버 오류'
        case _:
            return '알 수 없는 상태 코드'

# ---- 문제 8 정답 ----
def safe_divide_by_100(s):
    """문자열을 정수로 변환해 100으로 나눈 결과를 반환한다. 예외 시 안내 문자열 반환."""
    try:
        return int(s) // 100
    except ValueError:
        return '변환 오류'
    except ZeroDivisionError:
        return '0으로 나눔'
    except TypeError:
        return '타입 오류'

# ---- 문제 9 정답 ----
def contains(nums, target):
    """리스트에 target이 존재하면 True, 없으면 False를 반환. for-else 사용."""
    for num in nums:
        if num == target:
            return True
    else:
        return False

# ---- 문제 10 정답 ----
def word_length_dict(words):
    """길이 3 이상인 단어만 {단어: 길이} 딕셔너리로 반환한다."""
    return {w: len(w) for w in words if len(w) >= 3}


# ---- 전체 테스트 실행 ----
print('=== 문제 1 ===')
assert get_grade(95) == 'A'
assert get_grade(85) == 'B'
assert get_grade(75) == 'C'
assert get_grade(65) == 'D'
assert get_grade(55) == 'F'
assert get_grade(-5) == '잘못된 점수'
assert get_grade(105) == '잘못된 점수'
print('통과!')

print('=== 문제 2 ===')
assert make_ranking(['가나', '다라', '마바', '사아'], [90, 85, 70, 95]) == [
    '1위: 가나 - 90점', '2위: 다라 - 85점', '3위: 마바 - 70점', '4위: 사아 - 95점']
print('통과!')

print('=== 문제 3 ===')
assert collatz_steps(6) == 8
assert collatz_steps(1) == 0
assert collatz_steps(27) == 111
print('통과!')

print('=== 문제 4 ===')
assert find_first_multiple_of_7([3, -2, 14, 21, 7]) == 2
assert find_first_multiple_of_7([-1, -5, 7]) == 2
assert find_first_multiple_of_7([1, 2, 3, 4, 5]) == -1
assert find_first_multiple_of_7([]) == -1
print('통과!')

print('=== 문제 5 ===')
assert filter_words(['sky', 'apple', 'fly', 'orange', 'cry']) == ['APPLE', 'ORANGE']
assert filter_words(['b', 'echo', 'xyz', 'idea']) == ['ECHO', 'IDEA']
assert filter_words(['zzz', 'aaa']) == ['AAA']
print('통과!')

print('=== 문제 6 ===')
assert row_maxes([[3, 7, 2], [8, 1], [5, 9, 4]]) == [7, 8, 9]
assert row_maxes([[1], [2, 3], []]) == [1, 3]
assert row_maxes([]) == []
print('통과!')

print('=== 문제 7 ===')
assert http_status(200) == '성공'
assert http_status(404) == '찾을 수 없음'
assert http_status(999) == '알 수 없는 상태 코드'
print('통과!')

print('=== 문제 8 ===')
assert safe_divide_by_100('500') == 5
assert safe_divide_by_100('0') == 0
assert safe_divide_by_100('abc') == '변환 오류'
assert safe_divide_by_100(None) == '타입 오류'
print('통과!')

print('=== 문제 9 ===')
assert contains([1, 3, 5, 7], 5) == True
assert contains([1, 3, 5, 7], 4) == False
assert contains([], 1) == False
assert contains([2, 4, 6], 2) == True
print('통과!')

print('=== 문제 10 ===')
assert word_length_dict(['hi', 'hello', 'a', 'world', 'py']) == {'hello': 5, 'world': 5}
assert word_length_dict(['cat', 'dog', 'in', 'the', 'hat']) == {'cat': 3, 'dog': 3, 'the': 3, 'hat': 3}
assert word_length_dict(['a', 'b', 'c']) == {}
print('통과!')

print('\n✅ 모든 테스트 통과!')
````
