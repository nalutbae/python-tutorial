# 📘 예외 처리 (try-except)

프로그램 실행 중 발생하는 **오류(예외)**를 안전하게 처리하는 방법을 배웁니다.

**학습 목표:**
- try-except-else-finally 구조
- 특정 예외 잡기와 예외 계층
- 예외 발생시키기 (raise)와 사용자 정의 예외
- 예외 처리 모범 사례

## 1. try-except 기본

예외가 발생하면 프로그램이 종료됩니다.
`try-except`로 예외를 잡아 프로그램이 계속 실행되도록 합니다.

```python
# ┌──────────────────────────────────────────────────┐
# │  try-except-else-finally 구조                       │
# │  try:       오류가 날 수 있는 코드                    │
# │  except:    오류 발생 시 실행 (특정 예외 지정 가능)   │
# │  else:      오류 없이 정상 종료 시 실행               │
# │  finally:   오류 여부와 상관없이 항상 실행            │
# └──────────────────────────────────────────────────┘

# 0으로 나누기 — ZeroDivisionError
```

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다")    # ← 실행됨
```

```python
# 정상 실행 — else 블록
try:
    result = 10 / 2
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다")
else:
    print(f"결과: {result}")    # ← 실행됨 (예외 없을 때)
```

```python
# ┌─────────────────────────────────────────┐
# │  특정 예외부터 잡기!                        │
# │  구체적인 예외를 먼저,                       │
# │  범용 예외(Exception)를 나중에 작성           │
# │  except 절의 순서가 중요합니다               │
# └─────────────────────────────────────────┘

# 여러 예외 처리
try:
    value = int("abc")          # ValueError 발생
except ValueError as e:
    print(f"값 오류: {e}")       # ← 실행됨
except TypeError as e:
    print(f"타입 오류: {e}")
except Exception as e:
    print(f"기타 오류: {e}")
```

```python
# 여러 예외를 하나의 except로
try:
    lst = [1, 2, 3]
    print(lst[10])               # IndexError 발생
except (ValueError, TypeError) as e:
    print(f"변환 오류: {e}")
except Exception as e:
    print(f"기타 오류: {e}")       # ← 실행됨 (IndexError)
```

## 2. finally와 raise

`finally`는 **항상 실행**되므로 정리 작업에 적합합니다.
`raise`는 의도적으로 예외를 발생시킵니다.

```python
# finally: 예외 발생 여부와 상관없이 항상 실행
print("=== 정상 실행 시 ===")
try:
    print("try 블록 실행")
    result = 10 / 2
except ZeroDivisionError:
```

```python
    print("0으로 나누기 오류")
else:
    print(f"else 블록: 결과 = {result}")
finally:
    print("finally: 항상 실행됨\n")    # ← 항상!

print("=== 예외 발생 시 ===")
try:
    print("try 블록 실행")
    result = 10 / 0                 # 예외 발생!
except ZeroDivisionError:
    print("0으로 나누기 오류")
finally:
    print("finally: 예외가 발생해도 실행됨\n")
```

```python
# ┌─────────────────────────────────────────┐
# │  raise: 의도적으로 예외 발생시키기         │
# │  raise 예외타입("메시지")                  │
# │  raise: 현재 예외를 다시 발생시킴           │
# └─────────────────────────────────────────┘
```

```python
def check_age(age):
    if age < 0:
        raise ValueError("나이는 음수일 수 없습니다")
    if age < 18:
        raise ValueError("18세 이상만 접근 가능합니다")
    return "접근 허가됨"

try:
    print(check_age(25))     # 접근 허가됨
except ValueError as e:
    print(f"오류: {e}")

try:
    print(check_age(-5))    # ValueError!
except ValueError as e:
    print(f"오류: {e}")       # 나이는 음수일 수 없습니다

try:
    print(check_age(15))     # ValueError!
except ValueError as e:
    print(f"오류: {e}")       # 18세 이상만 접근 가능합니다
```

## 3. 사용자 정의 예외와 예외 체인

자신만의 예외 클래스를 만들면 더 명확한 에러 처리가 가능합니다.
`raise ... from`으로 예외의 원인을 연결할 수 있습니다.

```python
# ┌─────────────────────────────────────────────────┐
# │  사용자 정의 예외 만들기                             │
# │  Exception을 상속받아 커스텀 예외 클래스 생성         │
# │  __init__으로 에러 메시지를 지정                      │
# │  __str__로 에러 표현을 커스터마이징                   │
# └─────────────────────────────────────────────────┘

class InvalidScoreError(Exception):
    """0~100 범위를 벗어난 점수에 대한 예외"""
```

```python
    def __init__(self, score, message="점수는 0~100 사이여야 합니다"):
        self.score = score
        self.message = message
        super().__init__(self.message)
    
    def __str__(self):
        return f"{self.message} (입력값: {self.score})"

def validate_score(score):
    if not 0 <= score <= 100:
        raise InvalidScoreError(score)
    return score

try:
    validate_score(150)
except InvalidScoreError as e:
    print(f"커스텀 예외: {e}")     # 점수는 0~100 사이여야 합니다 (입력값: 150)
```

```python
# ┌─────────────────────────────────────────┐
# │  파이썬 내장 예외 계층 (일부)               │
# │  BaseException                             │
# │   ├── Exception                            │
# │   │    ├── ValueError   (잘못된 값)         │
# │   │    ├── TypeError    (잘못된 타입)         │
# │   │    ├── KeyError     (키 없음)             │
# │   │    ├── IndexError   (인덱스 초과)          │
# │   │    ├── FileNotFoundError (파일 없음)       │
# │   │    ├── ZeroDivisionError  (0 나눔)        │
# │   │    ├── AttributeError (속성 없음)           │
# │   │    └── RuntimeError (런타임 오류)         │
# │   ├── KeyboardInterrupt (Ctrl+C)          │
# │   └── SystemExit                            │
# └─────────────────────────────────────────┘

# 예외 체인: raise ... from
try:
    number = int("abc")
except ValueError as e:
    raise InvalidScoreError("abc", "숫자로 변환할 수 없습니다") from e
```

```python
# 실용 예시: 파일 읽기 안전 래퍼
def safe_read_file(filepath):
    try:
        with open(filepath, "r") as f:
            return f.read()
    except FileNotFoundError:
        print(f"파일을 찾을 수 없습니다: {filepath}")
        return None
    except PermissionError:
        print(f"파일 읽기 권한이 없습니다: {filepath}")
        return None
    except Exception as e:
        print(f"예상치 못한 오류: {e}")
        return None

content = safe_read_file("/tmp/nonexistent.txt")
print(f"파일 내용: {content}")    # None
```
