# 📘 while 반복문

`while`문은 **조건이 참인 동안** 계속 반복합니다.
반복 횟수를 미리 알 수 없을 때 유용합니다.

**학습 목표:**
- while문 기본 구조
- while-else 문
- 무한 루프와 break/continue
- while vs for 언제 쓸까?

## 1. while문 기본

`while 조건:` — 조건이 참인 동안 블록을 반복 실행합니다.

```python
# ┌───────────────────────────────────────────┐
# │  while 조건문 구조                           │
# │  while 조건:                                │
# │      실행할 코드  (조건이 참인 동안 반복)       │
# │                                             │
# │  주의: 조건이 항상 참이면 무한 루프!           │
# │       반드시 조건을 False로 만드는 코드 필요    │
# └───────────────────────────────────────────┘

# 기본 while 루프
count = 0
while count < 5:
    print(count, end=" ")
    count += 1     # ← 이게 없으면 무한 루프!
# 0 1 2 3 4
print()
```

```python
# 누적합 계산
total = 0
num = 1
while total < 50:
    total += num
    num += 1
print(f"누적합: {total}, 카운트: {num - 1}")  # 55, 10
```

```python
# 카운트다운
count = 5
while count > 0:
    print(f"{count}...")
    count -= 1
print("발사!")
```

## 2. while-else 문

`while-else`의 `else`는 **break 없이 조건이 False가 되어 종료**되었을 때 실행됩니다.
`for-else`와 동일한 개념입니다.

```python
# while-else: 정상 종료 시 else 실행
n = 0
while n < 3:
    print(n, end=" ")
    n += 1
else:
    print("| 정상 완료")    # ← 실행됨
```

```python
# while-else: break로 중단 시 else 생략
n = 0
while n < 10:
    if n == 3:
        print(f"| {n}에서 break!")
        break
    n += 1
else:
    print("이 줄은 실행 안 됨")
```

```python
# ┌───────────────────────────────────────────┐
# │  while-else 실용 예시                        │
# │  조건을 만족하는 첫 번째 요소를 찾을 때 유용    │
# │  break 없이 끝까지 탐색 → else 실행           │
# │  break로 중간에 발견 → else 생략              │
# └───────────────────────────────────────────┘

# 첫 번째 짝수 찾기
numbers = [1, 3, 5, 4, 7]
i = 0
while i < len(numbers):
    if numbers[i] % 2 == 0:
        print(f"첫 짝수: {numbers[i]} (인덱스 {i})")
        break
    i += 1
else:
    print("짝수가 없습니다")
```

## 3. break와 continue

- `break`: 반복문을 즉시 종료
- `continue`: 현재 반복을 건너뛰고 다음 반복으로

```python
# break: 반복문 즉시 종료
print("=== break 예시 ===")
for i in range(10):
    if i == 5:
        break      # 5에서 반복 종료
    print(i, end=" ")
# 0 1 2 3 4
print()
```

```python
# continue: 현재 반복 건너뛰기
print("=== continue 예시 ===")
for i in range(10):
    if i % 2 == 0:
        continue   # 짝수는 건너뜀
    print(i, end=" ")
# 1 3 5 7 9
print()
```

```python
# while에서 continue
print("=== while + continue ===")
n = 0
while n < 10:
    n += 1
    if n % 3 == 0:
        continue   # 3의 배수 건너뛰기
    print(n, end=" ")
# 1 2 4 5 7 8 10
print()
```

```python
# ┌─────────────────────────────────────────┐
# │  중첩 루프에서 break                        │
# │  break는 가장 안쪽 루프만 빠져나감            │
# │  바깥 루프까지 빠져나가려면 플래그 변수 필요    │
# └─────────────────────────────────────────┘

found = False
for i in range(3):
    for j in range(5):
        if j == 2:
            found = True
            break   # 안쪽 루프만 종료
        print(f"({i},{j})", end=" ")
    if found:
        break       # 바깥 루프도 종료
    print()
```

## 4. while vs for — 언제 무엇을 쓸까?

| 특징 | for | while |
|------|-----|-------|
| 반복 횟수 | 미리 알 때 | 모를 때 |
| 순회 대상 | 시퀀스/iterable | 조건식 |
| 전형적 용도 | 리스트, range 순회 | 사용자 입력, 상태 변화 대기 |
| 무한 루프 | `for _ in iter(int, 1)` | `while True` |

```python
# for가 적합한 경우: 반복 횟수가 정해져 있음
print("=== for: 1부터 10까지 합 ===")
total = 0
for i in range(1, 11):
    total += i
print(f"합: {total}")  # 55
```

```python
# while이 적합한 경우: 종료 조건이 상태에 따라 결정
print("\n=== while: 누적합이 100 초과될 때까지 ===")
total = 0
i = 1
while total <= 100:
    total += i
    i += 1
print(f"합: {total}, 카운트: {i - 1}")  # 105, 14
```

```python
# while이 적합한 경우: 시뮬레이션
import random
random.seed(42)

target = 50
attempts = 0
while True:
    guess = random.randint(1, 100)
    attempts += 1
    if guess == target:
        print(f"\n{attempts}번 만에 {target}을 찾았습니다!")
        break
```

```python
# while이 적합한 경우: 유효한 입력 받기 (예시)
# 실제로는 input()을 사용하지만 여기서는 시뮬레이션
valid_inputs = ["yes", "no", "quit"]
user_input = "maybe"
idx = 0
while user_input not in valid_inputs:
    # 실제: user_input = input("입력: ")
    user_input = valid_inputs[idx]   # 시뮬레이션
    idx += 1
print(f"유효한 입력: {user_input}")
```
