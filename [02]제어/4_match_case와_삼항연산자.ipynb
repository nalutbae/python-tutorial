# 📘 match-case와 삼항 연산자

파이썬 3.10에 추가된 **match-case** 문과 간결한 조건 표현인 **삼항 연산자**를 배웁니다.

**학습 목표:**
- match-case 기본 패턴 매칭
- OR 패턴, 변수 바인딩, 타입 매칭, 리스트/딕셔너리 패턴
- 삼항 조건 표현식
- 단락 평가(short-circuit)와 바다코끼리 연산자(:=)

## 1. match-case 기본 (Python 3.10+)

`match-case`는 다른 언어의 `switch-case`와 비슷하지만 **패턴 매칭** 기능이 더 강력합니다.

⚠️ Python 3.10 이상에서만 사용 가능합니다.

````python
# ┌───────────────────────────────────────────┐
# │  match-case 기본 구조                       │
# │  match 변수:                                │
# │      case 패턴1:                            │
# │          실행 코드                           │
# │      case 패턴2:                            │
# │          실행 코드                           │
# │      case _:           ← 와일드카드 (else)   │
# │          기본 실행 코드                       │
# └───────────────────────────────────────────┘

# 기본 match-case
command = "start"

match command:
    case "start":
        print("시작합니다")       # ← 실행됨
    case "stop":
        print("중지합니다")
    case "pause":
        print("일시정지합니다")
    case _:
        print("알 수 없는 명령")
````

````python
# OR 패턴: 여러 값에 같은 동작
status = "warn"

match status:
    case "error" | "critical":
        print("치명적 오류!")
    case "warn" | "warning":
        print("경고 발생")       # ← 실행됨
    case "ok" | "success":
        print("정상")
    case _:
        print("알 수 없는 상태")
````

## 2. 변수 바인딩과 타입 매칭

match-case는 매칭된 값을 **변수에 바인딩**하거나 **값의 타입**으로 분기할 수 있습니다.

````python
# 변수 바인딩: 패턴 매칭 결과를 변수에 저장
point = (3, 5)

match point:
    case (0, 0):
        print("원점")
    case (x, 0):
        print(f"x축 위: x={x}")
    case (0, y):
        print(f"y축 위: y={y}")
    case (x, y):
        print(f"좌표: ({x}, {y})")   # ← 실행됨
````

````python
# 타입 매칭: 값의 타입으로 분기
value = [1, 2, 3]

match value:
    case int(n):
        print(f"정수: {n}")
    case float(n):
        print(f"실수: {n}")
    case str(s):
        print(f"문자열: {s}")
    case list(l):
        print(f"리스트: {l}")       # ← 실행됨
    case dict(d):
        print(f"딕셔너리: {d}")
    case _:
        print(f"알 수 없는 타입: {type(value)}")
````

````python
# ┌─────────────────────────────────────────┐
# │  리스트 패턴 매칭                            │
# │  []: 빈 리스트                               │
# │  [x]: 요소 1개                               │
# │  [x, y]: 요소 2개                            │
# │  [x, y, *rest]: 앞 2개 + 나머지              │
# └─────────────────────────────────────────┘

data = [1, 2, 3, 4, 5]

match data:
    case []:
        print("빈 리스트")
    case [x]:
        print(f"요소 1개: {x}")
    case [x, y]:
        print(f"요소 2개: {x}, {y}")
    case [x, y, *rest]:
        print(f"첫 번째: {x}, 두 번째: {y}, 나머지: {rest}")  # ← 실행됨
````

````python
# 딕셔너리 패턴 매칭
person = {"name": "Alice", "age": 30}

match person:
    case {"name": name, "age": age}:
        print(f"이름: {name}, 나이: {age}")   # ← 실행됨
    case {"name": name}:
        print(f"이름: {name}")
````

````python
# 가드(guard) 조건 추가
score = 85

match score:
    case s if s >= 90:
        print(f"A학점 ({s}점)")
    case s if s >= 80:
        print(f"B학점 ({s}점)")     # ← 실행됨
    case s if s >= 70:
        print(f"C학점 ({s}점)")
    case _:
        print(f"F학점 ({score}점)")
````

## 3. 삼항 연산자 (조건 표현식)

삼항 연산자는 `if-else`를 한 줄로 표현합니다.

```python
값1 if 조건 else 값2
```

````python
# ┌───────────────────────────────────────────────┐
# │  삼항 연산자 (Ternary Conditional Expression)    │
# │  결과 = 값1 if 조건 else 값2                     │
# │  조건이 참이면 값1, 거짓이면 값2                  │
# └───────────────────────────────────────────────┘

# 기본 삼항 연산자
age = 20
status = "성인" if age >= 18 else "미성년"
print(f"상태: {status}")           # 성인
````

````python
# 숫자 비교
x, y = 10, 20
max_val = x if x > y else y
print(f"최댓값: {max_val}")        # 20
````

````python
# 합격/불합격
score = 75
result = "합격" if score >= 60 else "불합격"
print(f"결과: {result}")           # 합격
````

````python
# 리스트 컴프리헨션에서 삼항 연산자
numbers = range(10)
labels = ["짝수" if n % 2 == 0 else "홀수" for n in numbers]
print(f"라벨: {labels}")
````

````python
# None 확인과 기본값
value = None
result = value if value is not None else "기본값"
print(f"결과: {result}")           # 기본값
````

````python
# ┌─────────────────────────────────────────┐
# │  중첩 삼항 연산자 (가독성 주의!)           │
# │  너무 복잡하면 if-elif-else를 사용하세요   │
# └─────────────────────────────────────────┘

grade = 85
letter = "A" if grade >= 90 else "B" if grade >= 80 else "C" if grade >= 70 else "F"
print(f"학점: {letter}")           # B
````

## 4. 단락 평가와 바다코끼리 연산자

`and`/`or`은 마지막으로 평가된 값을 반환하며, 필요한 곳까지만 평가합니다(단락 평가).
`:=`(바다코끼리 연산자)는 할당과 평가를 동시에 합니다 (Python 3.8+).

````python
# ┌─────────────────────────────────────────┐
# │  단락 평가 (Short-circuit Evaluation)      │
# │  and: 첫 번째가 거짓이면 바로 반환            │
# │       둘 다 참이면 마지막 값을 반환            │
# │  or: 첫 번째가 참이면 바로 반환               │
# │      둘 다 거짓이면 마지막 값을 반환            │
# └─────────────────────────────────────────┘

# and: 둘 다 참이면 마지막 값, 아니면 첫 번째 거짓 값
print(1 and 2 and 3)    # 3 (마지막 참 값)
print(0 and 2 and 3)    # 0 (첫 번째 거짓 값에서 멈춤)
print("" and "hello")   # "" (빈 문자열은 Falsy)
print([] and "hello")   # [] (빈 리스트는 Falsy)
````

````python
# or: 첫 번째 참 값, 둘 다 거짓이면 마지막 값
print(0 or 1 or 2)      # 1 (첫 번째 참 값)
print("" or "기본값")    # 기본값
print(None or "대체")    # 대체
````

````python
# 실용 예시: 기본값 설정
name = ""
display = name or "익명"
print(f"표시: {display}")      # 익명

name = "김파이"
display = name or "익명"
print(f"표시: {display}")      # 김파이
````

````python
# 조건부 실행
count = 5
count > 0 and print("양수입니다")    # 실행됨
count == 0 and print("0입니다")        # 실행 안 됨
# ┌─────────────────────────────────────────┐
# │  바다코끼리 연산자 := (Python 3.8+)        │
# │  할당과 평가를 동시에 수행                   │
# │  (표현식 := 값) → 할당하면서 값 사용         │
# └─────────────────────────────────────────┘

data = [1, 2, 3, 4, 5]
# 할당하면서 조건 검사
if (length := len(data)) > 3:
    print(f"길이: {length}, 데이터가 깁니다")    # 길이: 5
````

````python
# while 루프에서 유용
inputs = ["hello", "world", ""]
idx = 0
results = []
while idx < len(inputs) and (word := inputs[idx]):
    results.append(word.upper())
    idx += 1
print(f"결과: {results}")    # ['HELLO', 'WORLD']
# 삼항 연산자와 함께
result = "김" if (length := len(data)) > 3 else "짧"
print(f"길이 {length}: {result}")     # 길이 5: 김
````
