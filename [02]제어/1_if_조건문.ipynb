# 📘 if 조건문

프로그램은 **조건에 따라 다른 동작**을 해야 할 때가 많습니다.
파이썬의 `if`, `elif`, `else`로 조건문을 작성합니다.

**학습 목표:**

- 중첩 조건문과 논리 연산자

## 1. if / elif / else 기본 구조

파이썬은 들여쓰기(indentation)로 블록을 구분합니다.
들여쓰기는 항상 **4칸**을 사용하는 것이 관례입니다.

```python
# ┌───────────────────────────────────────────┐
# │  if 조건문 기본 구조                        │
# │  if 조건:                                  │
# │      실행할 코드                             │
# │  elif 조건:    (선택)                        │
# │      실행할 코드                             │
# │  else:         (선택)                        │
# │      실행할 코드                             │
# └───────────────────────────────────────────┘

# 단순 if
x = 10
if x > 0:
    print("x는 양수입니다")
```

```python
# if-else
y = -5
if y > 0:
    print("y는 양수입니다")
else:
    print("y는 음수입니다")
```

```python
# if-elif-else (여러 조건)
z = 0
if z > 0:
    print("z는 양수")
elif z < 0:
    print("z는 음수")
else:
    print("z는 0")
```

```python
# 학점 판별 예시
grade = 85
if grade >= 90:
    print("A학점")
elif grade >= 80:
    print("B학점")   # ← 여기서 참!
elif grade >= 70:
    print("C학점")
elif grade >= 60:
    print("D학점")
else:
    print("F학점")
```

## 2. 중첩 조건문과 논리 연산자

조건문 안에 또 조건문을 넣을 수 있습니다.
하지만 `and`, `or`을 사용하면 더 간결하게 표현할 수 있습니다.

```python
# 중첩 if문
num = 15

if num > 0:
    if num % 2 == 0:
        print(f"{num}은(는) 양의 짝수")
    else:
        print(f"{num}은(는) 양의 홀수")   # ← 여기!
elif num < 0:
    print(f"{num}은(는) 음수")
else:
    print("0입니다")
```

```python
# ┌───────────────────────────────────────────────┐
# │  논리 연산자로 조건 간결하게 표현하기             │
# │  and: 둘 다 참이어야 참                         │
# │  or: 하나라도 참이면 참                          │
# │  not: 참→거짓, 거짓→참                          │
# │  복합 조건은 괄호로 명확히!                       │
# └───────────────────────────────────────────────┘

age = 25
has_id = True

if age >= 18 and has_id:
    print("입장 허용")      # ← 실행됨

if age < 13 or age > 65:
    print("할인 적용")
else:
    print("일반 요금")      # ← 실행됨
```

```python
# not 키워드
is_active = False
if not is_active:
    print("비활성 계정")     # ← 실행됨
```

## 3. in / not in 멤버십 검사

`in` 연산자로 리스트, 문자열, 딕셔너리에 값이 있는지 확인할 수 있습니다.

```python
# 리스트에서 멤버십 검사
fruits = ["apple", "banana", "cherry"]

if "banana" in fruits:
    print("바나나가 있습니다")    # ← 실행됨

if "grape" not in fruits:
    print("포도는 없습니다")       # ← 실행됨
```

```python
# 문자열에서 멤버십 검사
text = "Hello, Python!"
if "Python" in text:
    print("Python이 포함되어 있습니다")
```

```python
# 딕셔너리에서 키 검사
person = {"name": "Alice", "age": 30}

if "name" in person:
    print("name 키가 존재합니다")  # ← 실행됨

if "email" not in person:
    print("email 키가 없습니다")   # ← 실행됨
# ┌────────────────────────────────────────────┐
# │  멤버십 연산자                                │
# │  x in 시퀀스   → 시퀀스에 x가 있으면 True    │
# │  x not in 시퀀스 → 없으면 True               │
# │  적용 대상: 리스트, 튜플, 문자열, 집합, 딕셔너리│
# │  딕셔너리는 키(key)를 기준으로 검사            │
# └────────────────────────────────────────────┘

# range에서도 사용 가능
if 5 in range(10):
    print("5는 0~9 범위 안에 있습니다")
```

## 4. Truthy와 Falsy

파이썬은 `if` 조건에서 **0이 아닌 수, 비어있지 않은 컨테이너**를 참(True)으로 평가합니다.

```python
# ┌─────────────────────────────────────────┐
# │  Falsy 값 (거짓으로 평가되는 값들)          │
# │  False, None, 0, 0.0, "", [], {}, (), set()│
# │  나머지는 모두 Truthy (참으로 평가)          │
# └─────────────────────────────────────────┘

# Falsy 값
if 0:
    print("실행 안 됨")
else:
    print("0은 Falsy")        # ← 실행됨

if "":
    print("실행 안 됨")
else:
    print("빈 문자열은 Falsy")   # ← 실행됨

if []:
    print("실행 안 됨")
else:
    print("빈 리스트는 Falsy")   # ← 실행됨
```

```python
# Truthy 값
if 42:
    print("0이 아닌 정수는 Truthy")  # ← 실행됨

if "hello":
    print("비어있지 않은 문자열은 Truthy")  # ← 실행됨

if [1, 2]:
    print("요소가 있는 리스트는 Truthy")    # ← 실행됨
```

```python
# 실용적인 예시: 기본값 설정
name = ""
display = name or "익명"
print(f"이름: {display}")      # 익명 (빈 문자열은 Falsy)

name = "김파이"
display = name or "익명"
print(f"이름: {display}")      # 김파이
```
