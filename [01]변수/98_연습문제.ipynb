## 정답 확인

아래 셀을 실행하여 모든 문제의 정답을 확인하세요.

````python
# 정답

# ---- 문제 1 정답 ----
valid_names = ['my_var', '_private']
invalid_reasons = {
    '2nd_place': '숫자로 시작할 수 없음',
    'class': '파이썬 예약어(keyword)임',
    'my-var': '하이픈(-)은 변수 이름에 사용할 수 없음 (빼기 연산자와 혼동)',
}

# ---- 문제 2 정답 ----
def convert_values(data):
    result = []
    for item in data:
        if isinstance(item, int) and not isinstance(item, bool):
            result.append(float(item))
        elif isinstance(item, float):
            result.append(int(item))
        elif isinstance(item, str):
            try:
                result.append(float(item))
            except ValueError:
                result.append(item)
        elif isinstance(item, bool):
            result.append(int(item))
        else:
            result.append(item)
    return result

# ---- 문제 3 정답 ----
def analyze_text(text):
    words = text.split()
    upper = sum(1 for c in text if c.isupper())
    lower = sum(1 for c in text if c.islower())
    digits = sum(1 for c in text if c.isdigit())
    chars_no_space = len(text.replace(' ', ''))
    reversed_words = ' '.join(reversed(words))
    return {
        'words': len(words),
        'chars': chars_no_space,
        'upper': upper,
        'lower': lower,
        'digits': digits,
        'reversed_words': reversed_words,
    }

# ---- 문제 4 정답 ----
def slice_magic(lst):
    even_idx = lst[::2]
    odd_vals = [x for x in lst if x % 2 != 0]
    top3_rev = lst[::-1][:3]
    unique_sorted = sorted(set(lst))
    return (even_idx, odd_vals, top3_rev, unique_sorted)

# ---- 문제 5 정답 ----
def merge_student_records(records):
    result = {}
    for r in records:
        name = r['name']
        if name not in result:
            result[name] = {'scores': {}, 'average': 0, 'highest': 0}
        result[name]['scores'][r['subject']] = r['score']
    for name in result:
        scores = list(result[name]['scores'].values())
        result[name]['average'] = round(sum(scores) / len(scores), 2)
        result[name]['highest'] = max(scores)
    return result

# ---- 문제 6 정답 ----
def set_operations(a, b):
    sa, sb = set(a), set(b)
    return {
        'union': sorted(sa | sb),
        'intersection': sorted(sa & sb),
        'diff_a_b': sorted(sa - sb),
        'diff_b_a': sorted(sb - sa),
        'symmetric_diff': sorted(sa ^ sb),
        'is_subset': sa.issubset(sb),
    }

# ---- 문제 7 정답 ----
def format_report(data):
    name = data['name']
    id_ = data['id']
    score = data['score']
    ratio = data['ratio']
    items = data['items']
    return (
        f"[ID:{id_:04d}] {name}님\n"
        f"점수: {score:.1f}점 ({score:.2f}%)\n"
        f"비율: {ratio:.1%}\n"
        f"구매 항목: {', '.join(items)}"
    )

# ---- 문제 8 정답 ----
def unpack_data(items):
    names = []
    avg_scores = []
    heights = []
    bmi_list = []
    for name, (kor, eng, math), (height, weight) in items:
        names.append(name)
        avg_scores.append(round((kor + eng + math) / 3, 2))
        heights.append(height)
        bmi = round(weight / (height / 100) ** 2, 1)
        bmi_list.append(bmi)
    tallest = names[heights.index(max(heights))]
    return {
        'names': names,
        'avg_scores': avg_scores,
        'tallest': tallest,
        'bmi_list': bmi_list,
    }

# ---- 문제 9 정답 ----
def encode_decode(text, encoding='utf-8'):
    byte_data = text.encode(encoding)
    hex_str = byte_data.hex()
    decoded = byte_data.decode(encoding)
    match = (text == decoded)
    return {
        'bytes_len': len(byte_data),
        'hex_str': hex_str,
        'decoded': decoded,
        'match': match,
    }

# ---- 문제 10 정답 ----
def truth_table_demo():
    short_circuit = {
        'a': 0 or 42,
        'b': '' or 'hello',
        'c': 100 and 'yes',
        'd': 0 and 'no',
    }
    falsy = [bool(0), bool(''), bool([]), bool({}), bool(None)]
    truthy = [bool(1), bool('a'), bool([0]), bool({'a': 1})]
    compound = not (True and False) or (False or True)
    return {
        'short_circuit': short_circuit,
        'falsy': falsy,
        'truthy': truthy,
        'compound': compound,
    }

# ---- 전체 테스트 실행 ----
print('=' * 50)
print('문제 1: 변수 명명 규칙')
print('올바른 변수 이름:', valid_names)
print('잘못된 이름 이유:', invalid_reasons)

print('\n' + '=' * 50)
print('문제 2: 타입 변환')
print(convert_values([42, 3.14, 'hello', '99', True, False, None]))

print('\n' + '=' * 50)
print('문제 3: 문자열 메서드')
print(analyze_text('Hello World 2024'))

print('\n' + '=' * 50)
print('문제 4: 리스트 슬라이싱')
print(slice_magic([3, 1, 4, 1, 5, 9, 2, 6, 5, 3]))

print('\n' + '=' * 50)
print('문제 5: 딕셔너리 연산')
import json
records = [
    {'name': '김철수', 'subject': '수학', 'score': 90},
    {'name': '김철수', 'subject': '영어', 'score': 85},
    {'name': '이영희', 'subject': '수학', 'score': 95},
    {'name': '이영희', 'subject': '영어', 'score': 88},
    {'name': '박민수', 'subject': '수학', 'score': 78},
]
print(json.dumps(merge_student_records(records), ensure_ascii=False, indent=2))

print('\n' + '=' * 50)
print('문제 6: 집합 연산')
print(set_operations([1, 2, 3, 4], [3, 4, 5, 6]))

print('\n' + '=' * 50)
print('문제 7: f-string 포매팅')
data = {'name': '김철수', 'id': 42, 'score': 87.6543, 'ratio': 0.1234, 'items': ['사과', '배', '포도']}
print(format_report(data))

print('\n' + '=' * 50)
print('문제 8: 튜플 언패킹')
items = [
    ('김철수', (90, 85, 92), (175, 70)),
    ('이영희', (88, 92, 95), (162, 55)),
    ('박민수', (78, 80, 85), (180, 82)),
]
result = unpack_data(items)
print(result)

print('\n' + '=' * 50)
print('문제 9: 바이트 인코딩/디코딩')
print('ASCII:', encode_decode('Hello'))
print('한글 UTF-8:', encode_decode('안녕하세요', 'utf-8'))
print('한글 EUC-KR:', encode_decode('안녕하세요', 'euc-kr'))

print('\n' + '=' * 50)
print('문제 10: 불리언 연산')
result = truth_table_demo()
print(result)

print('\n' + '=' * 50)
print('모든 정답 확인 완료!')
````
