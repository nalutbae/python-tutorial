# 📘 변수와 자료형

이 노트북에서는 파이썬의 **변수(variable)**와 **자료형(data type)**의 기초를 배웁니다.

**학습 목표:**
- 변수를 선언하고 값을 할당하는 방법
- 파이썬의 기본 자료형 이해
- `type()` 함수로 자료형 확인하기
- 변수 이름 규칙

## 1. 변수란?

변수는 **데이터를 저장하는 이름표**입니다. 파이썬에서는 `=` 기호를 사용해 변수에 값을 할당합니다.

```python
변수이름 = 값
```

파이썬은 동적 타이핑 언어이므로, 변수의 자료형을 미리 선언할 필요가 없습니다.

````python
# 변수에 값 할당하기
name = "김파이"        # 문자열(string)
age = 25               # 정수(integer)
height = 175.5         # 실수(float)
is_student = True      # 불리언(boolean)
````

````python
# 변수 출력
print(f"이름: {name}")
print(f"나이: {age}")
print(f"키: {height}")
print(f"학생 여부: {is_student}")
````

````python
# type() 함수로 자료형 확인
print(f"\nname의 자료형: {type(name)}")
print(f"age의 자료형: {type(age)}")
print(f"height의 자료형: {type(height)}")
print(f"is_student의 자료형: {type(is_student)}")
````

## 2. 변수 이름 규칙

파이썬 변수 이름은 다음 규칙을 따라야 합니다:

- ✅ 문자(영문, 한글, 유니코드), 숫자, 밑줄(`_`) 사용 가능
- ❌ 숫자로 시작할 수 없음
- ❌ 파이썬 키워드(`if`, `for`, `class` 등) 사용 불가
- ❌ 공백, 특수문자 사용 불가
- 관례: `snake_case` (단어 사이 밑줄) 권장

````python
# ✅ 올바른 변수 이름
my_name = "파이썬"       # 밑줄 사용
age2 = 30               # 숫자 포함 (시작은 아님)
_private = 42           # 밑줄로 시작
변수 = "한글도 가능"     # 유니코드 문자

print(f"my_name = {my_name}")
print(f"age2 = {age2}")
print(f"_private = {_private}")
print(f"변수 = {변수}")

# ❌ 잘못된 변수 이름 (주석을 해제하면 에러 발생)
# 2name = "에러"        # SyntaxError: 숫자로 시작
# my-name = "에러"     # SyntaxError: 하이픈 사용 불가
# class = "에러"        # SyntaxError: 키워드 사용 불가
````

````python
# 파이썬 키워드 확인하기
import keyword

print(f"파이썬 키워드 개수: {len(keyword.kwlist)}")
print(f"\n키워드 목록:")
print(keyword.kwlist)

# 특정 단어가 키워드인지 확인
print(f"\n'if'는 키워드인가? {keyword.iskeyword('if')}")
print(f"'my_var'는 키워드인가? {keyword.iskeyword('my_var')}")
````

## 3. 변수 재할당과 자료형 변경

파이썬 변수는 언제든 다른 값을 할당할 수 있습니다. **다른 자료형의 값**도 할당 가능합니다.

````python
# 변수 재할당
x = 10
print(f"x = {x}, 자료형: {type(x)}")

x = 3.14          # 정수 → 실수
print(f"x = {x}, 자료형: {type(x)}")

x = "hello"       # 실수 → 문자열
print(f"x = {x}, 자료형: {type(x)}")
# 여러 변수 동시 할당
a, b, c = 1, 2, 3
print(f"\na={a}, b={b}, c={c}")
# 같은 값을 여러 변수에 할당
x = y = z = 0
print(f"x={x}, y={y}, z={z}")
````

````python
# 변수 값 교환 (swap) — 파이썬의 편리한 문법
a, b = 10, 20
print(f"\n교환 전: a={a}, b={b}")
a, b = b, a
print(f"교환 후: a={a}, b={b}")
````

## 4. None — 값이 없음을 나타내는 특수값

`None`은 "값이 없음"을 나타내는 파이썬의 특수값입니다.
다른 언어의 `null`과 비슷합니다.

````python
# None: 값이 할당되지 않았음을 나타냄
result = None
print(f"result = {result}")      # None
print(f"type(result) = {type(result)}")  # <class 'NoneType'>
# None인지 확인할 때는 'is' 연산자 사용
if result is None:
    print("result에 아직 값이 없습니다.")
````

````python
# 일반적인 사용 사례: 함수의 기본값
def greet(name=None):
    """name이 None이면 기본 인사, 아니면 이름으로 인사"""
    if name is None:
        return "안녕하세요!"
    return f"{name}님, 안녕하세요!"

print(greet())           # "안녕하세요!"
print(greet("김파이"))   # "김파이님, 안녕하세요!"
````

## 5. 자료형 요약

| 자료형 | 표기 | 예시 | 특징 |
|--------|------|------|------|
| 정수 | `int` | `42`, `-7`, `0` | 소수점 없는 수 |
| 실수 | `float` | `3.14`, `-0.5` | 소수점 있는 수 |
| 불리언 | `bool` | `True`, `False` | 참/거짓 |
| 문자열 | `str` | `"hello"`, `'파이썬'` | 텍스트 |
| 리스트 | `list` | `[1, 2, 3]` | 순서 있음, 변경 가능 |
| 튜플 | `tuple` | `(1, 2, 3)` | 순서 있음, 변경 불가 |
| 집합 | `set` | `{1, 2, 3}` | 순서 없음, 중복 불가 |
| 딕셔너리 | `dict` | `{"a": 1}` | 키-값 쌍 |
| 바이트 | `bytes` | `b"hello"` | 바이트 데이터 |
| None | `NoneType` | `None` | 값 없음

````python
# 자료형별 예시 한눈에 보기
examples = {
    "int (정수)": 42,
    "float (실수)": 3.14,
    "bool (불리언)": True,
    "str (문자열)": "파이썬",
    "list (리스트)": [1, 2, 3],
    "tuple (튜플)": (1, 2, 3),
    "set (집합)": {1, 2, 3},
    "dict (딕셔너리)": {"key": "value"},
    "bytes (바이트)": b"hello",
    "NoneType": None,
}

for name, value in examples.items():
    print(f"{name:20s} → 값: {str(value):20s} 자료형: {type(value).__name__}")
````
