# 📘 불리언(bool)과 실수(float)

파이썬의 **참/거짓**을 나타내는 `bool`型和 **소수점이 있는 수**인 `float`을 배웁니다.

**학습 목표:**
- `True`/`False`의 의미와 연산
- `bool()` 함수로 다른 자료형을 불리언으로 변환
- 실수의 연산과 부동소수점 오차
- 무한대(inf)와 NaN

## 1. 불리언(Boolean) 기초

`True`와 `False`는 파이썬의 예약어입니다.
내부적으로 `True = 1`, `False = 0`으로 처리됩니다.

```python
# 불리언 변수
x = True
y = False

print(f"x = {x}")    # True
print(f"y = {y}")    # False
```

```python
# 불리언은 정수와 연산 가능 (True=1, False=0)
print(f"x + y = {x + y}")    # 1 (True+False = 1+0)
print(f"x - y = {x - y}")    # 1 (True-False = 1-0)
print(f"x * y = {x * y}")    # 0 (True*False = 1*0)
print(f"x ** y = {x ** y}")  # 1 (True**False = 1**0)
```

```python
# ┌───────────────────────────────────────┐
# │  불리언 연산자                          │
# │  and: 둘 다 참이면 참                   │
# │  or: 하나라도 참이면 참                  │
# │  not: 참->거짓, 거짓->참                 │
# │  and/or은 마지막으로 평가된 값을 반환     │
# └───────────────────────────────────────┘

print(f"x and y = {x and y}")   # False
print(f"x or y  = {x or y}")    # True
print(f"not x   = {not x}")     # False
print(f"not y   = {not y}")     # True
# 연산자 우선순위: not > and > or
print(f"True or True and False = {True or True and False}")    # True
print(f"True or (True and False) = {True or (True and False)}")  # True (동일 결과)
```

## 2. bool() — 다른 자료형을 불리언으로 변환

파이썬에서 **0, 빈 컨테이너, None, 빈 문자열**은 `False`로 평가됩니다.
나머지는 모두 `True`입니다.

| 값 | bool() 결과 |
|------|----------|
| `0`, `0.0`, `0j` | `False` |
| ``, `[]`, `{}`, `()` | `False` |
| `None` | `False` |
| 나머지 모든 값 | `True` |

```python
# bool() 변환 — 거짓(False)으로 평가되는 값들
print("=== False로 평가되는 값들 ===")
print(f"bool(0)     = {bool(0)}")
print(f"bool(0.0)   = {bool(0.0)}")
print(f"bool(0j)    = {bool(0j)}")
print(f"bool('')    = {bool('')}")
print(f"bool([])    = {bool([])}")
print(f"bool({})    = {bool({})}")
print(f"bool(())    = {bool(())}")
print(f"bool(None)  = {bool(None)}")
```

```python
# True로 평가되는 값들
print("\n=== True로 평가되는 값들 ===")
print(f"bool(1)      = {bool(1)}")
print(f"bool(-1)     = {bool(-1)}")   # 음수도 True!
print(f"bool(0.1)    = {bool(0.1)}")
print(f"bool('hello') = {bool('hello')}")
print(f"bool([0])    = {bool([0])}")   # 0이 들어있어도 리스트는 비어있지 않음
print(f"bool(' ')    = {bool(' ')}")   # 공백도 True!
```

## 3. 실수(float) 연산과 부동소수점 오차

컴퓨터는 실수를 **부동소수점(IEEE 754)**으로 표현합니다.
이로 인해 `0.1 + 0.2 == 0.3`이 `False`가 되는 현상이 발생합니다.

⚠️ 실수 비교는 `==` 대신 허용 오차(epsilon)를 사용하세요.

```python
# 실수 변수
x = 0.1
y = 3.14
z = 30

print(f"x = {x}")     # 0.1
print(f"y = {y}")     # 3.14
print(f"x + y = {x + y}")   # 3.24
print(f"x - y = {x - y}")   # -3.04
print(f"x * y = {x * y}")   # 0.31400000000000006 ← 부동소수점 오차!
print(f"x / y = {x / y}")   # 0.031847...
print(f"x // y = {x // y}")  # 0.0
print(f"x % y = {x % y}")    # 0.1
print(f"x ** y = {x ** y}")   # 0.000724...
# ┌─────────────────────────────────────────────────┐
# │  부동소수점 오차 (Floating Point Error)           │
# │  0.1 + 0.2 가 정확히 0.3 이 아닙니다!            │
# │  해결: abs(a - b) < epsilon 으로 비교              │
# └─────────────────────────────────────────────────┘

print(f"\n0.1 + 0.2 = {0.1 + 0.2}")          # 0.30000000000000004
print(f"0.1 + 0.2 == 0.3 -> {0.1 + 0.2 == 0.3}")  # False!
```

```python
# 올바른 실수 비교 방법
epsilon = 1e-10  # 허용 오차
result = 0.1 + 0.2
print(f"abs(result - 0.3) < epsilon -> {abs(result - 0.3) < epsilon}")  # True
# math.isclose() 사용 (권장)
import math
print(f"math.isclose(0.1 + 0.2, 0.3) -> {math.isclose(0.1 + 0.2, 0.3)}")  # True
```

## 4. 무한대와 NaN

파이썬의 `float`은 **무한대(inf)**와 **NaN(Not a Number)**을 표현할 수 있습니다.

```python
# 무한대 (Infinity)
x = float("inf")
y = -float("inf")

print(f"양의 무한대: {x}")
print(f"음의 무한대: {y}")
print(f"무한대 + 100 = {x + 100}")     # 여전히 inf
print(f"무한대 + 음의 무한대 = {x + y}")  # nan (정의 불가)
```

```python
# NaN (Not a Number)
nan_val = float("nan")
print(f"\nNaN: {nan_val}")
print(f"NaN == NaN -> {nan_val == nan_val}")  # False! (NaN은 자기 자신과도 같지 않음)
print(f"math.isnan(nan_val) -> {math.isnan(nan_val)}")  # True
```

```python
# ┌─────────────────────────────────────────┐
# │  무한대와 NaN의 특징                       │
# │  inf + 어떤 수 = inf                      │
# │  inf + (-inf) = nan                       │
# │  NaN != NaN (자기 자신과도 다름)            │
# │  NaN 확인은 math.isnan() 사용              │
# └─────────────────────────────────────────┘

# 실수의 특수 값 확인
print(f"\nmath.isinf({x}) -> {math.isinf(x)}")   # True
print(f"math.isinf({y}) -> {math.isinf(y)}")    # True
print(f"math.isfinite(3.14) -> {math.isfinite(3.14)}")  # True
print(f"math.isfinite({x}) -> {math.isfinite(x)}")    # False
```
