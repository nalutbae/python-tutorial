# 📘 리스트(list)

파이썬의 **리스트**는 순서가 있는 변경 가능한(mutable) 시퀀스입니다.
가장 많이 사용되는 자료형 중 하나입니다.

**학습 목표:**
- 리스트 생성과 인덱싱/슬라이싱
- 요소 추가, 수정, 삭제
- 리스트 정렬과 검색
- 리스트 컴프리헨션

## 1. 리스트 생성과 인덱싱

리스트는 대괄호 `[]`로 만들며, 다양한 자료형을 섞어 넣을 수 있습니다.

````python
# 리스트 생성
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

print(f"리스트: {fruits}")
print(f"자료형: {type(fruits)}")
print(f"길이: {len(fruits)}")
````

````python
# ┌─────────────────────────────────────────────┐
# │  인덱싱 (Indexing)                            │
# │  양수 인덱스: 0부터 시작 (왼쪽→오른쪽)          │
# │  음수 인덱스: -1부터 시작 (오른쪽→왼쪽)          │
# │                                              │
# │  ["apple", "banana", "cherry", "date", "elder"]│
# │    0       1         2        3       4        │
# │   -5      -4        -3       -2      -1        │
# └─────────────────────────────────────────────┘

print(f"fruits[0]  = {fruits[0]}")    # apple
print(f"fruits[2]  = {fruits[2]}")    # cherry
print(f"fruits[-1] = {fruits[-1]}")   # elderberry
print(f"fruits[-2] = {fruits[-2]}")   # date
````

````python
# 슬라이싱
print(f"fruits[1:3]  = {fruits[1:3]}")   # ['banana', 'cherry']
print(f"fruits[2:]   = {fruits[2:]}")     # ['cherry', 'date', 'elderberry']
print(f"fruits[:3]   = {fruits[:3]}")     # ['apple', 'banana', 'cherry']
# 리스트 연산
print(f"fruits * 2 = {fruits * 2}")          # 반복
print(f"fruits + ['fig'] = {fruits + ['fig']}")  # 결합
````

## 2. 요소 추가, 수정, 삭제

리스트는 **변경 가능(mutable)**하므로 생성 후에도 요소를 수정할 수 있습니다.

````python
# 요소 수정
nums = [1, 2, 3, 4, 5]
print(f"원본: {nums}")

nums[0] = 10        # 단일 요소 수정
print(f"nums[0] = 10 후: {nums}")

nums[1:3] = [20, 30]  # 슬라이스 수정
print(f"nums[1:3] = [20, 30] 후: {nums}")

nums[1:3] = []       # 슬라이스로 요소 삭제
print(f"nums[1:3] = [] 후: {nums}")
````

````python
# ┌─────────────────────────────────────────┐
# │  요소 추가 메서드                          │
# │  append(x):  끝에 하나 추가               │
# │  insert(i,x): 인덱스 i에 x 삽입           │
# │  extend(iter): iterable을 끝에 추가        │
# └─────────────────────────────────────────┘

nums.append(40)
print(f"append(40) 후: {nums}")

nums.insert(0, 60)
print(f"insert(0, 60) 후: {nums}")

nums.extend([70, 80])
print(f"extend([70, 80]) 후: {nums}")
````

````python
# ┌─────────────────────────────────────────┐
# │  요소 삭제 메서드                          │
# │  remove(x):  값 x를 찾아 삭제             │
# │  pop():      마지막 요소 제거 후 반환       │
# │  pop(i):     인덱스 i 요소 제거 후 반환     │
# │  del lst[i]: 인덱스 i 요소 삭제            │
# │  clear():    모든 요소 삭제                │
# └─────────────────────────────────────────┘

nums.remove(70)
print(f"remove(70) 후: {nums}")

removed = nums.pop()
print(f"pop() 후: {nums}, 제거된 값: {removed}")

removed = nums.pop(0)
print(f"pop(0) 후: {nums}, 제거된 값: {removed}")
````

## 3. 리스트 정렬과 검색

리스트의 요소를 정렬하거나 특정 값을 찾을 수 있습니다.

````python
# 정렬
nums = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"원본: {nums}")
````

````python
# ┌───────────────────────────────────────────────┐
# │  sort() vs sorted()                            │
# │  sort():    리스트 자체를 정렬 (원본 변경)        │
# │  sorted(): 정렬된 새 리스트 반환 (원본 유지)     │
# └───────────────────────────────────────────────┘

nums_sorted = sorted(nums)       # 새 리스트 반환
print(f"sorted(nums): {nums_sorted}")
print(f"원본 유지: {nums}")          # 원본 그대로

nums.sort()                       # 리스트 자체 정렬
print(f"nums.sort() 후: {nums}")    # 원본이 변경됨

nums.sort(reverse=True)          # 내림차순
print(f"내림차순: {nums}")
````

````python
# 검색
nums = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"\n리스트: {nums}")
print(f"1의 위치(index): {nums.index(1)}")    # 1 (첫 번째 위치)
print(f"1의 개수(count): {nums.count(1)}")    # 2
print(f"9가 있는지: {9 in nums}")              # True
print(f"7이 있는지: {7 in nums}")              # False
````

````python
# 최댓값, 최솟값, 합계
print(f"\nmax(nums) = {max(nums)}")
print(f"min(nums) = {min(nums)}")
print(f"sum(nums) = {sum(nums)}")
````

## 4. 리스트 컴프리헨션(List Comprehension)

리스트 컴프리헨션은 리스트를 간결하게 생성하는 파이썬 특유의 문법입니다.

```python
[표현식 for 변수 in 반복가능객체 if 조건]
```

````python
# 기본 컴프리헨션
squares = [x ** 2 for x in range(1, 6)]
print(f"x^2 for x in 1..5: {squares}")  # [1, 4, 9, 16, 25]
# 조건 필터링
evens = [x for x in range(1, 11) if x % 2 == 0]
print(f"1~10 짝수: {evens}")  # [2, 4, 6, 8, 10]
````

````python
# 중첩 for문 (2차원 리스트 평탄화)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [x for row in matrix for x in row]
print(f"평탄화: {flat}")  # [1, 2, 3, 4, 5, 6, 7, 8, 9]
````

````python
# ┌───────────────────────────────────────────┐
# │  컴프리헨션 vs for문 비교                   │
# │  컴프리헨션이 더 간결하고 빠릅니다             │
# │  하지만 너무 복잡하면 가독성이 떨어집니다       │
# └───────────────────────────────────────────┘

# for문 버전 (동일 결과)
evens_for = []
for x in range(1, 11):
    if x % 2 == 0:
        evens_for.append(x)
print(f"for문 버전: {evens_for}")
# 딕셔너리 컴프리헨션도 가능
word_len = {w: len(w) for w in ["apple", "banana", "cherry"]}
print(f"단어 길이: {word_len}")
````

````python
# enumerate와 함께 사용
words = ["apple", "banana", "cherry"]
indexed = [(i, w) for i, w in enumerate(words)]
print(f"인덱스와 함께: {indexed}")
````

## 5. 리스트 복사와 얕은/깊은 복사

리스트를 대입하면 **같은 객체**를 참조합니다.
독립적인 복사본을 만들려면 `copy()`나 ` deepcopy()`를 사용하세요.

````python
# 얕은 복사 (shallow copy) vs 대입
a = [1, 2, 3]
b = a          # 대입: 같은 객체를 참조!
b[0] = 99
print(f"a = {a}")  # [99, 2, 3] — a도 변경됨!
print(f"b = {b}")  # [99, 2, 3]
````

````python
# ┌─────────────────────────────────────────┐
# │  대입(=)은 복사가 아닙니다!                 │
# │  같은 리스트 객체를 가리키게 됩니다           │
# │  독립적인 복사본을 만들려면 copy() 사용     │
# └─────────────────────────────────────────┘

# 얕은 복사 (1단계만 독립)
c = [1, 2, 3]
d = c.copy()    # 또는 d = list(c) 또는 d = c[:]
d[0] = 99
print(f"\nc = {c}")  # [1, 2, 3] — c는 그대로!
print(f"d = {d}")  # [99, 2, 3]
````

````python
# 중첩 리스트의 얕은 복사 문제
nested = [[1, 2], [3, 4]]
shallow = nested.copy()
shallow[0][0] = 99
print(f"\nnested = {nested}")  # [[99, 2], [3, 4]] — 원본도 변경됨!
````

````python
# 깊은 복사 (deep copy) — 중첩 구조까지 완전 복사
import copy
nested2 = [[1, 2], [3, 4]]
deep = copy.deepcopy(nested2)
deep[0][0] = 99
print(f"nested2 = {nested2}")  # [[1, 2], [3, 4]] — 원본 유지!
print(f"deep = {deep}")        # [[99, 2], [3, 4]]
````
