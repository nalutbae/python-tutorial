# 📘 정수(int)와 연산자

파이썬의 **정수형(int)**과 다양한 **연산자**를 배웁니다.

**학습 목표:**
- 정수 변수 선언 및 기본 산술 연산
- 비교 연산자와 논리 연산자
- 여러 조건 동시 비교
- 정수의 특수 표기법(2진수, 8진수, 16진수)

## 1. 정수 변수와 산술 연산자

파이썬은 정수끼리의 연산에서도 `/`(나눗셈)는 항상 실수를 반환합니다.
정수 나눗셈은 `//`를, 나머지는 `%`를 사용합니다.

```python
# 정수 변수 선언
x = 10
y = 20
z = 30

print(f"x = {x}")  # 10
print(f"y = {y}")  # 20
```

```python
# ┌───────────────────────────────────────┐
# │  산술 연산자 (Arithmetic Operators)    │
# │  + : 덧셈        - : 뺄셈             │
# │  * : 곱셈        / : 나눗셈(실수)     │
# │  //: 정수 나눗셈  %: 나머지            │
# │  **: 거듭제곱                          │
# └───────────────────────────────────────┘

print(f"x + y = {x + y}")     # 30
print(f"x - y = {x - y}")     # -10
print(f"x * y = {x * y}")     # 200
print(f"x / y = {x / y}")     # 0.5  ← 항상 float 반환!
print(f"x // y = {x // y}")   # 0    ← 정수 나눗셈(몫)
print(f"x % y = {x % y}")     # 10   ← 나머지
print(f"x ** y = {x ** y}")   # 10^20 = 100000000000000000000
```

## 2. 비교 연산자

비교 연산자는 **참(True)** 또는 **거짓(False)**을 반환합니다.

| 연산자 | 의미 | 예시 |
|--------|------|------|
| `==` | 같다 | `10 == 10` → True |
| `!=` | 다르다 | `10 != 20` → True |
| `>` | 크다 | `20 > 10` → True |
| `<` | 작다 | `10 < 20` → True |
| `>=` | 크거나 같다 | `10 >= 10` → True |
| `<=` | 작거나 같다 | `10 <= 20` → True |

```python
x = 10
y = 20
```

```python
# 비교 연산자
print(f"x == y -> {x == y}")   # False
print(f"x != y -> {x != y}")   # True
print(f"x > y  -> {x > y}")    # False
print(f"x < y  -> {x < y}")    # True
print(f"x >= y -> {x >= y}")   # False
print(f"x <= y -> {x <= y}")   # True
```

```python
# ┌───────────────────────────────────────────┐
# │  연쇄 비교 (Chained Comparison)            │
# │  x < y < z 은 (x < y) and (y < z) 과 동일  │
# │  여러 조건을 한 번에 비교할 수 있습니다       │
# └───────────────────────────────────────────┘

z = 30
print(f"x < y < z -> {x < y < z}")              # True
print(f"(x < y) and (y < z) -> {(x < y) and (y < z)}")  # True (동일)
```

## 3. 논리 연산자와 항등 연산자

- **논리 연산자**: `and`, `or`, `not`
- **항등 연산자**: `is`, `is not` (객체가 같은지 비교)

⚠️ `==`은 **값**을 비교, `is`는 **객체(메모리 주소)**를 비교합니다.

```python
x = 10
y = 20
```

```python
# 논리 연산자
# ┌──────────────────────────────────────────────┐
# │  and: 둘 다 참이면 참    or: 하나라도 참이면 참  │
# │  not: 참->거짓, 거짓->참                        │
# │  주의: and/or은 마지막으로 평가된 값을 반환함    │
# └──────────────────────────────────────────────┘

print(f"x and y -> {x and y}")     # 20 (둘 다 참이면 y 반환)
print(f"x or y  -> {x or y}")      # 10 (둘 다 참이면 x 반환)
print(f"not x   -> {not x}")       # False (0이 아니면 참)
print(f"not y   -> {not y}")       # False
# 항등 연산자 (identity)
print(f"x is y     -> {x is y}")      # False (다른 객체)
print(f"x is not y -> {x is not y}")  # True (다른 객체)
```

```python
# == vs is 의 차이
a = 256
b = 256
print(f"a == b -> {a == b}")  # True (값이 같음)
print(f"a is b -> {a is b}")  # True (작은 정수는 캐싱됨)

c = 257
d = 257
print(f"c == d -> {c == d}")  # True (값이 같음)
print(f"c is d -> {c is d}")  # False! (257은 캐싱 범위 밖)
```

## 4. 진수 표기법

파이썬은 2진수, 8진수, 16진수 리터럴을 지원합니다.

| 접두사 | 진수 | 예시 | 10진수 값 |
|--------|------|------|----------|
| `0b` | 2진수 | `0b1010` | 10 |
| `0o` | 8진수 | `0o12` | 10 |
| `0x` | 16진수 | `0xA` | 10 |

```python
# 다양한 진수 표기
binary = 0b1010      # 2진수: 1*8 + 0*4 + 1*2 + 0*1 = 10
octal = 0o12         # 8진수: 1*8 + 2*1 = 10
hexadecimal = 0xA    # 16진수: A = 10
decimal = 10

print(f"2진수  0b1010 = {binary}")
print(f"8진수  0o12   = {octal}")
print(f"10진수 10     = {decimal}")
print(f"16진수 0xA    = {hexadecimal}")
```

```python
# 진법 변환 함수
print(f"bin(10) = {bin(10)}")    # '0b1010'
print(f"oct(10) = {oct(10)}")    # '0o12'
print(f"hex(10) = {hex(10)}")    # '0xa'
```

```python
# int() 함수로 진법 변환
print(f"int('1010', 2) = {int('1010', 2)}")  # 2진수 -> 10진수
print(f"int('12', 8) = {int('12', 8)}")       # 8진수 -> 10진수
print(f"int('A', 16) = {int('A', 16)}")       # 16진수 -> 10진수
```

## 5. 할당 연산자와 비트 연산자

할당 연산자는 변수 값을 갱신하는 짧은 표기법입니다.
비트 연산자는 정수의 이진 표현을 직접 조작합니다.

```python
# 할당 연산자 (Augmented Assignment)
x = 10

x += 5    # x = x + 5 = 15
print(f"x += 5 -> {x}")

x -= 3    # x = x - 3 = 12
print(f"x -= 3 -> {x}")

x *= 2    # x = x * 2 = 24
print(f"x *= 2 -> {x}")

x //= 5   # x = x // 5 = 4
print(f"x //= 5 -> {x}")

x **= 3   # x = x ** 3 = 64
print(f"x **= 3 -> {x}")
```

```python
# ┌─────────────────────────────────────────┐
# │  비트 연산자 (Bitwise Operators)         │
# │  & : AND   | : OR   ^ : XOR            │
# │  ~ : NOT   <<: 좌측 시프트  >>: 우측 시프트  │
# │  정수의 이진 표현을 직접 조작합니다          │
# └─────────────────────────────────────────┘

a = 0b1100   # 12
b = 0b1010   # 10

print(f"a = {a} (0b{a:04b})")
print(f"b = {b} (0b{b:04b})")
print(f"a & b  = {a & b}  (0b{a & b:04b})")   # 0b1000 = 8
print(f"a | b  = {a | b}  (0b{a | b:04b})")   # 0b1110 = 14
print(f"a ^ b  = {a ^ b}  (0b{a ^ b:04b})")   # 0b0110 = 6
print(f"a << 1 = {a << 1} (0b{a << 1:08b})")  # 0b00011000 = 24
print(f"a >> 1 = {a >> 1} (0b{a >> 1:04b})")   # 0b0110 = 6
```
