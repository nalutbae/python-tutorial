# 📘 딕셔너리(dict)와 바이트(bytes)

**딕셔너리**는 키-값 쌍의 컬렉션이고, **바이트**는 이진 데이터를 다루는 자료형입니다.

**학습 목표:**
- 딕셔너리 생성, 접근, 수정, 삭제
- 딕셔너리 컴프리헨션
- 바이트 자료형의 생성과 변환

## 1. 딕셔너리 생성과 기본 조작

딕셔너리는 중괄호 `{}`로 만들며, **키:값** 쌍을 저장합니다.
리스트가 인덱스로 접근한다면, 딕셔너리는 **키**로 접근합니다.

```python
# 딕셔너리 생성
person = {
    "name": "김파이",
    "age": 25,
    "city": "서울"
}

print(f"딕셔너리: {person}")
print(f"name: {person['name']}")       # 키로 접근
print(f"age: {person['age']}")
```

```python
# ┌─────────────────────────────────────────────┐
# │  get() vs [] 접근                            │
# │  person['phone'] → KeyError (키가 없으면)      │
# │  person.get('phone') → None (키가 없어도 안전)  │
# │  person.get('phone', '없음') → 기본값 지정 가능  │
# └─────────────────────────────────────────────┘

print(f"get('city'): {person.get('city')}")             # 서울
print(f"get('phone'): {person.get('phone')}")           # None
print(f"get('phone', '없음'): {person.get('phone', '없음')}")  # 없음
```

```python
# 키, 값, 아이템 조회
print(f"\nkeys(): {list(person.keys())}")     # ['name', 'age', 'city']
print(f"values(): {list(person.values())}")   # ['김파이', 25, '서울']
print(f"items(): {list(person.items())}")      # [('name', '김파이'), ...]
```

```python
# 요소 추가와 수정
person["email"] = "python@example.com"   # 새 키 추가
person["age"] = 26                        # 기존 키 수정
print(f"\n수정 후: {person}")
```

```python
# 요소 삭제
removed = person.pop("city")            # 키 삭제하고 값 반환
print(f"pop('city'): {removed}")           # 서울
print(f"삭제 후: {person}")
# dict() 생성자로 만들기
scores = dict(korean=90, english=85, math=95)
print(f"\ndict()로 생성: {scores}")
```

## 2. 딕셔너리 반복과 컴프리헨션

딕셔너리를 반복문에서 사용하거나, 컴프리헨션으로 간결하게 만들 수 있습니다.

```python
# 딕셔너리 반복
scores = {"korean": 90, "english": 85, "math": 95, "science": 88}
```

```python
# 키로 반복
print("=== 키 반복 ===")
for key in scores:
    print(f"  {key}: {scores[key]}")
```

```python
# items()로 키-값 쌍 반복 (가장 파이썬적인 방법)
print("\n=== items() 반복 ===")
for subject, score in scores.items():
    print(f"  {subject}: {score}점")
# ┌─────────────────────────────────────────────┐
# │  딕셔너리 컴프리헨션 (Dict Comprehension)     │
# │  {키_표현식: 값_표현식 for 변수 in 반복가능객체} │
# │  조건 추가: ... if 조건                       │
# └─────────────────────────────────────────────┘

# 제곱 딕셔너리
squares = {x: x ** 2 for x in range(1, 7)}
print(f"\n제곱: {squares}")
# 조건 필터링
even_squares = {x: x ** 2 for x in range(1, 7) if x % 2 == 0}
print(f"짝수 제곱: {even_squares}")
```

```python
# 두 리스트로 딕셔너리 만들기
subjects = ["korean", "english", "math"]
points = [90, 85, 95]
combined = {s: p for s, p in zip(subjects, points)}
print(f"zip으로 결합: {combined}")
# 값 변환
score_letters = {k: "A" if v >= 90 else "B" for k, v in scores.items()}
print(f"등급 변환: {score_letters}")
```

## 3. 딕셔너리 메서드 정리

| 메서드 | 설명 | 예시 |
|--------|------|------|
| `keys()` | 모든 키 반환 | `d.keys()` |
| `values()` | 모든 값 반환 | `d.values()` |
| `items()` | 모든 (키,값) 쌍 반환 | `d.items()` |
| `get(k)` | 키 k의 값 반환 (없으면 None) | `d.get("age")` |
| `get(k, v)` | 키 k의 값 반환 (없으면 v) | `d.get("age", 0)` |
| `pop(k)` | 키 k 삭제 후 값 반환 | `d.pop("age")` |
| `update(d2)` | 다른 딕셔너리와 병합 | `d.update({"a": 1})` |
| `setdefault(k, v)` | 키가 없으면 v 설정 후 반환 | `d.setdefault("age", 0)` |
| `clear()` | 모든 요소 삭제 | `d.clear()` |

```python
# 딕셔너리 메서드 실습
student = {"name": "김파이", "age": 20}
```

```python
# setdefault: 키가 없으면 기본값 설정
grade = student.setdefault("grade", "A")
print(f"setdefault 후: {student}")
print(f"반환값: {grade}")          # A (새로 설정됨)

grade2 = student.setdefault("grade", "B")  # 이미 있으므로 기존 값 반환
print(f"다시 setdefault: {grade2}")  # A (기존값 유지)
```

```python
# update: 다른 딕셔너리와 병합
extra = {"city": "서울", "age": 21}
student.update(extra)    # age는 21로 덮어씀
print(f"update 후: {student}")
# 키 존재 확인
print(f"\n'name' in student -> {'name' in student}")    # True
print(f"'phone' in student -> {'phone' in student}")    # False
```

```python
# popitem: 마지막 요소 제거 (Python 3.7+ 삽입 순서 유지)
last = student.popitem()
print(f"popitem(): {last}")
print(f"popitem 후: {student}")
```

```python
# ┌─────────────────────────────────────────┐
# │  딕셔너리 병합 (Python 3.9+)              │
# │  d1 | d2 → 두 딕셔너리를 병합한 새 딕셔너리 │
# │  d1 |= d2 → d1에 d2를 병합 (d1이 변경됨)  │
# └─────────────────────────────────────────┘

d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}
print(f"\nd1 | d2 = {d1 | d2}")   # {'a': 1, 'b': 3, 'c': 4} (d2가 우선)
```

```python
# 중첩 딕셔너리
school = {
    "student1": {"name": "김파이", "age": 20},
    "student2": {"name": "이파이", "age": 22},
}
print(f"\n중첩 딕셔너리: {school}")
print(f"student1 이름: {school['student1']['name']}")
```

## 4. 바이트(bytes) 자료형

바이트는 **이진 데이터**를 다루는 변경 불가능한 시퀀스입니다.
파일 I/O, 네트워크 통신, 암호화 등에서 사용됩니다.

```python
# 바이트 생성
b1 = bytes([0, 16, 32, 48, 64])   # 정수 리스트로 생성
print(f"bytes([0,16,32,48,64]): {b1}")
print(f"자료형: {type(b1)}")
```

```python
# 인덱싱 (정수로 반환)
print(f"b1[0] = {b1[0]}")    # 0
print(f"b1[1] = {b1[1]}")    # 16
print(f"b1[2:4] = {b1[2:4]}")  # b' 0'
# 16진수로 바이트 생성
b2 = bytes([0x00, 0x10, 0x20, 0x30, 0x40])
print(f"\n16진수 바이트: {b2}")
```

```python
# ┌─────────────────────────────────────────┐
# │  문자열 ↔ 바이트 변환                       │
# │  인코딩: 문자열 -> 바이트 (encode)          │
# │  디코딩: 바이트 -> 문자열 (decode)          │
# │  UTF-8이 가장 널리 쓰이는 인코딩 방식        │
# └─────────────────────────────────────────┘

# 문자열 → 바이트 (인코딩)
text = "Hello"
encoded = text.encode()             # 기본 UTF-8
print(f"\n'Hello'.encode() = {encoded}")
print(f"b'Hello' = {b'Hello'}")         # 바이트 리터럴
# 바이트 → 문자열 (디코딩)
decoded = encoded.decode()
print(f"b'Hello'.decode() = {decoded!r}")
```

```python
# 한글 인코딩
korean = "파이썬"
korean_bytes = korean.encode("utf-8")
print(f"\n'파이썬'.encode('utf-8') = {korean_bytes}")
print(f"디코딩: {korean_bytes.decode('utf-8')}")
```

```python
# ┌─────────────────────────────────────────┐
# │  bytes vs bytearray                       │
# │  bytes:     변경 불가 (immutable)          │
# │  bytearray: 변경 가능 (mutable)            │
# └─────────────────────────────────────────┘

ba = bytearray(b"Hello")
ba[0] = 74   # 'H' → 'J' (ASCII 74)
print(f"\nbytearray 수정: {ba}")       # bytearray(b'Jello')
print(f"bytearray → bytes: {bytes(ba)}")  # b'Jello'
```
