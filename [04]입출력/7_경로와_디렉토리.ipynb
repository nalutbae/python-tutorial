# 📘 경로와 디렉토리

파일 시스템의 **경로와 디렉토리**를 다루는 방법을 배웁니다.
`os.path`와 `pathlib.Path` 두 가지 방식을 비교합니다.

**학습 목표:**
- os.path로 경로 조작하기
- pathlib.Path로 현대적 경로 처리
- 디렉토리 생성/삭제/순회

## 1. os.path — 경로 조작

`os.path`는 파일 경로를 조작하는 전통적인 모듈입니다.
운영체제별 경로 구분자를 자동으로 처리합니다.

```python
import os

# ┌─────────────────────────────────────────┐
# │  os.path 주요 함수                         │
# │  join()    → 경로 결합                      │
# │  basename() → 파일명 추출                   │
# │  dirname()  → 디렉토리 추출                 │
# │  splitext() → 확장자 분리                   │
# │  exists()   → 경로 존재 여부                │
# │  isfile()   → 파일 여부                     │
# │  isdir()    → 디렉토리 여부                 │
# └─────────────────────────────────────────┘

# 경로 결합
path1 = os.path.join("folder", "subfolder", "file.txt")
print(f"결합: {path1}")

# 경로 분리
full_path = os.path.join(os.getcwd(), "data", "report.pdf")
print(f"전체 경로: {full_path}")
print(f"파일명: {os.path.basename(full_path)}")
print(f"디렉토리: {os.path.dirname(full_path)}")

# 확장자 분리
print(f"\n확장자 분리:")
print(f"  report.pdf → {os.path.splitext('report.pdf')}")
print(f"  archive.tar.gz → {os.path.splitext('archive.tar.gz')}")

# 경로 존재 확인
print(f"\n현재 디렉토리 존재: {os.path.exists(os.getcwd())}")
print(f"가짜 경로: {os.path.exists('/nonexistent/path')}")
```

## 2. pathlib — 현대적 경로 처리 (Python 3.4+)

`pathlib.Path`는 객체 지향적이고 직관적인 경로 처리를 제공합니다.
`/` 연산자로 경로를 결합할 수 있어 가독성이 좋습니다.

```python
from pathlib import Path
import tempfile

# ┌─────────────────────────────────────────┐
# │  pathlib.Path vs os.path                  │
# │  os.path.join("a", "b")                   │
# │  Path("a") / "b"                          │
# │  os.path.exists(path)                     │
# │  Path(path).exists()                      │
# │  → pathlib이 더 직관적이고 읽기 쉬움        │
# └─────────────────────────────────────────┘

# Path 객체 생성
p = Path(tempfile.gettempdir())
print(f"경로: {p}")
print(f"이름: {p.name}")
print(f"존재: {p.exists()}")
print(f"디렉토리: {p.is_dir()}")

# / 연산자로 경로 결합
data_path = p / "data" / "raw"
print(f"\n결합 경로: {data_path}")

# 디렉토리 생성
data_path.mkdir(parents=True, exist_ok=True)
print(f"생성 완료: {data_path.exists()}")

# 경로 구성 요소
sample = Path("/home/user/documents/report.txt")
print(f"\n경로 분석: {sample}")
print(f"  부모: {sample.parent}")
print(f"  파일명: {sample.name}")
print(f"  스템: {sample.stem}")
print(f"  확장자: {sample.suffix}")
print(f"  구성요소: {sample.parts}")
```

```python
# 파일 읽기/쓰기도 Path 객체로 가능
from pathlib import Path
import tempfile

tmp = Path(tempfile.gettempdir())
test_file = tmp / "pathlib_test.txt"

# 쓰기
test_file.write_text("PathLib으로 파일 쓰기\n두 번째 줄", encoding="utf-8")
print(f"파일 생성: {test_file.exists()}")

# 읽기
content = test_file.read_text(encoding="utf-8")
print(f"파일 내용:\n{content}")

# ┌─────────────────────────────────────────┐
# │  디렉토리 순회                              │
# │  iterdir()  → 디렉토리 내 파일/폴더 목록    │
# │  glob()     → 패턴 매칭 파일 검색           │
# │  rglob()    → 하위 디렉토리까지 재귀 검색    │
# └─────────────────────────────────────────┘

# 디렉토리 내 파일 목록
print("\n임시 디렉토리 내 .txt 파일:")
for f in sorted(tmp.glob("*.txt"))[:5]:
    print(f"  {f.name}")

# 파일 정보
if test_file.exists():
    stat = test_file.stat()
    print(f"\n파일 크기: {stat.st_size} 바이트")
```
