## 정답 확인

아래 셀을 실행하여 모든 문제의 정답을 확인하세요.

````python
# 정답
import json, csv, os, tempfile, sys
from pathlib import Path
from string import Template

# ---- 문제 1 정답 ----
def get_positive_int(prompt):
    """양의 정수를 입력받아 반환한다. 잘못된 입력은 재요청."""
    while True:
        try:
            value = int(input(prompt))
            if value <= 0:
                print('양의 정수를 입력하세요.')
                continue
            return value
        except ValueError:
            print('정수를 입력하세요.')

# ---- 문제 2 정답 ----
def format_currency(amount):
    """정수를 ₩1,234,567 형식으로 반환."""
    return f'₩{amount:,}'

def format_percent(value):
    """0~1 사이 값을 85.50% 형식으로 반환."""
    return f'{value*100:.2f}%'

def format_table_row(name, score, width=10):
    """'|Kim       |       95.3|' 형식의 문자열 반환."""
    return f'|{name:<{width}}|{score:>{width}.1f}|'

# ---- 문제 3 정답 ----
def write_lines(filepath, lines):
    """문자열 리스트를 파일에 한 줄씩 쓴다."""
    with open(filepath, 'w', encoding='utf-8') as f:
        for line in lines:
            f.write(line + '\n')

def read_lines(filepath):
    """파일을 읽어 줄 단위 리스트로 반환 (끝 \n 제거)."""
    with open(filepath, 'r', encoding='utf-8') as f:
        return [line.rstrip('\n') for line in f]

# ---- 문제 4 정답 ----
def read_csv(filepath):
    """CSV 파일을 읽어 (header, rows) 튜플을 반환한다."""
    with open(filepath, 'r', encoding='utf-8', newline='') as f:
        reader = csv.DictReader(f)
        header = reader.fieldnames
        rows = [row for row in reader]
    return header, rows

# ---- 문제 5 정답 ----
def extract_json_values(json_str, *keys):
    """JSON 문자열에서 지정한 키의 값들을 튜플로 반환. 없으면 None."""
    data = json.loads(json_str)
    return tuple(data.get(key) for key in keys)

# ---- 문제 6 정답 ----
def get_path_info(filepath):
    """pathlib을 사용해 파일 경로 정보를 딕셔너리로 반환."""
    p = Path(filepath)
    return {
        'parent': p.parent.as_posix(),
        'stem': p.stem,
        'suffix': p.suffix,
        'name': p.name
    }

# ---- 문제 7 정답 ----
def append_line(filepath, line):
    """with문으로 파일 끝에 한 줄을 추가한다. 파일이 없으면 생성."""
    with open(filepath, 'a', encoding='utf-8') as f:
        f.write(line + '\n')

# ---- 문제 8 정답 ----
def format_all(name, age, height):
    """4가지 포매팅 방법으로 동일한 문자열을 생성해 리스트로 반환."""
    # 1. % 포매팅
    s1 = '%s은(는) %d세, 키 %.1fcm' % (name, age, height)
    # 2. str.format()
    s2 = '{}은(는) {}세, 키 {:.1f}cm'.format(name, age, height)
    # 3. f-string
    s3 = f'{name}은(는) {age}세, 키 {height:.1f}cm'
    # 4. Template
    t = Template('${name}은(는) ${age}세, 키 ${height}cm')
    s4 = t.substitute(name=name, age=age, height=f'{height:.1f}')
    return [s1, s2, s3, s4]

# ---- 문제 9 정답 ----
def safe_read(filepath):
    """파일을 안전하게 읽고, 예외 발생 시 오류 메시지를 반환."""
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            return f.read()
    except FileNotFoundError:
        return f'[오류] 파일을 찾을 수 없습니다: {filepath}'
    except IsADirectoryError:
        return f'[오류] 디렉토리입니다: {filepath}'
    except PermissionError:
        return f'[오류] 파일 읽기 권한이 없습니다: {filepath}'
    except Exception as e:
        return f'[오류] 알 수 없는 오류: {e}'

# ---- 문제 10 정답 ----
def parse_args(argv=None):
    """명령행 인수를 처리해 인사 문자열을 반환. argv=None이면 sys.argv 사용."""
    if argv is None:
        argv = sys.argv
    args = argv[1:]  # 스크립트명 제외
    if len(args) == 0:
        return '사용법: python script.py <이름> [나이]'
    name = args[0]
    if len(args) == 1:
        return f'안녕하세요, {name}님!'
    age_str = args[1]
    try:
        age = int(age_str)
    except ValueError:
        return '[오류] 나이는 숫자여야 합니다.'
    return f'안녕하세요, {name}님! ({age}세)'

# ---- 전체 테스트 실행 ----
print('=' * 50)
print('문제 2 테스트')
print(format_currency(1234567))
print(format_percent(0.855))
print(format_table_row('Kim', 95.3))

print('=' * 50)
print('문제 3 테스트')
tmp3 = os.path.join(tempfile.gettempdir(), 'test_lines.txt')
write_lines(tmp3, ['hello', 'world', 'python'])
assert read_lines(tmp3) == ['hello', 'world', 'python']
print('통과!')
os.remove(tmp3)

print('=' * 50)
print('문제 4 테스트')
tmp4 = os.path.join(tempfile.gettempdir(), 'test.csv')
with open(tmp4, 'w', encoding='utf-8', newline='') as f:
    f.write('name,math,english\nKim,90,85\nLee,75,92\n')
header, rows = read_csv(tmp4)
assert header == ['name', 'math', 'english']
assert rows[0]['name'] == 'Kim'
print('통과!')
os.remove(tmp4)

print('=' * 50)
print('문제 5 테스트')
s = '{"name": "Kim", "age": 25, "city": "Seoul"}'
assert extract_json_values(s, 'name', 'age', 'country') == ('Kim', 25, None)
assert extract_json_values(s, 'city') == ('Seoul',)
print('통과!')

print('=' * 50)
print('문제 6 테스트')
info = get_path_info('/home/user/docs/report.csv')
assert info['parent'] == '/home/user/docs'
assert info['stem'] == 'report'
assert info['suffix'] == '.csv'
assert info['name'] == 'report.csv'
info2 = get_path_info('data')
assert info2['suffix'] == ''
print('통과!')

print('=' * 50)
print('문제 7 테스트')
tmp7 = os.path.join(tempfile.gettempdir(), 'test_append.txt')
if os.path.exists(tmp7):
    os.remove(tmp7)
append_line(tmp7, 'first line')
append_line(tmp7, 'second line')
with open(tmp7, 'r', encoding='utf-8') as f:
    content = f.read()
assert content == 'first line\nsecond line\n'
print('통과!')
os.remove(tmp7)

print('=' * 50)
print('문제 8 테스트')
results = format_all('Kim', 25, 175.5)
expected = 'Kim은(는) 25세, 키 175.5cm'
for i, r in enumerate(results, 1):
    assert r == expected, f'방법 {i}: {repr(r)}'
print('통과!')

print('=' * 50)
print('문제 9 테스트')
err1 = safe_read('/nonexistent/path/file.txt')
assert '파일을 찾을 수 없습니다' in err1
# Windows에서 디렉토리 열기 시 IsADirectoryError 또는 PermissionError 발생 가능
err_dir = safe_read(tempfile.gettempdir())
assert '디렉토리' in err_dir or '권한' in err_dir
tmp9 = os.path.join(tempfile.gettempdir(), 'test_safe.txt')
with open(tmp9, 'w', encoding='utf-8') as f:
    f.write('Hello, World!')
assert safe_read(tmp9) == 'Hello, World!'
os.remove(tmp9)
print('통과!')

print('=' * 50)
print('문제 10 테스트')
assert '사용법' in parse_args(['script.py'])
assert parse_args(['script.py', 'Kim']) == '안녕하세요, Kim님!'
assert parse_args(['script.py', 'Kim', '25']) == '안녕하세요, Kim님! (25세)'
assert '오류' in parse_args(['script.py', 'Kim', 'abc'])
print('통과!')

print('=' * 50)
print('🎉 모든 테스트 통과!')
````
