# 📘 파일 입출력

파일에서 데이터를 **읽고 쓰는** 방법을 배웁니다.
`with`문으로 안전하게 파일을 다루는 것이 핵심입니다.

**학습 목표:**
- 텍스트 파일 읽기/쓰기 (read, write, append)
- 줄 단위 읽기 (readline, readlines, for line in file)
- 파일 모드와 with문 활용

## 1. 텍스트 파일 쓰기와 읽기

`open()` 함수로 파일을 열고, `with`문으로 안전하게 관리합니다.
파일 모드: `"r"`(읽기), `"w"`(쓰기), `"a"`(추가), `"x"`(독점 생성).

```python
import os, tempfile

# ┌─────────────────────────────────────────┐
# │  파일 모드                                 │
# │  "r"  → 읽기 (파일이 없으면 에러)           │
# │  "w"  → 쓰기 (파일이 있으면 덮어쓰기)       │
# │  "a"  → 추가 (파일 끝에 내용 추가)          │
# │  "x"  → 독점 생성 (파일이 있으면 에러)       │
# └─────────────────────────────────────────┘

# 테스트용 임시 디렉토리
tmp = tempfile.gettempdir()

# 텍스트 파일 쓰기 ("w" 모드)
file_path = os.path.join(tmp, "tutorial_test.txt")
with open(file_path, "w", encoding="utf-8") as f:
    f.write("첫 번째 줄\n")
    f.write("두 번째 줄\n")
    f.write("세 번째 줄\n")

print("파일 쓰기 완료")
```

```python
# 파일 읽기 — 전체 읽기
import os, tempfile

tmp = tempfile.gettempdir()
file_path = os.path.join(tmp, "tutorial_test.txt")

# read(): 전체 내용을 하나의 문자열로
with open(file_path, "r", encoding="utf-8") as f:
    content = f.read()
    print(f"=== 전체 읽기 ===")
    print(content)

# read(n): 처음 n글자만 읽기
with open(file_path, "r", encoding="utf-8") as f:
    first_5 = f.read(5)
    print(f"처음 5글자: {first_5!r}")
```

```python
# ┌─────────────────────────────────────────┐
# │  줄 단위 읽기                               │
# │  readline()   → 한 줄 읽기                  │
# │  readlines()  → 모든 줄을 리스트로          │
# │  for line in f → 한 줄씩 순회 (메모리 효율) │
# └─────────────────────────────────────────┘

import os, tempfile

tmp = tempfile.gettempdir()
file_path = os.path.join(tmp, "tutorial_test.txt")

# readline(): 한 줄씩 읽기
with open(file_path, "r", encoding="utf-8") as f:
    line1 = f.readline()
    line2 = f.readline()
    print(f"줄 1: {line1!r}")
    print(f"줄 2: {line2!r}")

# readlines(): 모든 줄을 리스트로
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
    print(f"\n줄 수: {len(lines)}")
    for i, line in enumerate(lines, 1):
        print(f"  {i}: {line.strip()}")

# for line in f: 메모리 효율적 순회
with open(file_path, "r", encoding="utf-8") as f:
    print("\n=== 줄별 순회 ===")
    for line in f:
        print(f"  {line.strip()}")
```

```python
# 파일 추가 모드와 독점 생성
import os, tempfile

tmp = tempfile.gettempdir()

# 추가 모드 ("a") — 파일 끝에 내용 추가
append_path = os.path.join(tmp, "tutorial_append.txt")
with open(append_path, "w", encoding="utf-8") as f:
    f.write("첫 번째 줄\n")

with open(append_path, "a", encoding="utf-8") as f:
    f.write("추가된 줄\n")

with open(append_path, "r", encoding="utf-8") as f:
    print(f"추가 모드 결과:\n{f.read()}")

# 독점 생성 모드 ("x") — 파일이 이미 있으면 에러
exclusive_path = os.path.join(tmp, "tutorial_exclusive.txt")
if os.path.exists(exclusive_path):
    os.remove(exclusive_path)

with open(exclusive_path, "x", encoding="utf-8") as f:
    f.write("새 파일\n")
print("독점 생성 완료")

# 두 번째 시도 → FileExistsError 발생
try:
    with open(exclusive_path, "x", encoding="utf-8") as f:
        f.write("실패\n")
except FileExistsError:
    print("이미 파일이 존재합니다 (FileExistsError)")
```

```python
# ┌─────────────────────────────────────────┐
# │  writelines() — 여러 줄 한 번에 쓰기       │
# │  with문 없이 파일 다루기 (권장하지 않음)      │
# └─────────────────────────────────────────┘

import os, tempfile

tmp = tempfile.gettempdir()
lines_path = os.path.join(tmp, "tutorial_lines.txt")

# writelines()로 여러 줄 쓰기
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]
with open(lines_path, "w", encoding="utf-8") as f:
    f.writelines(lines)

with open(lines_path, "r", encoding="utf-8") as f:
    print(f"writelines 결과:\n{f.read()}")

# with문 없이 파일 다루기 (수동 close 필요)
f = open(lines_path, "r", encoding="utf-8")
content = f.read()
f.close()  # 반드시 close!
print(f"수동 close 결과: {content!r}")
```
