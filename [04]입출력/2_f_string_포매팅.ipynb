# 📘 f-string 포매팅

Python 3.6+의 **f-string**은 문자열 포매팅의 가장 강력하고 읽기 쉬운 방법입니다.

**학습 목표:**
- f-string 기본 문법과 표현식
- 숫자 포매팅 (소수점, 자릿수, 천 단위 구분)
- 정렬과 날짜 포매팅
- 디버깅 포맷 (`{변수=}`)

## 1. f-string 기본

f-string은 `f"..."` 형태로 작성하며, 중괄호 `{}` 안에 변수나 표현식을 넣습니다.

```python
# f-string 기본
name = "Alice"
age = 30
height = 1.75

print(f"이름: {name}")
print(f"{name}은(는) {age}세입니다")
print(f"키: {height}m")
```

```python
# f-string 안에 표현식 넣기
print(f"2 + 3 = {2 + 3}")
print(f"{'hello'.upper()}")
print(f"이름 길이: {len(name)}")
print(f"{'홀수' if age % 2 else '짝수'}")

# 딕셔너리 접근
person = {"name": "Bob", "age": 25}
print(f"이름: {person['name']}, 나이: {person['age']}")
```

## 2. 숫자 포매팅

f-string으로 소수점, 자릿수, 천 단위 구분 등을 제어합니다.

```python
# ┌─────────────────────────────────────────┐
# │  숫자 포매팅 형식 지정자                    │
# │  {값:포맷}                                │
# │  .2f  → 소수점 2자리                        │
# │  ,    → 천 단위 구분                        │
# │  05d  → 5자리 0으로 채움                    │
# │  .1%  → 퍼센트 (소수점 1자리)              │
# └─────────────────────────────────────────┘

pi = 3.14159265
num = 42
big = 1000000

# 소수점 자릿수
print(f"Pi: {pi:.2f}")          # 3.14
print(f"Pi: {pi:.4f}")          # 3.1416
print(f"Pi: {pi:.0f}")          # 3

# 정렬과 폭 지정
print(f"{'왼쪽':<10}|")          # 왼쪽      |
print(f"{'오른쪽':>10}|")        #      오른쪽|
print(f"{'가운데':^10}|")        #   가운데  |

# 숫자 포매팅
print(f"2진수: {num:b}")         # 101010
print(f"16진수: {num:x}")        # 2a
print(f"0 채움: {num:05d}")      # 00042

# 천 단위 구분과 퍼센트
print(f"천 단위: {big:,}")       # 1,000,000
print(f"퍼센트: {0.856:.1%}")   # 85.6%
```

```python
# ┌───────────────────────────────────────────────┐
# │  날짜 포매팅                                    │
# │  f-string 안에서 날짜 형식을 직접 지정            │
# │  {now:%Y-%m-%d}  ← strftime 없이 포매팅          │
# └───────────────────────────────────────────────┘

from datetime import datetime

now = datetime(2025, 7, 31, 14, 30, 0)
print(f"날짜: {now:%Y-%m-%d}")        # 2025-07-31
print(f"시간: {now:%H:%M:%S}")        # 14:30:00
print(f"전체: {now:%Y-%m-%d %H:%M}")  # 2025-07-31 14:30

# ┌─────────────────────────────────────────┐
# │  디버깅 포맷 (Python 3.8+)                │
# │  {변수=} → 변수명=값 형태로 출력           │
# └─────────────────────────────────────────┘

x = 42
y = 3.14
name = "Alice"

print(f"{x=}")            # x=42
print(f"{y=}")            # y=3.14
print(f"{name=}")         # name='Alice'
print(f"{x + y=}")        # x + y=45.14
```

## 3. 구식 포매팅과 비교

f-string 이전에 사용하던 `%` 포맷팅과 `str.format()`을 알아봅니다.
**새 코드에서는 f-string을 사용하세요.**

```python
name = "Alice"
age = 30
pi = 3.14159

# ┌─────────────────────────────────────────┐
# │  세 가지 포매팅 방식 비교                   │
# │  1. % 포맷팅 (구식)                        │
# │  2. str.format() (중간 세대)               │
# │  3. f-string (최신, 권장!)                 │
# └─────────────────────────────────────────┘

# 1. % 포맷팅 (Python 초기)
print("이름: %s, 나이: %d" % (name, age))
print("Pi: %.2f" % pi)

# 2. str.format() (Python 2.6+)
print("이름: {}, 나이: {}".format(name, age))
print("Pi: {:.2f}".format(pi))

# 3. f-string (Python 3.6+, 권장!)
print(f"이름: {name}, 나이: {age}")
print(f"Pi: {pi:.2f}")

# f-string이 가장 읽기 쉽고 간결합니다
# 새 코드에서는 항상 f-string을 사용하세요
```
