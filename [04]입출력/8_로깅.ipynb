# 📘 로깅 (logging)

`print()` 대신 **logging 모듈**로 체계적인 로그를 관리합니다.
디버깅, 운영 모니터링, 에러 추적에 필수적입니다.

**학습 목표:**
- 로깅 레벨과 기본 설정
- 파일 로깅과 콘솔 로깅 동시에
- 로거 계층 구조

## 1. 로깅 기본

5가지 로그 레벨: `DEBUG` < `INFO` < `WARNING` < `ERROR` < `CRITICAL`.
설정된 레벨 이상의 메시지만 출력됩니다.

```python
import logging

# ┌─────────────────────────────────────────┐
# │  로그 레벨 (심각도 순)                      │
# │  DEBUG    = 10  → 상세 진단 정보            │
# │  INFO     = 20  → 정상 작동 확인            │
# │  WARNING  = 30  → 예상치 못한 상황          │
# │  ERROR    = 40  → 기능 실패                 │
# │  CRITICAL = 50  → 프로그램 동작 불가        │
# └─────────────────────────────────────────┘

# 기본 로깅 설정
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(name)s - %(levelname)s - %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S",
)

# 각 레벨별 로그 출력
logging.debug("디버그 — 상세 진단 정보")
logging.info("정보 — 정상 작동 확인")
logging.warning("경고 — 예상치 못한 상황")
logging.error("오류 — 기능 실패")
logging.critical("심각 — 프로그램 동작 불가")

# 레벨 숫자 확인
print(f"\nDEBUG={logging.DEBUG}, INFO={logging.INFO}, WARNING={logging.WARNING}")
print(f"ERROR={logging.ERROR}, CRITICAL={logging.CRITICAL}")
```

```python
# 로깅으로 예외 기록
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)s: %(message)s")

try:
    result = 10 / 0
except ZeroDivisionError:
    logging.error("0으로 나누기 오류 발생", exc_info=True)

# 변수 치환 (지연 포매팅)
user = "Alice"
action = "login"
logging.info(f"사용자 {user}이(가) {action} 수행")   # f-string 방식
logging.info("사용자 %s이(가) %s 수행", user, action)  # % 포매팅 (지연 평가)
```

## 2. 파일 로깅과 핸들러

로그를 파일과 콘솔에 동시에 출력할 수 있습니다.
핸들러로 출력 대상과 레벨을 개별적으로 제어합니다.

```python
import logging, os, tempfile

# ┌─────────────────────────────────────────────┐
# │  핸들러 종류                                  │
# │  StreamHandler  → 콘솔(터미널) 출력            │
# │  FileHandler     → 파일 출력                    │
# │  RotatingFileHandler → 파일 크기 제한 + 순환   │
# │  NullHandler     → 로그 무시                    │
# └─────────────────────────────────────────────┘

# 기존 핸들러 제거 (재실행 시 중복 방지)
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

# 파일 핸들러 (모든 레벨 기록)
log_path = os.path.join(tempfile.gettempdir(), "tutorial_app.log")
file_handler = logging.FileHandler(log_path, encoding="utf-8")
file_handler.setLevel(logging.DEBUG)
file_handler.setFormatter(logging.Formatter("%(asctime)s - %(levelname)s - %(message)s"))

# 콘솔 핸들러 (WARNING 이상만 출력)
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.WARNING)
console_handler.setFormatter(logging.Formatter("%(levelname)s: %(message)s"))

# 로거 설정
logger = logging.getLogger("myapp")
logger.setLevel(logging.DEBUG)
logger.addHandler(file_handler)
logger.addHandler(console_handler)

# 로그 기록
logger.debug("파일에만 기록됨")
logger.info("파일에만 기록됨")
logger.warning("파일과 콘솔에 모두 기록됨")
logger.error("파일과 콘솔에 모두 기록됨")

# 파일 내용 확인
with open(log_path, "r", encoding="utf-8") as f:
    print("\n=== 로그 파일 내용 ===")
    print(f.read())
```
