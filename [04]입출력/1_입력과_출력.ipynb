# 📘 입력과 출력 (input/output)

프로그램의 **입력과 출력**을 다루는 방법을 배웁니다.
`input()`으로 사용자 입력을 받고, `print()`로 결과를 출력합니다.

**학습 목표:**
- input()으로 사용자 입력 받기 및 타입 변환
- 여러 값 입력 파싱
- 입력 검증 패턴
- print()의 sep, end, file 옵션

## 1. input() — 사용자 입력

`input()`은 사용자로부터 한 줄을 입력받아 **문자열**로 반환합니다.
숫자가 필요하면 `int()`, `float()`로 변환해야 합니다.

```python
# input()은 항상 문자열을 반환합니다
# 주피터 노트북에서는 input()이 대화형으로 작동하므로
# 여기서는 변수에 직접 값을 할당하는 방식으로 설명합니다

name = "김파이"
print(f"이름: {name}")

# input()의 실제 사용법 (주석으로 표시):
# name = input("이름을 입력하세요: ")
# print(f"안녕하세요, {name}!")

# input()은 항상 문자열을 반환하므로 변환이 필요
age_str = "25"
age = int(age_str)
print(f"나이: {age} (타입: {type(age).__name__})")

height_str = "175.5"
height = float(height_str)
print(f"키: {height} (타입: {type(height).__name__})")
```

```python
# ┌─────────────────────────────────────────┐
# │  여러 값 입력받기                           │
# │  공백/콤마로 구분된 입력을 파싱하는 방법       │
# └─────────────────────────────────────────┘

# 공백으로 구분된 여러 숫자
line = "10 20 30"
numbers = list(map(int, line.split()))
print(f"숫자 리스트: {numbers}")
print(f"합계: {sum(numbers)}")

# 콤마로 구분된 값
line2 = "김파이,25,서울"
name, age, city = line2.split(",")
print(f"이름: {name}, 나이: {age}, 도시: {city}")

# 정수 리스트 입력
line3 = "1 2 3 4 5"
nums = list(map(int, line3.split()))
print(f"리스트: {nums}, 평균: {sum(nums)/len(nums)}")
```

```python
# ┌─────────────────────────────────────────┐
# │  입력 검증 (try-except 활용)               │
# │  잘못된 입력을 안전하게 처리하는 패턴        │
# └─────────────────────────────────────────┘

test_inputs = ["abc", "-5", "25"]

for value_str in test_inputs:
    try:
        value = int(value_str)
        if value <= 0:
            print(f"  '{value_str}' → 양수만 가능합니다")
        else:
            print(f"  '{value_str}' → 유효한 값: {value}")
    except ValueError:
        print(f"  '{value_str}' → 정수가 아닙니다")

# getpass 모듈 — 비밀번호 입력 (화면에 안 보임)
# from getpass import getpass
# password = getpass("비밀번호: ")
# print(f"비밀번호 길이: {len(password)}")
```

## 2. print()와 출력 제어

`print()`는 `sep`, `end`, `file` 매개변수로 출력 형식을 제어합니다.

```python
# ┌─────────────────────────────────────────┐
# │  print() 매개변수                          │
# │  sep: 여러 값 사이 구분자 (기본: 공백)        │
# │  end: 출력 끝에 추가할 문자 (기본: 줄바꿈)     │
# │  file: 출력 대상 (기본: sys.stdout)          │
# └─────────────────────────────────────────┘

# 기본 출력
print("Hello, World!")

# 여러 값 출력 (공백으로 구분)
print("이름:", "김파이", "나이:", 25)

# sep: 구분자 변경
print("2025", "07", "31", sep="-")       # 2025-07-31
print("a", "b", "c", sep=", ")            # a, b, c
print("a", "b", "c", sep="")              # abc

# end: 끝 문자 변경
print("Hello", end=" ")
print("World")                              # Hello World (같은 줄)

print("Loading", end="...")
print("Done")                               # Loading...Done

# StringIO로 출력 캡처
from io import StringIO
buffer = StringIO()
print("캡처된 내용", file=buffer)
print(f"버퍼: {buffer.getvalue()!r}")
```
