# 📘 CSV와 JSON 파일 다루기

**CSV**(표 형식 데이터)와 **JSON**(계층 구조 데이터)은 가장 널리 쓰이는 데이터 형식입니다.

**학습 목표:**
- CSV 파일 읽기/쓰기 (csv 모듈)
- JSON 파일 읽기/쓰기 (json 모듈)
- 파이썬 객체와 JSON 간 변환

## 1. CSV 파일

CSV(Comma-Separated Values)는 표 형식의 데이터를 저장하는 간단한 형식입니다.
`csv` 모듈로 읽고 쓸 수 있습니다.

```python
import csv, os, tempfile

tmp = tempfile.gettempdir()
csv_path = os.path.join(tmp, "tutorial_data.csv")

# ┌─────────────────────────────────────────┐
# │  CSV 파일 쓰기                              │
# │  csv.writer: 리스트를 CSV 행으로 작성         │
# │  csv.DictWriter: 딕셔너리를 CSV로 작성       │
# └─────────────────────────────────────────┘

# 기본 writer로 쓰기
with open(csv_path, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["이름", "나이", "도시"])
    writer.writerow(["김파이", 25, "서울"])
    writer.writerow(["이코딩", 30, "부산"])
    writer.writerow(["박개발", 28, "대전"])

print("CSV 파일 쓰기 완료")
```

```python
# CSV 파일 읽기
import csv, os, tempfile

tmp = tempfile.gettempdir()
csv_path = os.path.join(tmp, "tutorial_data.csv")

# 기본 reader로 읽기
with open(csv_path, "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)

print()

# DictReader로 읽기 (헤더를 키로 사용)
with open(csv_path, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['이름']}({row['나이']}세), {row['도시']}")
```

```python
# DictWriter로 쓰기
import csv, os, tempfile

tmp = tempfile.gettempdir()
csv_path2 = os.path.join(tmp, "tutorial_scores.csv")

students = [
    {"이름": "김파이", "수학": 90, "영어": 85},
    {"이름": "이코딩", "수학": 78, "영어": 92},
    {"이름": "박개발", "수학": 95, "영어": 88},
]

with open(csv_path2, "w", newline="", encoding="utf-8") as f:
    fieldnames = ["이름", "수학", "영어"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(students)

print("DictWriter 쓰기 완료")

# 확인
with open(csv_path2, "r", encoding="utf-8") as f:
    print(f.read())
```

## 2. JSON 파일

JSON(JavaScript Object Notation)은 계층 구조 데이터를 저장하는 표준 형식입니다.
파이썬의 `dict`, `list`와 자연스럽게 변환됩니다.

```python
import json, os, tempfile

tmp = tempfile.gettempdir()
json_path = os.path.join(tmp, "tutorial_data.json")

# ┌─────────────────────────────────────────┐
# │  JSON ↔ 파이썬 타입 매핑                  │
# │  JSON object  ↔ dict                       │
# │  JSON array   ↔ list                      │
# │  JSON string  ↔ str                       │
# │  JSON number  ↔ int/float                 │
# │  JSON true/false ↔ True/False              │
# │  JSON null     ↔ None                      │
# └─────────────────────────────────────────┘

# 파이썬 데이터
data = {
    "name": "김파이",
    "age": 25,
    "city": "서울",
    "skills": ["Python", "JavaScript", "SQL"],
    "employed": True,
    "salary": None
}

# JSON 파일 쓰기
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print("JSON 파일 쓰기 완료")
```

```python
# JSON 파일 읽기
import json, os, tempfile

tmp = tempfile.gettempdir()
json_path = os.path.join(tmp, "tutorial_data.json")

with open(json_path, "r", encoding="utf-8") as f:
    loaded = json.load(f)

print(f"이름: {loaded['name']}")
print(f"나이: {loaded['age']}")
print(f"기술: {loaded['skills']}")
print(f"재직: {loaded['employed']}")
print(f"급여: {loaded['salary']}")

# ┌─────────────────────────────────────────┐
# │  json.dumps() vs json.dump()               │
# │  json.dumps() → JSON 문자열 반환            │
# │  json.dump()  → 파일에 직접 쓰기            │
# │  json.loads() → JSON 문자열을 파이썬 객체로 │
# │  json.load()  → 파일에서 파이썬 객체로       │
# └─────────────────────────────────────────┘

# JSON 문자열 ↔ 파이썬 객체
json_str = '{"name": "Alice", "age": 30}'
parsed = json.loads(json_str)
print(f"\n파싱 결과: {parsed}")
print(f"이름: {parsed['name']}")

# 파이썬 객체 → JSON 문자열
json_str2 = json.dumps(parsed, ensure_ascii=False, indent=2)
print(f"JSON 문자열:\n{json_str2}")
```
