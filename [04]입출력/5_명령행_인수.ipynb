# 📘 명령행 인수 (sys.argv & argparse)

스크립트를 실행할 때 **명령행 인수**를 전달받아 처리하는 방법을 배웁니다.

**학습 목표:**
- sys.argv로 기본 인수 처리
- argparse로 구조화된 인수 파싱
- 위치 인수, 옵션 인수, 선택 제한

## 1. sys.argv — 기본 명령행 인수

`sys.argv`는 스크립트 이름과 명령행 인수를 리스트로 제공합니다.
간단한 스크립트에서 유용하지만, 복잡한 인수 처리에는 argparse가 더 적합합니다.

```python
# ┌───────────────────────────────────────────────┐
# │  sys.argv 구조                                  │
# │  python script.py arg1 arg2 --verbose            │
# │  sys.argv[0]    = "script.py"                   │
# │  sys.argv[1]    = "arg1"                        │
# │  sys.argv[2]    = "arg2"                        │
# │  sys.argv[3]    = "--verbose"                   │
# │  len(sys.argv)  = 4                             │
# └───────────────────────────────────────────────┘

import sys

# 실제 명령행 인수 대신 시뮬레이션
original_argv = sys.argv
sys.argv = ["my_script.py", "input.txt", "--verbose", "100"]

print(f"스크립트 이름: {sys.argv[0]}")
print(f"전체 인수: {sys.argv}")
print(f"인수만: {sys.argv[1:]}")
print(f"인수 개수: {len(sys.argv) - 1}")

sys.argv = original_argv  # 복원
```

## 2. argparse — 구조화된 인수 파싱

`argparse`는 도움말 생성, 타입 변환, 기본값, 선택 제한 등을 자동으로 처리합니다.

```python
import argparse

# ┌─────────────────────────────────────────┐
# │  argparse 기본 사용법                      │
# │  1. ArgumentParser() 객체 생성              │
# │  2. add_argument()로 인수 정의               │
# │  3. parse_args()로 인수 파싱                  │
# └─────────────────────────────────────────┘

# 기본 argparse 예시
parser = argparse.ArgumentParser(description="샘플 프로그램")
parser.add_argument("filename", help="입력 파일 경로")
parser.add_argument("-v", "--verbose", action="store_true", help="상세 출력")
parser.add_argument("-n", "--count", type=int, default=1, help="반복 횟수")

# 명령행 인수 시뮬레이션
args = parser.parse_args(["data.txt", "-v", "-n", "5"])
print(f"파일명: {args.filename}")
print(f"상세 모드: {args.verbose}")
print(f"반복 횟수: {args.count}")
```

```python
# ┌─────────────────────────────────────────┐
# │  argparse 인수 종류                        │
# │  위치 인수: 필수, 순서대로                    │
# │  옵션 인수: 선택적, - 또는 --로 시작           │
# │  기본값: default 매개변수                    │
# │  타입: type 매개변수 (int, float 등)          │
# │  선택 제한: choices 매개변수                  │
# └─────────────────────────────────────────┘

import argparse

# 선택 제한 (choices)
parser2 = argparse.ArgumentParser()
parser2.add_argument("--mode", choices=["train", "test", "validate"], default="train")
parser2.add_argument("--log-level", choices=["DEBUG", "INFO", "WARNING", "ERROR"], default="INFO")

args2 = parser2.parse_args(["--mode", "test", "--log-level", "DEBUG"])
print(f"모드: {args2.mode}")
print(f"로그 레벨: {args2.log_level}")

# nargs: 여러 개의 인수
parser3 = argparse.ArgumentParser()
parser3.add_argument("files", nargs="+", help="하나 이상의 입력 파일")
parser3.add_argument("--tags", nargs="*", help="0개 이상의 태그")

args3 = parser3.parse_args(["a.txt", "b.txt", "--tags", "python", "ml"])
print(f"파일: {args3.files}")
print(f"태그: {args3.tags}")
```
