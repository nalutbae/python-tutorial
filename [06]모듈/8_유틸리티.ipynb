# 📘 유틸리티 모듈

파이썬 표준 라이브러리의 유용한 **유틸리티 모듈**을 다룹니다.

**학습 목표:**
- random: 난수 생성과 샘플링
- math: 수학 함수와 상수
- json: JSON 직렬화/역직렬화
- hashlib: 해시 함수

## 1. random — 난수 생성

`random` 모듈은 난수 생성, 샘플링, 섞기 등을 제공합니다.

```python
import random

# ┌─────────────────────────────────────────┐
# │  random 주요 함수                         │
# │  random()        → 0~1 사이 난수             │
# │  randint(a, b)   → a~b 사이 정수             │
# │  choice(seq)     → 시퀀스에서 무작위 선택    │
# │  choices(seq, k) → 복원 추출 (k개)          │
# │  sample(seq, k)  → 비복원 추출 (k개)         │
# │  shuffle(seq)    → 시퀀스 섞기              │
# └─────────────────────────────────────────┘

# 시드 설정 (재현 가능한 난수)
random.seed(42)

print(f"random(): {random.random()}")
print(f"randint(1, 10): {random.randint(1, 10)}")
print(f"randint(1, 100): {random.randint(1, 100)}")

# 시퀀스에서 선택
fruits = ["사과", "바나나", "체리", "포도", "오렌지"]
print(f"\nchoice: {random.choice(fruits)}")
print(f"choices(3): {random.choices(fruits, k=3)}")  # 복원 추출
print(f"sample(3): {random.sample(fruits, k=3)}")     # 비복원 추출

# 섞기
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(f"\nshuffle: {numbers}")

# 균등/정규 분포
print(f"\nuniform(0, 10): {random.uniform(0, 10):.2f}")
print(f"gauss(0, 1): {random.gauss(0, 1):.2f}")
```

## 2. math — 수학 함수

`math` 모듈은 수학 상수와 함수를 제공합니다.

```python
import math

# ┌─────────────────────────────────────────┐
# │  math 주요 상수와 함수                     │
# │  math.pi       → 원주율 (3.14159...)        │
# │  math.e        → 자연상수 (2.71828...)       │
# │  math.sqrt()   → 제곱근                     │
# │  math.ceil()   → 올림                       │
# │  math.floor()  → 내림                       │
# │  math.log()    → 자연로그                   │
# │  math.gcd()    → 최대공약수                  │
# └─────────────────────────────────────────┘

print(f"pi: {math.pi}")
print(f"e: {math.e}")

print(f"\nsqrt(16): {math.sqrt(16)}")
print(f"ceil(4.3): {math.ceil(4.3)}")
print(f"floor(4.7): {math.floor(4.7)}")

print(f"\nlog(e): {math.log(math.e)}")  # 1.0
print(f"log10(100): {math.log10(100)}")  # 2.0
print(f"gcd(12, 8): {math.gcd(12, 8)}")  # 4

# 삼각함수 (라디안)
print(f"\nsin(pi/2): {math.sin(math.pi/2)}")  # 1.0
print(f"cos(0): {math.cos(0)}")  # 1.0

# 각도 변환
print(f"\ndegrees(pi): {math.degrees(math.pi)}")  # 180.0
print(f"radians(180): {math.radians(180)}")  # ~3.14159
```

## 3. json과 hashlib

`json`은 데이터 직렬화, `hashlib`은 해시 함수를 제공합니다.

```python
import json
import hashlib

# ┌─────────────────────────────────────────┐
# │  json — 데이터 직렬화/역직렬화             │
# │  json.dumps()  → 파이썬 객체 → JSON 문자열 │
# │  json.loads()  → JSON 문자열 → 파이썬 객체 │
# │  json.dump()   → 파이썬 객체 → 파일        │
# │  json.load()   → 파일 → 파이썬 객체        │
# └─────────────────────────────────────────┘

data = {
    "name": "김파이",
    "age": 25,
    "skills": ["Python", "SQL"],
    "employed": True
}

# JSON 문자열로 변환
json_str = json.dumps(data, ensure_ascii=False, indent=2)
print("JSON 문자열:")
print(json_str)

# JSON 문자열에서 파이썬 객체로
parsed = json.loads(json_str)
print(f"\n이름: {parsed['name']}")
print(f"기술: {parsed['skills']}")

# ┌─────────────────────────────────────────┐
# │  hashlib — 해시 함수                      │
# │  SHA-256, MD5 등으로 데이터 무결성 확인      │
# └─────────────────────────────────────────┘

message = "Hello, Python!"

# SHA-256 해시
sha256 = hashlib.sha256(message.encode()).hexdigest()
print(f"\nSHA-256: {sha256}")

# MD5 해시
md5 = hashlib.md5(message.encode()).hexdigest()
print(f"MD5: {md5}")

# 비밀번호 해시 (실제로는 bcrypt 사용 권장)
password = "my_password"
salt = "random_salt"
hashed = hashlib.sha256((password + salt).encode()).hexdigest()
print(f"\n비밀번호 해시: {hashed[:16]}...")
```
