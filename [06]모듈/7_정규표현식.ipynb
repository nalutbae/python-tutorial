# 📘 정규표현식 (Regular Expression)

**정규표현식**은 문자열 패턴을 검색, 치환, 검증하는 강력한 도구입니다.
`re` 모듈로 파이썬에서 정규표현식을 사용합니다.

**학습 목표:**
- 기본 패턴 매칭: search, match, fullmatch
- 문자 클래스와 수량자
- 그룹과 캡처
- findall, sub, compile

## 1. 기본 패턴 매칭

`re.search()`는 문자열 전체에서 첫 번째 매치를 찾고,
`re.match()`는 문자열 시작에서만 매치를 찾습니다.

```python
import re

# ┌─────────────────────────────────────────┐
# │  re 모듈 주요 함수                        │
# │  re.search()   → 전체에서 첫 매치           │
# │  re.match()    → 시작 위치에서만 매치        │
# │  re.fullmatch() → 전체 문자열 매치          │
# │  re.findall()  → 모든 매치 목록             │
# │  re.sub()      → 치환                       │
# └─────────────────────────────────────────┘

text = "Hello, Python! Python is great."

# search: 전체에서 첫 매치
result = re.search(r"Python", text)
if result:
    print(f"search: '{result.group()}' at {result.start()}-{result.end()}")

# match: 시작 위치에서만
result = re.match(r"Hello", text)
print(f"match: {result.group() if result else 'None'}")

result = re.match(r"Python", text)
print(f"match(Python): {result}")  # None (시작이 아님)

# findall: 모든 매치
matches = re.findall(r"Python", text)
print(f"\nfindall: {matches}")

# sub: 치환
replaced = re.sub(r"Python", "Java", text)
print(f"sub: {replaced}")
```

## 2. 문자 클래스와 수량자

문자 클래스는 문자 집합을, 수량자는 반복 횟수를 지정합니다.

```python
import re

# ┌─────────────────────────────────────────┐
# │  문자 클래스                                │
# │  [abc]  → a, b, c 중 하나                   │
# │  [a-z]  → a부터 z                           │
# │  [0-9]  → 숫자                              │
# │  \d     → 숫자 ([0-9]와 동일)               │
# │  \w     → 문자+숫자+밑줄                    │
# │  \s     → 공백                               │
# │  .      → 임의의 문자                        │
# └─────────────────────────────────────────┘

# ┌─────────────────────────────────────────┐
# │  수량자                                    │
# │  *      → 0번 이상                          │
# │  +      → 1번 이상                          │
# │  ?      → 0번 또는 1번                      │
# │  {n}    → 정확히 n번                        │
# │  {n,m}  → n번 이상 m번 이하                 │
# └─────────────────────────────────────────┘

# 이메일 패턴
email_pattern = r"[\w.-]+@[\w.-]+\.\w+"
text = "연락처: user@example.com, admin@test.org"
emails = re.findall(email_pattern, text)
print(f"이메일: {emails}")

# 전화번호 패턴
phone_pattern = r"\d{3}-\d{3,4}-\d{4}"
text2 = "전화: 010-1234-5678, 02-123-4567"
phones = re.findall(phone_pattern, text2)
print(f"전화번호: {phones}")

# 숫자만 추출
text3 = "가격: 15,000원, 수량: 3개"
numbers = re.findall(r"\d+", text3)
print(f"숫자: {numbers}")
```

## 3. 그룹과 캡처

괄호 `()`로 패턴의 일부를 그룹으로 캡처할 수 있습니다.

```python
import re

# ┌─────────────────────────────────────────┐
# │  그룹과 캡처                               │
# │  (pattern)   → 캡처 그룹                   │
# │  (?P<name>) → 이름 붙은 그룹              │
# │  (?:pattern) → 비캡처 그룹                │
# └─────────────────────────────────────────┘

# 날짜 캡처
date_pattern = r"(\d{4})-(\d{2})-(\d{2})"
text = "2025-07-31과 2025-12-25"

matches = re.findall(date_pattern, text)
print(f"날짜 매치: {matches}")

# 이름 붙은 그룹
named_pattern = r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})"
match = re.search(named_pattern, text)
if match:
    print(f"\n년: {match.group('year')}")
    print(f"월: {match.group('month')}")
    print(f"일: {match.group('day')}")

# compile — 패턴 컴파일 (재사용 시 성능 향상)
email_re = re.compile(r"([\w.-]+)@([\w.-]+\.\w+)")
text2 = "문의: user@example.com, admin@test.org"
for match in email_re.finditer(text2):
    print(f"\n이메일: {match.group()}")
    print(f"  아이디: {match.group(1)}")
    print(f"  도메인: {match.group(2)}")
```
