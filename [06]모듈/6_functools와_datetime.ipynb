# 📘 functools와 datetime

**functools**는 고차 함수와 함수 데코레이터를, **datetime**은
날짜와 시간 처리를 제공합니다.

**학습 목표:**
- functools: reduce, partial, lru_cache, wraps
- datetime: date, time, datetime, timedelta
- 시간대 처리와 실용 패턴

## 1. functools — reduce, partial, lru_cache

`reduce`는 누적 연산, `partial`은 인수 고정, `lru_cache`는 메모이제이션입니다.

```python
from functools import reduce, partial, lru_cache

# ┌─────────────────────────────────────────┐
# │  reduce — 누적 연산                       │
# │  리스트의 모든 요소를 순차적으로 연산        │
# └─────────────────────────────────────────┘

numbers = [1, 2, 3, 4, 5]

# 누적 합
total = reduce(lambda x, y: x + y, numbers)
print(f"누적 합: {total}")  # 15

# 누적 곱
product = reduce(lambda x, y: x * y, numbers)
print(f"누적 곱: {product}")  # 120

# ┌─────────────────────────────────────────┐
# │  partial — 인수 고정                      │
# │  함수의 일부 인수를 미리 설정                │
# └─────────────────────────────────────────┘

def power(base, exponent):
    return base ** exponent

square = partial(power, exponent=2)
cube = partial(power, exponent=3)

print(f"\nsquare(4): {square(4)}")  # 16
cube(3): {cube(3)}")  # 27

# ┌─────────────────────────────────────────┐
# │  lru_cache — 메모이제이션                  │
# │  함수 결과를 캐시하여 중복 계산 방지         │
# └─────────────────────────────────────────┘

@lru_cache(maxsize=128)
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(f"\nfibonacci(30): {fibonacci(30)}")  # 캐시 덕분에 빠름
print(f"캐시 정보: {fibonacci.cache_info()}")
```

## 2. datetime — 날짜와 시간

`datetime` 모듈은 날짜, 시간, 시간 연산을 제공합니다.

```python
from datetime import date, time, datetime, timedelta

# ┌─────────────────────────────────────────┐
# │  datetime 주요 클래스                      │
# │  date      → 날짜 (년, 월, 일)              │
# │  time      → 시간 (시, 분, 초)              │
# │  datetime  → 날짜 + 시간                    │
# │  timedelta → 날짜/시간 차이                 │
# └─────────────────────────────────────────┘

# date 객체
today = date.today()
print(f"오늘: {today}")
print(f"년: {today.year}, 월: {today.month}, 일: {today.day}")

# 특정 날짜 생성
birthday = date(2025, 12, 25)
print(f"\n크리스마스: {birthday}")

# datetime 객체
now = datetime.now()
print(f"\n현재: {now}")
print(f"포맷: {now.strftime('%Y-%m-%d %H:%M:%S')}")

# 문자열 → datetime
parsed = datetime.strptime("2025-07-31 14:30", "%Y-%m-%d %H:%M")
print(f"파싱: {parsed}")

# f-string으로 날짜 포매팅
print(f"\nf-string: {now:%Y년 %m월 %d일 %H시 %M분}")
```

```python
# ┌─────────────────────────────────────────┐
# │  timedelta — 날짜/시간 연산                │
# │  두 날짜의 차이를 계산하거나 날짜를 이동     │
# └─────────────────────────────────────────┘

from datetime import date, timedelta

today = date.today()
print(f"오늘: {today}")

# 날짜 이동
tomorrow = today + timedelta(days=1)
next_week = today + timedelta(weeks=1)
print(f"내일: {tomorrow}")
print(f"다음 주: {next_week}")

# 날짜 차이
birthday = date(2025, 12, 25)
diff = birthday - today
print(f"\n크리스마스까지: {diff.days}일")

# 날짜 범위 생성
start = date(2025, 1, 1)
end = date(2025, 1, 7)
print("\n1월 첫 주:")
current = start
while current <= end:
    print(f"  {current:%A, %B %d}")
    current += timedelta(days=1)
```
