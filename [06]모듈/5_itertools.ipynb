# 📘 itertools 모듈

**itertools**는 반복자를 효율적으로 다루는 함수를 제공합니다.
메모리를 절약하면서 대량의 데이터를 처리할 수 있습니다.

**학습 목표:**
- 무한 반복자: count, cycle, repeat
- 누적과 필터: accumulate, chain, compress
- 순열과 조합: permutations, combinations
- 그룹화와 슬라이싱: groupby, islice

## 1. 무한 반복자와 체인

`count`, `cycle`, `repeat`은 무한히 반복하는 반복자를 만듭니다.
`chain`은 여러 반복자를 연결합니다.

```python
from itertools import count, cycle, repeat, chain, islice

# ┌─────────────────────────────────────────┐
# │  무한 반복자 — 반드시 islice 등으로 제한!  │
# │  count(start, step) → 무한 카운터           │
# │  cycle(iterable)   → 무한 순환              │
# │  repeat(elem, n)   → n번 반복               │
# └─────────────────────────────────────────┘

# count: 무한 카운터 (islice로 제한)
for i in islice(count(1, 2), 5):  # 1부터 2씩, 5개
    print(f"  count: {i}")

# cycle: 무한 순환
print("\ncycle:")
colors = cycle(["빨강", "초록", "파랑"])
for i, color in enumerate(colors):
    if i >= 6:
        break
    print(f"  {i}: {color}")

# repeat: 반복
print("\nrepeat:")
for x in repeat("안녕", 3):
    print(f"  {x}")

# chain: 여러 반복자 연결
print("\nchain:")
combined = chain([1, 2], [3, 4], [5])
print(f"  결과: {list(combined)}")
```

## 2. 누적, 필터, 그룹화

`accumulate`는 누적 계산, `compress/filterfalse`는 필터링,
`groupby`는 연속된 요소를 그룹화합니다.

```python
from itertools import accumulate, compress, groupby
from itertools import filterfalse, takewhile, dropwhile

# accumulate: 누적 계산
import operator
numbers = [1, 2, 3, 4, 5]
print(f"누적 합: {list(accumulate(numbers))}")
print(f"누적 곱: {list(accumulate(numbers, operator.mul))}")

# compress: 선택자로 필터링
data = ['a', 'b', 'c', 'd']
selectors = [True, False, True, False]
print(f"\ncompress: {list(compress(data, selectors))}")

# filterfalse: 조건이 거짓인 요소
print(f"filterfalse(짝수 아닌): {list(filterfalse(lambda x: x % 2 == 0, range(10)))}")

# takewhile / dropwhile
print(f"takewhile(<5): {list(takewhile(lambda x: x < 5, range(10)))}")
print(f"dropwhile(<5): {list(dropwhile(lambda x: x < 5, range(10)))}")

# groupby: 연속된 요소 그룹화
data = [('A', 1), ('A', 2), ('B', 3), ('B', 4), ('C', 5)]
print("\ngroupby:")
for key, group in groupby(data, key=lambda x: x[0]):
    print(f"  {key}: {list(group)}")
```

## 3. 순열과 조합

`permutations`은 순서가 있는 선택, `combinations`은 순서 없는 선택입니다.

```python
from itertools import permutations, combinations, combinations_with_replacement

# ┌─────────────────────────────────────────┐
# │  순열과 조합                               │
# │  permutations(n, r) → 순서 있는 r개 선택    │
# │  combinations(n, r) → 순서 없는 r개 선택   │
# │  combinations_with_replacement → 중복 허용  │
# └─────────────────────────────────────────┘

# permutations: 순서가 있는 선택
items = ['A', 'B', 'C']
print("permutations(A,B,C, 2):")
for p in permutations(items, 2):
    print(f"  {p}")
print(f"총 {len(list(permutations(items, 2)))}개")

# combinations: 순서 없는 선택
print("\ncombinations(A,B,C, 2):")
for c in combinations(items, 2):
    print(f"  {c}")
print(f"총 {len(list(combinations(items, 2)))}개")

# combinations_with_replacement: 중복 허용
print("\ncombinations_with_replacement(A,B,C, 2):")
for c in combinations_with_replacement(items, 2):
    print(f"  {c}")
print(f"총 {len(list(combinations_with_replacement(items, 2)))}개")
```
