# 📘 collections 모듈

**collections** 모듈은 특수 컨테이너 데이터 타입을 제공합니다.
기본 dict, list, tuple의 한계를 극복하는 유용한 자료구조입니다.

**학습 목표:**
- Counter: 요소 카운팅
- defaultdict: 기본값 딕셔너리
- deque: 양방향 큐
- namedtuple: 이름 붙은 튜플

## 1. Counter — 요소 카운팅

`Counter`는 반복 가능한 객체에서 요소의 개수를 셉니다.

```python
from collections import Counter

# ┌─────────────────────────────────────────┐
# │  Counter 주요 기능                        │
# │  요소 카운팅, 최빈값, 연산                 │
# └─────────────────────────────────────────┘

# 문자열에서 글자 빈도
text = "hello world"
letter_count = Counter(text.replace(" ", ""))
print(f"글자 빈도: {letter_count}")
print(f"가장 많은 글자 3개: {letter_count.most_common(3)}")

# 리스트에서 요소 빈도
fruits = ["사과", "바나나", "사과", "오렌지", "바나나", "사과"]
fruit_count = Counter(fruits)
print(f"\n과일 빈도: {fruit_count}")
print(f"사과 개수: {fruit_count['사과']}")

# Counter 연산
c1 = Counter(a=3, b=1, c=2)
c2 = Counter(a=1, b=2, d=1)
print(f"\n합: {c1 + c2}")       # 요소 더하기
print(f"차: {c1 - c2}")       # 요소 빼기 (음수는 제거)
print(f"교집합: {c1 & c2}")   # 최솟값
```

## 2. defaultdict — 기본값 딕셔너리

`defaultdict`는 존재하지 않는 키에 접근할 때 기본값을 자동 생성합니다.

```python
from collections import defaultdict

# ┌─────────────────────────────────────────┐
# │  defaultdict 사용법                       │
# │  defaultdict(list)  → 기본값: 빈 리스트     │
# │  defaultdict(int)   → 기본값: 0             │
# │  defaultdict(set)   → 기본값: 빈 집합       │
# └─────────────────────────────────────────┘

# 리스트 defaultdict — 단어 그룹화
words = ["apple", "banana", "avocado", "blueberry", "cherry"]
groups = defaultdict(list)
for word in words:
    groups[word[0]].append(word)

print("단어 그룹화:")
for key, values in sorted(groups.items()):
    print(f"  {key}: {values}")

# int defaultdict — 카운터
counter = defaultdict(int)
for word in words:
    counter[word[0]] += 1
print(f"\n첫 글자 카운트: {dict(counter)}")

# set defaultdict
unique = defaultdict(set)
for word in words:
    unique[len(word)].add(word)
print(f"\n길이별 단어: {dict(unique)}")
```

## 3. deque — 양방향 큐

`deque`(double-ended queue)는 양쪽 끝에서 빠른 추가/삭제가 가능합니다.

```python
from collections import deque

# ┌─────────────────────────────────────────┐
# │  deque vs list                             │
# │  deque: 양쪽 끝 O(1), 중간 O(n)            │
# │  list:  끝 O(1), 맨 앞 O(n)               │
# └─────────────────────────────────────────┘

# deque 생성
d = deque([1, 2, 3])
print(f"초기 deque: {d}")

# 양쪽 추가
d.append(4)         # 오른쪽 추가
d.appendleft(0)      # 왼쪽 추가
print(f"추가 후: {d}")   # deque([0, 1, 2, 3, 4])

# 양쪽 제거
right = d.pop()       # 오른쪽 제거
left = d.popleft()    # 왼쪽 제거
print(f"제거 후: {d}, 오른쪽: {right}, 왼쪽: {left}")

# 회전
d.extend([5, 6, 7])
print(f"\nextend 후: {d}")
d.rotate(2)          # 오른쪽으로 2칸 회전
print(f"rotate(2) 후: {d}")
d.rotate(-2)         # 왼쪽으로 2칸 회전
print(f"rotate(-2) 후: {d}")

# 최대 길이 제한 (슬라이딩 윈도우)
recent = deque(maxlen=3)
for i in range(1, 6):
    recent.append(i)
    print(f"  추가: {i} → {list(recent)}")
```

## 4. namedtuple — 이름 붙은 튜플

`namedtuple`은 튜플에 이름을 붙여 읽기 쉽게 만듭니다.

```python
from collections import namedtuple

# namedtuple 정의
Point = namedtuple('Point', ['x', 'y'])
Student = namedtuple('Student', ['name', 'age', 'grade'])

# ┌─────────────────────────────────────────┐
# │  namedtuple vs 일반 튜플                   │
# │  일반 튜플: p[0], p[1] (인덱스 접근)        │
# │  namedtuple: p.x, p.y (이름 접근)          │
# │  인덱스 접근도 가능: p[0], p[1]              │
# └─────────────────────────────────────────┘

p = Point(3, 4)
print(f"이름 접근: p.x={p.x}, p.y={p.y}")
print(f"인덱스 접근: p[0]={p[0]}, p[1]={p[1]}")

s = Student("김파이", 25, 4.3)
print(f"\n학생: {s.name}, {s.age}세, 학점: {s.grade}")

# 딕셔너리로 변환
print(f"딕셔너리: {s._asdict()}")

# 필드 교체 (불변이므로 새 객체 반환)
s2 = s._replace(age=26)
print(f"교체 후: {s2}")
print(f"원본不变: {s}")
```
