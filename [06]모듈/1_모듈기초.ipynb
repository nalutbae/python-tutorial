# 📘 모듈 기초

**모듈**은 파이썬 코드를 논리적으로 그룹화한 파일입니다.
`import`로 모듈을 불러와 재사용할 수 있습니다.

**학습 목표:**
- 모듈이란 무엇인가
- import의 다양한 형태
- 모듈 속성 확인
- 커스텀 모듈 생성
- __all__과 조건부 import

## 1. 모듈이란?

모듈은 `.py` 파일 하나입니다. 함수, 클래스, 변수를 모듈에 정의하고
다른 파일에서 `import`로 불러와 사용합니다.

```python
# ┌─────────────────────────────────────────┐
# │  import의 다양한 형태                       │
# │  import module          → 전체 모듈 임포트   │
# │  from module import foo → 특정 함수만 임포트│
# │  import module as m     → 별칭 지정          │
# └─────────────────────────────────────────┘

import math
print(f"math.pi: {math.pi}")
print(f"math.sqrt(16): {math.sqrt(16)}")

from math import ceil, floor
print(f"ceil(4.3): {ceil(4.3)}")
print(f"floor(4.7): {floor(4.7)}")

import math as m
print(f"m.e: {m.e}")
```

```python
# ┌─────────────────────────────────────────┐
# │  모듈 속성 확인                             │
# │  __name__    → 모듈 이름                    │
# │  __file__    → 모듈 파일 경로               │
# │  __doc__     → 모듈 독스트링                │
# │  dir()       → 모듈의 모든 속성 목록         │
# └─────────────────────────────────────────┘

import math

print(f"모듈 이름: {math.__name__}")
print(f"모듈 파일: {math.__file__}")
print(f"독스트링: {math.__doc__[:80]}...")

# 모듈의 모든 속성
math_attrs = [attr for attr in dir(math) if not attr.startswith('_')]
print(f"\nmath 모듈 속성 (처음 10개): {math_attrs[:10]}")
print(f"전체 속성 수: {len(math_attrs)}")
```

## 2. 커스텀 모듈과 조건부 import

직접 만든 `.py` 파일도 모듈로 사용할 수 있습니다.
`if __name__ == "__main__"` 패턴으로 테스트 코드를 보호합니다.

```python
# ┌─────────────────────────────────────────┐
# │  if __name__ == "__main__" 패턴            │
# │  모듈로 임포트될 때는 실행되지 않고           │
# │  직접 실행할 때만 실행되는 코드               │
# └─────────────────────────────────────────┘

# mymath.py 파일이 있다고 가정:
# def add(a, b):
#     return a + b
#
# def multiply(a, b):
#     return a * b
#
# if __name__ == "__main__":
#     # 직접 실행할 때만 테스트
#     print(add(2, 3))      # 5
#     print(multiply(4, 5)) # 20

# 다른 파일에서 임포트
# from mymath import add, multiply
# result = add(10, 20)  # 30

# __all__: from module import * 시 노출할 목록
# __all__ = ["add", "multiply"]  # add, multiply만 노출

print("커스텀 모듈 패턴을 이해했습니다")

# 조건부 import — 모듈이 없을 때 대체
try:
    import numpy as np
    print(f"NumPy 사용 가능: {np.__version__}")
except ImportError:
    print("NumPy가 설치되어 있지 않습니다. 순수 Python으로 대체합니다.")
```
