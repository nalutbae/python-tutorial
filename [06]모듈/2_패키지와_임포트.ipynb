# 📘 패키지와 임포트

**패키지**는 모듈들을 디렉토리 구조로 그룹화한 것입니다.
`__init__.py` 파일로 패키지를 정의합니다.

**학습 목표:**
- 패키지 구조와 __init__.py
- 절대/상대 임포트
- sys.path와 모듈 검색 경로

## 1. 패키지 구조

패키지는 디렉토리 안에 `__init__.py`가 있는 구조입니다.
모듈을 계층적으로组织할 수 있습니다.

```python
# ┌─────────────────────────────────────────┐
# │  패키지 구조 예시                           │
# │  mypackage/                                │
# │  ├── __init__.py                           │
# │  ├── math_utils.py                         │
# │  ├── string_utils.py                       │
# │  └── subpackage/                           │
# │      ├── __init__.py                       │
# │      └── advanced.py                       │
# └─────────────────────────────────────────┘

# 패키지 임포트 방식
# 1) 전체 패키지 임포트
# import mypackage

# 2) 특정 모듈 임포트
# from mypackage import math_utils

# 3) 특정 함수/클래스 임포트
# from mypackage.math_utils import add, multiply

# 4) 하위 패키지 임포트
# from mypackage.subpackage import advanced

print("패키지 임포트 방식을 이해했습니다")

# ┌─────────────────────────────────────────┐
# │  __init__.py의 역할                        │
# │  • 패키지로 인식되게 함 (Python 3.3+ 선택)  │
# │  • 패키지 임포트 시 초기화 코드 실행          │
# │  • __all__로 from package import * 제어     │
# └─────────────────────────────────────────┘

# sys.path — 모듈 검색 경로
import sys
print(f"\n모듈 검색 경로 (처음 5개):")
for p in sys.path[:5]:
    print(f"  {p}")
```

## 2. 절대 임포트와 상대 임포트

절대 임포트는 전체 경로를, 상대 임포트는 현재 위치 기준으로 임포트합니다.

```python
# ┌─────────────────────────────────────────┐
# │  절대 임포트 vs 상대 임포트                 │
# │  절대: from mypackage.math_utils import add │
# │  상대: from .math_utils import add            │
# │        from ..subpackage import advanced      │
# └─────────────────────────────────────────┘

# 절대 임포트 (권장)
# from mypackage.math_utils import add
# from mypackage.subpackage.advanced import calculate

# 상대 임포트 (패키지 내부에서만 사용)
# from .math_utils import add        # 같은 패키지 내
# from . import math_utils             # 같은 패키지 전체
# from ..subpackage import advanced    # 상위 패키지에서

print("절대/상대 임포트를 이해했습니다")

# __all__로 임포트 제어
# __init__.py에 정의:
# __all__ = ["math_utils", "string_utils"]
# from mypackage import * 시 __all__에 있는 것만 임포트

print("__all__ 임포트 제어를 이해했습니다")
```
