## 문제 1: Counter — 빈도수 분석

문장에서 각 단어의 빈도수를 구하고, 가장 많이 등장하는 단어 상위 3개를 출력하세요.

```python
text = 'the cat sat on the mat the cat ate the rat the cat is fat'
```

````python
# 문제 1: 답을 작성하세요
from collections import Counter
text = 'the cat sat on the mat the cat ate the rat the cat is fat'
# Counter로 빈도수 분석 + 상위 3개
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from collections import Counter
text = 'the cat sat on the mat the cat ate the rat the cat is fat'
words = text.split()
counter = Counter(words)
print(counter.most_common(3))
# [('the', 4), ('cat', 3), ('sat', 1)]
```

## 문제 2: defaultdict — 중첩 딕셔너리

학생들의 과목별 점수를 `defaultdict(list)`에 그룹화하세요.

```python
records = [('Kim', 'math', 90), ('Lee', 'math', 85), ('Kim', 'eng', 88), ('Lee', 'eng', 92)]
```

````python
# 문제 2: 답을 작성하세요
from collections import defaultdict
records = [('Kim', 'math', 90), ('Lee', 'math', 85), ('Kim', 'eng', 88), ('Lee', 'eng', 92)]
# 이름별 점수 그룹화
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from collections import defaultdict
records = [('Kim', 'math', 90), ('Lee', 'math', 85), ('Kim', 'eng', 88), ('Lee', 'eng', 92)]
groups = defaultdict(list)
for name, subj, score in records:
    groups[name].append((subj, score))
for name, scores in groups.items():
    print(f'{name}: {scores}')
```

</details>

## 문제 3: deque — 슬라이딩 윈도우 최댓값

`deque`를 사용해 정수 리스트에서 크기 3의 슬라이딩 윈도우 최댓값을 구하세요.

```python
nums = [1, 3, -1, -3, 5, 3, 6, 7]
# 결과: [3, 3, 5, 5, 6, 7]
```

````python
# 문제 3: 답을 작성하세요
from collections import deque
nums = [1, 3, -1, -3, 5, 3, 6, 7]
# 슬라이딩 윈도우 최댓값
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from collections import deque
nums = [1, 3, -1, -3, 5, 3, 6, 7]
k = 3
result = []
dq = deque()  # 인덱스 저장
for i, n in enumerate(nums):
    while dq and nums[dq[-1]] < n:
        dq.pop()
    dq.append(i)
    if dq[0] <= i - k:
        dq.popleft()
    if i >= k - 1:
        result.append(nums[dq[0]])
print(result)  # [3, 3, 5, 5, 6, 7]
```

</details>

## 문제 4: itertools.groupby — 연속 그룹화

정렬된 리스트에서 `groupby`로 연속된 값의 그룹과 개수를 구하세요.

```python
data = [1, 1, 1, 2, 2, 3, 3, 3, 3, 1, 1]
# 결과: (1, 3), (2, 2), (3, 4), (1, 2)
```

````python
# 문제 4: 답을 작성하세요
from itertools import groupby
data = [1, 1, 1, 2, 2, 3, 3, 3, 3, 1, 1]
# groupby로 연속 그룹화
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from itertools import groupby
data = [1, 1, 1, 2, 2, 3, 3, 3, 3, 1, 1]

    length = len(list(group))
    print(f'값 {key}: {length}개')
```

## 문제 5: permutations/combinations — 확률 계산

1~5 숫자에서 3개를 선택하는 모든 조합을 구하고, 합이 10 이상인 조합의 수를 세세요.

````python
# 문제 5: 답을 작성하세요
from itertools import combinations
# 1~5에서 3개 선택, 합 10 이상인 조합 수
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from itertools import combinations
nums = range(1, 6)
total = 0
for combo in combinations(nums, 3):
    if sum(combo) >= 10:
        print(combo, f'합={sum(combo)}')
        total += 1
print(f'합이 10 이상인 조합 수: {total}')
```

</details>

## 문제 7: lru_cache — 피보나치 메모이제이션

`@lru_cache`를 사용해 피보나치 수열의 50번째 항을 계산하세요.
캐시 정보(`cache_info()`)도 출력하세요.

````python
# 문제 7: 답을 작성하세요
from functools import lru_cache
# lru_cache 피보나치 50번째
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from functools import lru_cache
@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)
print(f'fib(50) = {fib(50)}')
print(f'캐시 정보: {fib.cache_info()}')
```

</details>

## 문제 9: 정규표현식 — 그룹과 치환

날짜 문자열 `YYYY-MM-DD`를 `DD/MM/YYYY` 형식으로 변환하세요.

```python
dates = ['2025-01-15', '2024-12-25', '2025-03-01']
```

````python
# 문제 9: 답을 작성하세요
import re
dates = ['2025-01-15', '2024-12-25', '2025-03-01']
# YYYY-MM-DD → DD/MM/YYYY 변환
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import re
dates = ['2025-01-15', '2024-12-25', '2025-03-01']
pattern = r'(\d{4})-(\d{2})-(\d{2})'
for d in dates:
    print(re.sub(pattern, r'\3/\2/\1', d))
# 15/01/2025
# 25/12/2024
# 01/03/2025
```

</details>

## 문제 10: itertools.product — 행렬 인덱스 생성

`itertools.product`를 사용해 3×3 행렬의 모든 (행, 열) 인덱스를 생성하고,
대각선 인덱스만 필터링하세요.

````python
# 문제 10: 답을 작성하세요
from itertools import product
# 3×3 행렬 인덱스 + 대각선 필터링
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from itertools import product
indices = list(product(range(3), repeat=2))
print(f'전체 인덱스: {indices}')
diagonal = [(i, j) for i, j in indices if i == j]
print(f'대각선: {diagonal}')  # [(0,0), (1,1), (2,2)]
```

</details>
