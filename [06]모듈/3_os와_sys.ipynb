# 📘 os와 sys 모듈

**os** 모듈은 운영체제 기능을, **sys** 모듈은 파이썬 런타임 정보를 제공합니다.
**pathlib**은 현대적인 경로 처리 방식입니다.

**학습 목표:**
- os 모듈: 환경변수, 경로, 파일/디렉토리
- sys 모듈: argv, path, 버전 정보
- pathlib: 현대적 경로 처리

## 1. os 모듈 — 환경변수와 경로

`os` 모듈은 파일 시스템, 환경변수, 프로세스 관리 등을 제공합니다.

```python
import os

# ┌─────────────────────────────────────────┐
# │  os 모듈 주요 기능                        │
# │  os.environ   → 환경변수 딕셔너리          │
# │  os.getcwd()  → 현재 작업 디렉토리          │
# │  os.listdir() → 디렉토리 내용 목록         │
# │  os.path.join() → 경로 결합                 │
# │  os.path.exists() → 경로 존재 확인          │
# └─────────────────────────────────────────┘

# 현재 작업 디렉토리
print(f"현재 디렉토리: {os.getcwd()}")

# 디렉토리 내용
print(f"\n현재 디렉토리 파일 (처음 10개):")
for f in os.listdir('.')[:10]:
    print(f"  {f}")

# 경로 결합
path = os.path.join('folder', 'subfolder', 'file.txt')
print(f"\n경로 결합: {path}")

# 환경변수
home = os.getenv('HOME') or os.getenv('USERPROFILE')
print(f"홈 디렉토리: {home}")

# 경로 존재 확인
print(f"현재 디렉토리 존재: {os.path.exists('.')}")
print(f"가짜 경로: {os.path.exists('/nonexistent/path')}")
```

## 2. sys 모듈 — 파이썬 런타임

`sys` 모듈은 파이썬 인터프리터와 관련된 정보를 제공합니다.

```python
import sys

# ┌─────────────────────────────────────────┐
# │  sys 모듈 주요 기능                       │
# │  sys.argv     → 명령행 인수 목록            │
# │  sys.path     → 모듈 검색 경로              │
# │  sys.version  → 파이썬 버전                │
# │  sys.platform → 운영체제 정보               │
# │  sys.exit()   → 프로그램 종료               │
# └─────────────────────────────────────────┘

print(f"파이썬 버전: {sys.version}")
print(f"플랫폼: {sys.platform}")
print(f"명령행 인수: {sys.argv}")

# 모듈 검색 경로
print(f"\n모듈 검색 경로 (처음 3개):")
for p in sys.path[:3]:
    print(f"  {p}")

# 표준 인코딩
print(f"\n기본 인코딩: {sys.getdefaultencoding()}")
print(f"파일 시스템 인코딩: {sys.getfilesystemencoding()}")
```

## 3. pathlib — 현대적 경로 처리

`pathlib.Path`는 객체 지향적이고 직관적인 경로 처리를 제공합니다.

```python
from pathlib import Path
import tempfile

# ┌───────────────────────────────────────────────┐
# │  pathlib.Path vs os.path                       │
# │  os.path.join("a", "b")                       │
# │  Path("a") / "b"                              │
# │  os.path.exists(path)                         │
# │  Path(path).exists()                          │
# │  → pathlib이 더 직관적이고 읽기 쉬움           │
# └───────────────────────────────────────────────┘

# Path 객체 생성
tmp = Path(tempfile.gettempdir())
print(f"경로: {tmp}")
print(f"이름: {tmp.name}")
print(f"존재: {tmp.exists()}")

# / 연산자로 경로 결합
data_path = tmp / "tutorial_data"
data_path.mkdir(parents=True, exist_ok=True)

# 파일 읽기/쓰기
test_file = data_path / "test.txt"
test_file.write_text("pathlib 테스트", encoding="utf-8")
content = test_file.read_text(encoding='utf-8')
print(f"\n파일 내용: {content}")

# 경로 분석
sample = Path("/home/user/documents/report.txt")
print(f"\n경로 분석:")
print(f"  부모: {sample.parent}")
print(f"  파일명: {sample.name}")
print(f"  스템: {sample.stem}")
print(f"  확장자: {sample.suffix}")

# 정리
import shutil
shutil.rmtree(data_path, ignore_errors=True)
```
