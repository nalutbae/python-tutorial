## 2. 가장 간단한 창 만들기

Tkinter GUI의 가장 기본 구조는 3단계입니다:

1. **메인 창 생성** — `tk.Tk()`
2. **위젯 추가** — 버튼, 레이블 등
3. **이벤트 루프 실행** — `mainloop()`

`mainloop()`는 창이 닫힐 때까지 프로그램을 유지시키는 역할을 합니다.
이 함수가 호출되지 않으면 창이 순식간에 사라집니다.

````python
# 가장 간단한 Tkinter 창
# 이 셀을 실행하면 빈 창이 나타납니다. 닫으려면 X 버튼을 누르세요.
import tkinter as tk

# 1단계: 메인 창 생성
root = tk.Tk()

# 창 제목 설정
root.title('나의 첫 Tkinter 창')

# 창 크기 설정 (너비x높이)
root.geometry('400x300')

# 3단계: 이벤트 루프 실행
root.mainloop()
````

---
## 3. 창 속성 설정

창의 제목, 크기, 배경색, 리사이즈 여부 등을 설정할 수 있습니다.

````python
# 창 속성 설정 예시
import tkinter as tk

root = tk.Tk()
root.title('창 속성 예시')

# 창 크기: 너비x높이+X좌표+Y좌표
root.geometry('500x400+100+100')

# 배경색 설정
root.configure(bg='lightblue')

# 최소 창 크기 설정
root.minsize(300, 200)

# 리사이즈 비활성화 (너비, 높이)
root.resizable(False, False)

# 창을 화면 중앙에 배치하는 함수
def center_window(win, width, height):
    screen_w = win.winfo_screenwidth()
    screen_h = win.winfo_screenheight()
    x = (screen_w - width) // 2
    y = (screen_h - height) // 2
    win.geometry(f'{width}x{height}+{x}+{y}')

center_window(root, 500, 400)
root.mainloop()
````

## 4. 창 닫기 이벤트 처리

사용자가 창의 X 버튼을 누를 때 정리 작업을 수행할 수 있습니다.
`protocol("WM_DELETE_WINDOW", 콜백함수)`를 사용합니다.

````python
# 창 닫기 이벤트 처리
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.title('닫기 이벤트')
root.geometry('300x200')

def on_closing():
    if messagebox.askokcancel('종료', '정말 종료하시겠습니까?'):
        print('프로그램을 종료합니다.')
        root.destroy()  # 창 파괴

# WM_DELETE_WINDOW 프로토콜에 함수 연결
root.protocol('WM_DELETE_WINDOW', on_closing)

tk.Label(root, text='X 버튼을 눌러보세요', font=('Arial', 14)).pack(expand=True)
root.mainloop()
````

## 5. 여러 창 띄우기 (Toplevel)

메인 창 외에 추가 창을 띄울 때 `tk.Toplevel()`을 사용합니다.
메인 창(`tk.Tk()`)은 반드시 하나만 존재해야 합니다.

````python
# Toplevel로 추가 창 띄우기
import tkinter as tk

root = tk.Tk()
root.title('메인 창')
root.geometry('300x200')

def open_sub_window():
    sub = tk.Toplevel(root)  # 메인 창의 자식 창
    sub.title('서브 창')
    sub.geometry('200x150')
    sub.configure(bg='lightyellow')
    tk.Label(sub, text='서브 창입니다!', bg='lightyellow').pack(expand=True)
    # 서브 창만 닫기 (메인 창은 유지)
    tk.Button(sub, text='닫기', command=sub.destroy).pack(pady=10)

tk.Button(root, text='서브 창 열기', command=open_sub_window).pack(expand=True)
root.mainloop()
````

| 추가 창 | `tk.Toplevel(root)` | 서브 창 |
| 창 닫기 | `root.destroy()` | 창 파괴 |
| 닫기 이벤트 | `root.protocol('WM_DELETE_WINDOW', fn)` | X 버튼 콜백 |
