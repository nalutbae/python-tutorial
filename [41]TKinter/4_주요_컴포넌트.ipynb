# 🧩 주요 컴포넌트 — Checkbutton, Radiobutton, Listbox, Combobox, Scale, Menu, 대화상자, Notebook

Tkinter에서 자주 사용하는 핵심 위젯들을 하나씩 배웁니다. 각 위젯은 **설명 + 실행 가능한 코드 예제**로 구성되어 있습니다. 코드를 직접 실행해 보고 동작을 확인하세요.

> 💡 코드 셀의 `root.mainloop()`가 실행되면 창이 열립니다. 창을 닫아야 다음 셀을 실행할 수 있습니다.

## 1. Checkbutton (체크박스)

여러 항목을 **중복 선택**할 수 있는 위젯입니다. 각 체크박스는 `BooleanVar()`와 연결되어 선택 여부를 저장합니다.

| 주요 옵션 | 설명 |
|-----------|------|
| `text` | 옆에 표시할 텍스트 |
| `variable` | 선택 상태를 저장할 `BooleanVar` |
| `command` | 클릭 시 실행할 함수 |
| `onvalue` / `offvalue` | 선택/해제 시 변수에 저장할 값 |

> 💡 `BooleanVar`는 Tkinter 전용 변수로, `var.get()`으로 현재 값을 얻고 `var.set()`으로 변경할 수 있습니다.


```python
# ┌─────────────────────────────────────────────┐
# │ Checkbutton 예제: 취미 중복 선택                │
# └─────────────────────────────────────────────┘
import tkinter as tk

root = tk.Tk()
root.title("Checkbutton 예제")
root.geometry("300x250")

# 선택 상태를 저장할 변수들
hobby_read = tk.BooleanVar(value=True)
hobby_game = tk.BooleanVar(value=False)
hobby_music = tk.BooleanVar(value=True)

def show_selection():
    selected = []
    if hobby_read.get():
        selected.append("독서")
    if hobby_game.get():
        selected.append("게임")
    if hobby_music.get():
        selected.append("음악")
    result_label.config(text=f"선택: {', '.join(selected) if selected else '없음'}")

tk.Label(root, text="취미를 선택하세요", font=("Arial", 12, "bold")).pack(pady=10)

tk.Checkbutton(root, text="독서", variable=hobby_read).pack(anchor="w", padx=50)
tk.Checkbutton(root, text="게임", variable=hobby_game).pack(anchor="w", padx=50)
tk.Checkbutton(root, text="음악", variable=hobby_music).pack(anchor="w", padx=50)

tk.Button(root, text="확인", command=show_selection).pack(pady=10)
result_label = tk.Label(root, text="선택: 독서, 음악")
result_label.pack()

root.mainloop()
```

## 2. Radiobutton (라디오 버튼)

여러 항목 중 **하나만 선택**할 수 있는 위젯입니다. 모든 라디오 버튼이 **같은 변수**(`StringVar` 또는 `IntVar`)를 공유하면 자동으로 하나만 선택됩니다.

| 주요 옵션 | 설명 |
|-----------|------|
| `text` | 옆에 표시할 텍스트 |
| `value` | 이 버튼 선택 시 변수에 저장될 값 |
| `variable` | 선택 값을 공유할 변수 |
| `command` | 선택 시 실행할 함수 |

> 💡 `variable`이 같으면 하나의 그룹으로 동작합니다. 다른 그룹을 만들려면 다른 변수를 사용하세요.


```python
# ┌─────────────────────────────────────────────┐
# │ Radiobutton 예제: 성별 1개 선택                  │
# └─────────────────────────────────────────────┘
import tkinter as tk

root = tk.Tk()
root.title("Radiobutton 예제")
root.geometry("300x200")

gender = tk.StringVar(value="none")  # 기본값: 선택 없음

def show_selection():
    result_label.config(text=f"선택: {gender.get()}")

tk.Label(root, text="성별을 선택하세요", font=("Arial", 12, "bold")).pack(pady=10)

tk.Radiobutton(root, text="남성", value="남성", variable=gender, command=show_selection).pack(anchor="w", padx=50)
tk.Radiobutton(root, text="여성", value="여성", variable=gender, command=show_selection).pack(anchor="w", padx=50)
tk.Radiobutton(root, text="선택 안 함", value="none", variable=gender, command=show_selection).pack(anchor="w", padx=50)

result_label = tk.Label(root, text="선택: none")
result_label.pack(pady=10)

root.mainloop()
```

## 3. Listbox (목록 상자)

여러 항목을 **스크롤 가능한 목록**으로 표시합니다. 단일 선택 또는 다중 선택이 가능합니다.

| 주요 옵션/메서드 | 설명 |
|------------------|------|
| `selectmode` | `single`(단일), `multiple`(다중), `extended`(Ctrl/Shift) |
| `insert(index, item)` | 항목 추가 |
| `delete(start, end)` | 항목 삭제 |
| `curselection()` | 현재 선택된 항목의 인덱스 튜플 반환 |
| `get(index)` | 인덱스의 항목 텍스트 반환 |
| `height` | 표시할 행 수 |

> 💡 항목이 많으면 `tk.Scrollbar`와 연결해 스크롤을 추가하세요.


```python
# ┌─────────────────────────────────────────────┐
# │ Listbox 예제: 과일 목록에서 선택                │
# └─────────────────────────────────────────────┘
import tkinter as tk

root = tk.Tk()
root.title("Listbox 예제")
root.geometry("300x300")

tk.Label(root, text="과일을 선택하세요", font=("Arial", 12, "bold")).pack(pady=5)

# Listbox와 Scrollbar 생성
listbox = tk.Listbox(root, selectmode="single", height=8)
scrollbar = tk.Scrollbar(root, orient="vertical", command=listbox.yview)
listbox.config(yscrollcommand=scrollbar.set)

# 항목 추가
fruits = ["사과", "배", "포도", "수박", "딸기", "키위", "바나나", "오렌지", "망고", "복숭아"]
for fruit in fruits:
    listbox.insert(tk.END, fruit)

listbox.pack(side="left", padx=(20, 0), pady=10)
scrollbar.pack(side="right", fill="y", pady=10)

def show_selection():
    sel = listbox.curselection()
    if sel:
        result_label.config(text=f"선택: {listbox.get(sel[0])}")
    else:
        result_label.config(text="선택: 없음")

tk.Button(root, text="확인", command=show_selection).pack(pady=5)
result_label = tk.Label(root, text="선택: 없음")
result_label.pack()

root.mainloop()
```

## 4. ttk.Combobox (콤보박스)

`ttk`(Themed Tkinter) 모듈의 콤보박스는 **드롭다운 목록 + 직접 입력**이 모두 가능한 위젯입니다. Listbox보다 공간을 적게 차지합니다.

| 주요 옵션/메서드 | 설명 |
|------------------|------|
| `values` | 드롭다운 항목 목록 (리스트/튜플) |
| `state` | `"normal"`(입력 가능), `"readonly"`(입력 불가) |
| `get()` | 현재 입력/선택된 값 반환 |
| `set(value)` | 값 설정 |
| `current(index)` | 인덱스로 항목 선택 |

> 💡 `from tkinter import ttk`로 ttk를 임포트해야 사용할 수 있습니다.


```python
# ┌─────────────────────────────────────────────┐
# │ ttk.Combobox 예제: 언어 선택                    │
# └─────────────────────────────────────────────┘
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Combobox 예제")
root.geometry("300x200")

tk.Label(root, text="프로그래밍 언어를 선택하세요", font=("Arial", 12, "bold")).pack(pady=10)

languages = ["Python", "Java", "C++", "JavaScript", "Rust", "Go"]

combo = ttk.Combobox(root, values=languages, state="readonly", width=20)
combo.current(0)  # 기본값: 첫 번째 항목
combo.pack(pady=10)

def show_selection():
    result_label.config(text=f"선택: {combo.get()}")

tk.Button(root, text="확인", command=show_selection).pack(pady=5)
result_label = tk.Label(root, text=f"선택: {combo.get()}")
result_label.pack(pady=10)

# 선택 변경 시 자동으로 결과 업데이트
combo.bind("<<ComboboxSelected>>", lambda e: result_label.config(text=f"선택: {combo.get()}"))

root.mainloop()
```

## 5. Scale (슬라이더)

마우스로 **슬라이더를 움직여 숫자 값**을 선택하는 위젯입니다. 온도, 볼륨, 점수 등 범위가 정해진 값 입력에 적합합니다.

| 주요 옵션 | 설명 |
|-----------|------|
| `from_` | 최소값 |
| `to` | 최대값 |
| `orient` | `"horizontal"` 또는 `"vertical"` |
| `resolution` | 값의 간격 (기본 1) |
| `command` | 값 변경 시 실행할 함수 (값이 인수로 전달됨) |
| `variable` | 값을 저장할 `IntVar` / `DoubleVar` |
| `length` | 슬라이더 길이 (픽셀) |

> 💡 `command`에 전달되는 함수는 **문자열 형태의 값**을 인수로 받으므로 필요시 `float()` 변환이 필요합니다.


```python
# ┌─────────────────────────────────────────────┐
# │ Scale 예제: 온도 슬라이더                       │
# └─────────────────────────────────────────────┘
import tkinter as tk

root = tk.Tk()
root.title("Scale 예제")
root.geometry("350x200")

tk.Label(root, text="온도를 조절하세요", font=("Arial", 12, "bold")).pack(pady=10)

temp_var = tk.IntVar(value=20)

def on_change(val):
    temp_label.config(text=f"현재 온도: {int(float(val))}°C")

scale = tk.Scale(
    root, from_=0, to=100, orient="horizontal",
    variable=temp_var, command=on_change, length=250,
    label="온도 (°C)"
)
scale.set(20)
scale.pack(pady=10)

temp_label = tk.Label(root, text="현재 온도: 20°C", font=("Arial", 11))
temp_label.pack(pady=5)

root.mainloop()
```

## 6. Menu (메뉴 바)

창 상단에 **파일, 편집, 도움말** 등의 메뉴를 추가합니다. `root.config(menu=menubar)`로 창에 연결합니다.

| 주요 메서드 | 설명 |
|-------------|------|
| `Menu(master)` | 메뉴 객체 생성 |
| `add_cascade(label=, menu=)` | 하위 메뉴 추가 (상위 항목) |
| `add_command(label=, command=)` | 클릭 가능한 메뉴 항목 |
| `add_separator()` | 구분선 추가 |
| `add_checkbutton()` / `add_radiobutton()` | 메뉴 안에 체크/라디오 추가 |

> 💡 메뉴 바에 하위 메뉴를 "cascade"로 연결하고, 각 하위 메뉴에 `add_command`로 항목을 채웁니다.


```python
# ┌─────────────────────────────────────────────┐
# │ Menu 예제: 파일/도움말 메뉴 바                  │
# └─────────────────────────────────────────────┘
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.title("Menu 예제")
root.geometry("400x300")

# 메뉴 바 생성
menubar = tk.Menu(root)

# 파일 메뉴
file_menu = tk.Menu(menubar, tearoff=0)
file_menu.add_command(label="새 파일", command=lambda: messagebox.showinfo("메뉴", "새 파일을 만듭니다."))
file_menu.add_command(label="열기", command=lambda: messagebox.showinfo("메뉴", "파일을 엽니다."))
file_menu.add_separator()
file_menu.add_command(label="종료", command=root.quit)
menubar.add_cascade(label="파일", menu=file_menu)

# 도움말 메뉴
help_menu = tk.Menu(menubar, tearoff=0)
help_menu.add_command(label="정보", command=lambda: messagebox.showinfo("정보", "주요 컴포넌트 예제 v1.0"))
menubar.add_cascade(label="도움말", menu=help_menu)

root.config(menu=menubar)

tk.Label(root, text="상단 메뉴 바를 클릭해 보세요", font=("Arial", 12)).pack(expand=True)

root.mainloop()
```

## 7. messagebox (메시지 대화상자)

사용자에게 **정보, 경고, 질문**을 알려주는 팝업 창입니다. `from tkinter import messagebox`로 임포트합니다.

| 함수 | 용도 | 반환값 |
|------|------|--------|
| `showinfo(title, msg)` | 정보 알림 | `"ok"` |
| `showwarning(title, msg)` | 경고 | `"ok"` |
| `showerror(title, msg)` | 오류 | `"ok"` |
| `askquestion(title, msg)` | 예/아니오 질문 | `"yes"` / `"no"` |
| `askokcancel(title, msg)` | 확인/취소 | `True` / `False` |
| `askyesno(title, msg)` | 예/아니오 | `True` / `False` |
| `askretrycancel(title, msg)` | 재시도/취소 | `True` / `False` |

> 💡 질문형 대화상자는 반환값을 `if`로 검사하여 사용자의 선택에 따라 동작을 분기할 수 있습니다.


```python
# ┌─────────────────────────────────────────────┐
# │ messagebox 예제: 다양한 대화상자                │
# └─────────────────────────────────────────────┘
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.title("messagebox 예제")
root.geometry("350x300")

tk.Label(root, text="버튼을 눌러 대화상자를 띄워보세요", font=("Arial", 11, "bold")).pack(pady=15)

def btn_info():
    messagebox.showinfo("정보", "작업이 완료되었습니다!")

def btn_warning():
    messagebox.showwarning("경고", "디스크 공간이 부족합니다!")

def btn_error():
    messagebox.showerror("오류", "파일을 찾을 수 없습니다!")

def btn_question():
    answer = messagebox.askyesno("질문", "저장하시겠습니까?")
    result_label.config(text=f"응답: {'예(저장)' if answer else '아니오(취소)'}")

tk.Button(root, text="showinfo", width=15, command=btn_info).pack(pady=3)
tk.Button(root, text="showwarning", width=15, command=btn_warning).pack(pady=3)
tk.Button(root, text="showerror", width=15, command=btn_error).pack(pady=3)
tk.Button(root, text="askyesno", width=15, command=btn_question).pack(pady=3)

result_label = tk.Label(root, text="응답: (대기 중)")
result_label.pack(pady=15)

root.mainloop()
```

## 8. filedialog (파일 대화상자)

파일 **열기/저장** 대화상자를 띄웁니다. `from tkinter import filedialog`로 임포트합니다.

| 함수 | 용도 | 반환값 |
|------|------|--------|
| `askopenfilename()` | 단일 파일 열기 | 파일 경로 문자열 |
| `askopenfilenames()` | 여러 파일 열기 | 경로 튜플 |
| `asksaveasfilename()` | 저장 위치 선택 | 파일 경로 문자열 |
| `askdirectory()` | 폴더 선택 | 폴더 경로 문자열 |

| 주요 옵션 | 설명 |
|-----------|------|
| `title` | 대화상자 제목 |
| `filetypes` | 파일 유형 필터 (예: `[("텍스트", "*.txt"), ("모든 파일", "*.*")]`) |
| `initialdir` | 시작 디렉토리 |
| `defaultextension` | 저장 시 기본 확장자 |

> 💡 사용자가 취소하면 빈 문자열 `""`이 반환됩니다. 항상 빈 문자열인지 확인하세요.


```python
# ┌─────────────────────────────────────────────┐
# │ filedialog 예제: 파일 열기/저장                 │
# └─────────────────────────────────────────────┘
import tkinter as tk
from tkinter import filedialog, messagebox

root = tk.Tk()
root.title("filedialog 예제")
root.geometry("400x250")

tk.Label(root, text="파일 대화상자 테스트", font=("Arial", 12, "bold")).pack(pady=15)

def open_file():
    path = filedialog.askopenfilename(
        title="파일 열기",
        filetypes=[("텍스트 파일", "*.txt"), ("Python 파일", "*.py"), ("모든 파일", "*.*")]
    )
    if path:  # 빈 문자열이 아닐 때만
        result_label.config(text=f"선택: {path}")
    else:
        result_label.config(text="취소됨")

def save_file():
    path = filedialog.asksaveasfilename(
        title="다른 이름으로 저장",
        defaultextension=".txt",
        filetypes=[("텍스트 파일", "*.txt"), ("모든 파일", "*.*")]
    )
    if path:
        result_label.config(text=f"저장 경로: {path}")
    else:
        result_label.config(text="취소됨")

def open_dir():
    path = filedialog.askdirectory(title="폴더 선택")
    if path:
        result_label.config(text=f"폴더: {path}")
    else:
        result_label.config(text="취소됨")

tk.Button(root, text="파일 열기", width=15, command=open_file).pack(pady=3)
tk.Button(root, text="저장", width=15, command=save_file).pack(pady=3)
tk.Button(root, text="폴더 선택", width=15, command=open_dir).pack(pady=3)

result_label = tk.Label(root, text="(대기 중)", wraplength=380)
result_label.pack(pady=15)

root.mainloop()
```

## 9. ttk.Notebook (탭)

여러 페이지를 **탭**으로 전환하며 보여주는 컨테이너 위젯입니다. 설정 창이나 마법사 UI에 적합합니다.

| 주요 메서드 | 설명 |
|-------------|------|
| `ttk.Notebook(master)` | 노트북(탭 컨테이너) 생성 |
| `add(child, text=)` | 탭 추가 (child는 Frame 등) |
| `select(tab_id)` | 탭 전환 (인덱스 또는 위젯) |
| `tab(tab_id, option=value)` | 탭 속성 변경 |
| `bind("<<NotebookTabChanged>>", ...)` | 탭 전환 이벤트 바인딩 |

> 💡 각 탭의 내용은 `tk.Frame`으로 만들고, 그 안에 위젯들을 배치한 뒤 `notebook.add(frame, text="탭명")`으로 추가합니다.


```python
# ┌─────────────────────────────────────────────┐
# │ ttk.Notebook 예제: 탭 UI                       │
# └─────────────────────────────────────────────┘
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Notebook 예제")
root.geometry("400x300")

notebook = ttk.Notebook(root)
notebook.pack(expand=True, fill="both", padx=10, pady=10)

# --- 탭 1: 개인 정보 ---
tab1 = tk.Frame(notebook)
notebook.add(tab1, text="개인 정보")

tk.Label(tab1, text="이름:").grid(row=0, column=0, padx=10, pady=10, sticky="e")
tk.Entry(tab1, width=25).grid(row=0, column=1, padx=10, pady=10)
tk.Label(tab1, text="이메일:").grid(row=1, column=0, padx=10, pady=10, sticky="e")
tk.Entry(tab1, width=25).grid(row=1, column=1, padx=10, pady=10)

# --- 탭 2: 설정 ---
tab2 = tk.Frame(notebook)
notebook.add(tab2, text="설정")

tk.Label(tab2, text="테마:").pack(pady=10)
theme_var = tk.StringVar(value="light")
tk.Radiobutton(tab2, text="라이트", value="light", variable=theme_var).pack(anchor="w", padx=50)
tk.Radiobutton(tab2, text="다크", value="dark", variable=theme_var).pack(anchor="w", padx=50)

tk.Label(tab2, text="글자 크기:").pack(pady=10)
tk.Scale(tab2, from_=8, to=24, orient="horizontal", length=200).pack()

# --- 탭 3: 정보 ---
tab3 = tk.Frame(notebook)
notebook.add(tab3, text="정보")

tk.Label(tab3, text="앱 버전 1.0", font=("Arial", 14, "bold")).pack(pady=30)
tk.Label(tab3, text="Tkinter Notebook 예제\n탭을 클릭하여 페이지를 전환하세요.").pack(pady=10)

root.mainloop()
```

## 🎯 정리

| 위젯 | 용도 | 핵심 포인트 |
|------|------|-------------|
| `Checkbutton` | 중복 선택 | 각각 `BooleanVar` 연결, 여러 개 선택 가능 |
| `Radiobutton` | 단일 선택 | 같은 `variable` 공유 → 그룹 내 1개만 선택 |
| `Listbox` | 목록 선택 | `insert()`로 추가, `curselection()`으로 인덱스 확인, 스크롤바 연결 가능 |
| `ttk.Combobox` | 드롭다운 입력 | `values` 목록, `state="readonly"`로 입력 제한, `get()`/`set()` |
| `Scale` | 슬라이더 값 선택 | `from_`/`to` 범위, `command`로 실시간 변경 감지 |
| `Menu` | 상단 메뉴 바 | `add_cascade`로 하위 메뉴, `add_command`로 항목, `root.config(menu=)` |
| `messagebox` | 팝업 알림 | `showinfo`/`showwarning`/`showerror`, 질문형은 반환값으로 분기 |
| `filedialog` | 파일/폴더 선택 | `askopenfilename`/`asksaveasfilename`/`askdirectory`, 취소 시 빈 문자열 |
| `ttk.Notebook` | 탭 UI | 각 탭은 `Frame`, `add(frame, text=)`로 추가, `<<NotebookTabChanged>>` 이벤트 |

> 💡 **핵심 팁**: 사용자 입력 값을 저장하려면 `tk.StringVar()`, `tk.IntVar()`, `tk.BooleanVar()` 등 Tkinter 변수를 위젯의 `variable` 옵션에 연결하세요. `var.get()`으로 값을 읽고 `var.set()`으로 변경합니다.
