# 🧩 위젯 기초 — 레이블, 버튼, 입력

Tkinter의 가장 기본적인 위젯 3가지를 배웁니다.

| 위젯 | 클래스 | 용도 |
|------|--------|------|
| 레이블 | `tk.Label` | 텍스트/이미지 표시 |
| 버튼 | `tk.Button` | 클릭 이벤트 |
| 입력 | `tk.Entry` | 한 줄 텍스트 입력 |

---
## 1. Label — 텍스트 표시

`tk.Label`은 화면에 텍스트나 이미지를 표시하는 위젯입니다.
가장 자주 사용되는 기본 위젯입니다.

```python
# Label 기본 사용법
import tkinter as tk

root = tk.Tk()
root.title('Label 예시')
root.geometry('400x300')

# 기본 레이블
label1 = tk.Label(root, text='안녕하세요, Tkinter!')
label1.pack(pady=10)

# 폰트, 크기, 색상 설정
label2 = tk.Label(root, text='큰 빨간 글씨',
                  font=('Arial', 20, 'bold'),
                  fg='red',           # 글자색
                  bg='yellow',        # 배경색
                  padx=20, pady=10)   # 내부 여백
label2.pack(pady=10)

# 여러 줄 텍스트
label3 = tk.Label(root, text='첫 번째 줄\n두 번째 줄\n세 번째 줄',
                  justify='center',  # 정렬: left, center, right
                  font=('Consolas', 12))
label3.pack(pady=10)

root.mainloop()
```

---
## 2. Button — 클릭 이벤트

`tk.Button`은 사용자가 클릭할 수 있는 버튼입니다.
`command` 매개변수로 클릭 시 실행할 함수를 지정합니다.

```python
# Button 기본 사용법
import tkinter as tk

root = tk.Tk()
root.title('Button 예시')
root.geometry('400x300')

# 클릭 횟수를 저장할 변수
click_count = {'count': 0}  # 딕셔너리로 감싸서 클로저에서 수정 가능

def on_click():
    click_count['count'] += 1
    label.config(text=f'클릭 횟수: {click_count["count"]}')

# 레이블
label = tk.Label(root, text='클릭 횟수: 0', font=('Arial', 16))
label.pack(pady=20)

# 기본 버튼
btn1 = tk.Button(root, text='클릭하세요', command=on_click)
btn1.pack(pady=10)

# 스타일이 적용된 버튼
btn2 = tk.Button(root, text='파란 버튼',
                 command=on_click,
                 bg='blue', fg='white',
                 font=('Arial', 14, 'bold'),
                 padx=20, pady=10,
                 activebackground='darkblue',  # 클릭 시 배경색
                 activeforeground='yellow')     # 클릭 시 글자색
btn2.pack(pady=10)

# 종료 버튼
btn3 = tk.Button(root, text='종료', command=root.quit,
                 bg='red', fg='white', font=('Arial', 12))
btn3.pack(side='bottom', pady=20)

root.mainloop()
```

## 3. Entry — 한 줄 텍스트 입력

`tk.Entry`는 사용자로부터 한 줄 텍스트를 입력받는 위젯입니다.
입력된 값은 `.get()`으로 가져오고, `.delete()`로 지웁니다.

```python
# Entry 기본 사용법
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.title('Entry 예시')
root.geometry('400x250')

# 입력 필드
tk.Label(root, text='이름:').pack(pady=5)
entry_name = tk.Entry(root, font=('Arial', 14), width=20)
entry_name.pack(pady=5)
entry_name.focus()  # 자동으로 커서 위치

# 비밀번호 입력 (마스킹)
tk.Label(root, text='비밀번호:').pack(pady=5)
entry_pw = tk.Entry(root, font=('Arial', 14), width=20, show='*')  # show로 마스킹 문자 지정
entry_pw.pack(pady=5)

def submit():
    name = entry_name.get()
    pw = entry_pw.get()
    if name and pw:
        messagebox.showinfo('입력 결과', f'이름: {name}\n비밀번호: {"*" * len(pw)}')
    else:
        messagebox.showwarning('경고', '모든 필드를 입력하세요!')

def clear():
    entry_name.delete(0, tk.END)  # 0번부터 끝까지 삭제
    entry_pw.delete(0, tk.END)
    entry_name.focus()

# 버튼 프레임
frame = tk.Frame(root)
frame.pack(pady=20)
tk.Button(frame, text='제출', command=submit, width=10).pack(side='left', padx=5)
tk.Button(frame, text='지우기', command=clear, width=10).pack(side='left', padx=5)

root.mainloop()
```

## 4. StringVar와 양방향 바인딩

`tk.StringVar`를 사용하면 변수와 위젯을 연결할 수 있습니다.
변수가 변경되면 위젯이 자동으로 업데이트됩니다.

```python
# StringVar로 양방향 바인딩
import tkinter as tk

root = tk.Tk()
root.title('StringVar 예시')
root.geometry('400x200')

# StringVar 생성
name_var = tk.StringVar(value='초기값')

# Entry와 StringVar 연결 (textvariable)
entry = tk.Entry(root, textvariable=name_var, font=('Arial', 14), width=25)
entry.pack(pady=20)

# Label도 같은 StringVar를 사용 → 자동 동기화
label = tk.Label(root, textvariable=name_var, font=('Arial', 18, 'bold'), fg='blue')
label.pack(pady=10)

# Entry를 수정하면 Label이 자동으로 변경됩니다!
tk.Label(root, text='↑ 입력창을 수정하면 위 텍스트가 실시간으로 바뀝니다',
        fg='gray').pack()

root.mainloop()
```

## 5. Text — 여러 줄 텍스트 입력

`tk.Text`는 여러 줄의 텍스트를 입력/표시하는 위젯입니다.
간단한 메모장이나 에디터를 만들 때 사용합니다.

```python
# Text 위젯 기본
import tkinter as tk

root = tk.Tk()
root.title('Text 위젯 예시')
root.geometry('400x300')

# Text 위젯 생성
text = tk.Text(root, font=('Consolas', 12), width=40, height=10,
               wrap='word',        # 단어 단위 줄바꿈
               padx=10, pady=10,
               insertbackground='blue')  # 커서 색상
text.pack(pady=10)

# 초기 텍스트 삽입
text.insert('1.0', '여기에 자유롭게\n여러 줄을\n입력할 수 있습니다.\n')
# '1.0' = 1번째 줄, 0번째 문자

def get_text():
    # 전체 텍스트 가져오기 (1.0부터 끝까지)
    content = text.get('1.0', tk.END)
    print(f'입력된 텍스트:\n{content}')

def clear_text():
    text.delete('1.0', tk.END)

frame = tk.Frame(root)
frame.pack(pady=5)
tk.Button(frame, text='가져오기', command=get_text).pack(side='left', padx=5)
tk.Button(frame, text='모두 지우기', command=clear_text).pack(side='left', padx=5)

root.mainloop()
```

## 🎯 정리

| 위젯 | 클래스 | 주요 메서드/속성 |
|------|--------|-----------------|
| 레이블 | `tk.Label` | `text`, `font`, `fg`, `bg`, `padx`, `pady` |
| 버튼 | `tk.Button` | `command`, `activebackground`, `activeforeground` |
| 입력(1줄) | `tk.Entry` | `.get()`, `.delete()`, `.insert()`, `show='*'` |
| 입력(여러줄) | `tk.Text` | `.get('1.0', END)`, `.delete()`, `.insert('1.0', ...)` |
| 변수 바인딩 | `tk.StringVar` | `textvariable=` 연결, 실시간 동기화 |
