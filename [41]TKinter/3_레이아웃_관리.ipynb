# 📐 레이아웃 관리 — pack, grid, place

위젯을 창 안에 어떻게 배치할지 결정하는 3가지 방법을 배웁니다.

| 방식 | 개념 | 적합한 경우 |
|------|------|-------------|
| `pack()` | 위에서 아래로 순차 배치 | 간단한 세로/가로 배치 |
| `grid()` | 표(행/열) 형태 배치 | 폼, 입력 양식 |
| `place()` | 절대 좌표 배치 | 정밀한 위치 지정 |

> ⚠️ 한 부모 안에 pack과 grid를 섞어 쓰면 안 됩니다 (충돌 발생).

---
## 1. pack() — 순차적 배치

가장 간단한 레이아웃 방식입니다.
위젯을 위에서 아래로 (또는 좌에서 우로) 차례로 쌓습니다.

**주요 매개변수:**
- `side`: `top`(기본), `bottom`, `left`, `right`
- `fill`: `x`(가로 채우기), `y`(세로), `both`(둘 다)
- `expand`: `True`면 남은 공간 모두 차지
- `padx`, `pady`: 외부 여백

```python
# pack() 레이아웃 예시
import tkinter as tk

root = tk.Tk()
root.title('pack() 예시')
root.geometry('400x300')

# side='top' (기본) — 위에서 아래로
tk.Label(root, text='top 1', bg='red', fg='white').pack(side='top', fill='x')
tk.Label(root, text='top 2', bg='orange').pack(side='top', fill='x')

# side='bottom' — 아래에서 위로
tk.Label(root, text='bottom 1', bg='blue', fg='white').pack(side='bottom', fill='x')
tk.Label(root, text='bottom 2', bg='cyan').pack(side='bottom', fill='x')

# side='left' — 왼쪽에서 오른쪽으로
tk.Label(root, text='left', bg='green', fg='white', width=10).pack(side='left', pady=20)
tk.Label(root, text='center\n(expand)', bg='yellow').pack(side='left', expand=True, fill='both')
tk.Label(root, text='right', bg='magenta', fg='white', width=10).pack(side='right', pady=20)

root.mainloop()
```

## 2. pack() 실전 — 간단한 로그인 폼

`Frame`을 활용해 입력 필드와 버튼을 그룹화합니다.

```python
# pack() + Frame으로 간단한 로그인 폼
import tkinter as tk

root = tk.Tk()
root.title('로그인')
root.geometry('300x200')

# 제목
tk.Label(root, text='로그인', font=('Arial', 18, 'bold')).pack(pady=15)

# 아이디 프레임
id_frame = tk.Frame(root)
id_frame.pack(pady=5)
tk.Label(id_frame, text='아이디: ', width=8).pack(side='left')
tk.Entry(id_frame, width=20).pack(side='left')

# 비밀번호 프레임
pw_frame = tk.Frame(root)
pw_frame.pack(pady=5)
tk.Label(pw_frame, text='비밀번호: ', width=8).pack(side='left')
tk.Entry(pw_frame, width=20, show='*').pack(side='left')

# 버튼 프레임
btn_frame = tk.Frame(root)
btn_frame.pack(pady=15)
tk.Button(btn_frame, text='로그인', width=10).pack(side='left', padx=5)
tk.Button(btn_frame, text='취소', width=10).pack(side='left', padx=5)

root.mainloop()
```

## 3. grid() — 표 형태 배치

창을 표(행/열)처럼 나누어 위젯을 배치합니다.
폼이나 입력 양식에 가장 적합합니다.

**주요 매개변수:**
- `row`: 행 번호 (0부터 시작)
- `column`: 열 번호 (0부터 시작)
- `rowspan`, `columnspan`: 병합할 행/열 수
- `sticky`: `n`, `s`, `e`, `w` 조합으로 셀 내 정렬
- `padx`, `pady`: 셀 간 여백

```python
# grid() 레이아웃 예시
import tkinter as tk

root = tk.Tk()
root.title('grid() 예시')
root.geometry('400x300')

# 3x3 그리드에 레이블 배치
colors = ['red', 'green', 'blue', 'yellow', 'cyan', 'magenta', 'white', 'gray', 'orange']
for i in range(3):
    for j in range(3):
        idx = i * 3 + j
        tk.Label(root, text=f'({i},{j})', bg=colors[idx],
                font=('Arial', 14), width=10, height=3).grid(
            row=i, column=j, padx=2, pady=2)

root.mainloop()
```

## 4. grid() 실전 — 회원 가입 폼

`grid()`로 정렬된 입력 폼을 만듭니다.
`columnspan`으로 버튼을 가로로 병합합니다.

```python
# grid()로 회원 가입 폼
import tkinter as tk

root = tk.Tk()
root.title('회원 가입')
root.geometry('350x250')

# 열 가중치 설정 (오른쪽 열이 늘어나도록)
root.columnconfigure(1, weight=1)

# 행 0: 이름
tk.Label(root, text='이름:').grid(row=0, column=0, sticky='e', padx=10, pady=8)
tk.Entry(root).grid(row=0, column=1, sticky='ew', padx=10, pady=8)

# 행 1: 이메일
tk.Label(root, text='이메일:').grid(row=1, column=0, sticky='e', padx=10, pady=8)
tk.Entry(root).grid(row=1, column=1, sticky='ew', padx=10, pady=8)

# 행 2: 전화번호
tk.Label(root, text='전화번호:').grid(row=2, column=0, sticky='e', padx=10, pady=8)
tk.Entry(root).grid(row=2, column=1, sticky='ew', padx=10, pady=8)

# 행 3: 성별 (라디오 버튼)
tk.Label(root, text='성별:').grid(row=3, column=0, sticky='e', padx=10, pady=8)
gender_frame = tk.Frame(root)
gender_frame.grid(row=3, column=1, sticky='w', padx=10, pady=8)
tk.Radiobutton(gender_frame, text='남', value='M').pack(side='left')
tk.Radiobutton(gender_frame, text='여', value='F').pack(side='left')

# 행 4: 버튼 (columnspan=2로 가로 병합)
btn_frame = tk.Frame(root)
btn_frame.grid(row=4, column=0, columnspan=2, pady=15)
tk.Button(btn_frame, text='가입', width=12, bg='lightgreen').pack(side='left', padx=5)
tk.Button(btn_frame, text='취소', width=12, bg='lightcoral').pack(side='left', padx=5)

root.mainloop()
```

## 5. place() — 절대 좌표 배치

`place()`는 정확한 픽셀 좌표에 위젯을 배치합니다.
정밀한 제어가 가능하지만, 창 크기가 변해도 위치가 고정됩니다.

**주요 매개변수:**
- `x`, `y`: 좌상단 기준 픽셀 좌표
- `relx`, `rely`: 0.0~1.0 비율 좌표 (창 크기에 비례)
- `anchor`: 기준점 (`nw` 좌상단, `center` 중앙, `se` 우하단)

```python
# place() 레이아웃 예시
import tkinter as tk

root = tk.Tk()
root.title('place() 예시')
root.geometry('400x300')

# 절대 좌표 (x, y 픽셀)
tk.Label(root, text='(50, 50)', bg='red', fg='white').place(x=50, y=50)
tk.Label(root, text='(200, 100)', bg='blue', fg='white').place(x=200, y=100)

# 비율 좌표 (relx, rely — 0.0~1.0)
tk.Label(root, text='정중앙', bg='yellow', font=('Arial', 16)).place(
    relx=0.5, rely=0.5, anchor='center')

# 오른쪽 아래
tk.Label(root, text='우하단', bg='green', fg='white').place(
    relx=0.95, rely=0.95, anchor='se')

# 왼쪽 위
tk.Label(root, text='좌상단', bg='orange').place(
    relx=0.05, rely=0.05, anchor='nw')

root.mainloop()
```

## 6. Frame으로 영역 나누기

`Frame`은 위젯을 그룹화하는 컨테이너입니다.
복잡한 레이아웃을 구성할 때 필수적입니다.
서로 다른 Frame에 서로 다른 레이아웃 방식을 사용할 수 있습니다.

```python
# Frame으로 영역 나누기 (pack + grid 혼합)
import tkinter as tk

root = tk.Tk()
root.title('Frame 예시')
root.geometry('400x300')

# 상단 프레임 (pack으로 배치)
top_frame = tk.Frame(root, bg='lightblue', height=50)
top_frame.pack(fill='x')
tk.Label(top_frame, text='상단 영역', bg='lightblue', font=('Arial', 14)).pack(pady=15)

# 중앙 프레임 (grid로 배치)
center_frame = tk.Frame(root, bg='white')
center_frame.pack(expand=True, fill='both', padx=10, pady=10)

tk.Label(center_frame, text='이름:').grid(row=0, column=0, sticky='e', padx=5, pady=5)
tk.Entry(center_frame, width=25).grid(row=0, column=1, padx=5, pady=5)
tk.Label(center_frame, text='나이:').grid(row=1, column=0, sticky='e', padx=5, pady=5)
tk.Entry(center_frame, width=25).grid(row=1, column=1, padx=5, pady=5)

# 하단 프레임 (pack으로 배치)
bottom_frame = tk.Frame(root, bg='lightgray', height=50)
bottom_frame.pack(fill='x', side='bottom')
tk.Button(bottom_frame, text='저장').pack(side='right', padx=10, pady=10)
tk.Button(bottom_frame, text='취소').pack(side='right', padx=5, pady=10)

root.mainloop()
```

---
## 🎯 정리

| 방식 | 핵심 매개변수 | 장단점 |
|------|--------------|--------|
| `pack()` | `side`, `fill`, `expand` | 간단하지만 복잡한 배치 어려움 |
| `grid()` | `row`, `column`, `columnspan`, `sticky` | 표 형태 배치에 최적 |
| `place()` | `x`, `y`, `relx`, `rely`, `anchor` | 정밀하지만 반응형 아님 |
| `Frame` | 위젯 그룹화 컨테이너 | 각 Frame마다 다른 레이아웃 가능 |

> 💡 **팁**: 복잡한 UI는 Frame으로 영역을 나누고, 각 Frame에 pack 또는 grid를 적용하세요.
