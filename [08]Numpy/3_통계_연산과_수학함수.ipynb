# 📘 통계 연산과 수학 함수

NumPy는 기술 통계, 수학 함수, 누적 연산 등 다양한 계산을 제공합니다.
데이터 분석의 기본이 되는 연산들을 배웁니다.

**학습 목표:**
- 기술 통계 함수 (합, 평균, 분산, 표준편차)
- 축(axis)을 지정한 연산
- 누적 연산 (cumsum, cumprod)
- 수학 함수 (abs, round, where)

## 1. 기술 통계 함수

NumPy의 통계 함수는 **axis** 매개변수로 연산 방향을 지정할 수 있습니다.

```python
# ┌─────────────────────────────────────────┐
# │  1차원 배열 통계 연산                     │
# │  합, 평균, 분산, 표준편차, 최대/최소     │
# └─────────────────────────────────────────┘

arr = np.array([23, 45, 12, 67, 34, 89, 56, 11, 78, 43])
print(f"배열: {arr}")

print(f"\n=== 기술 통계 ===")
print(f"합계:     np.sum(arr)     = {np.sum(arr)}")
print(f"평균:     np.mean(arr)    = {np.mean(arr):.2f}")
print(f"중앙값:   np.median(arr)  = {np.median(arr):.2f}")
print(f"분산:     np.var(arr)     = {np.var(arr):.2f}     # 모분산 (N)")
print(f"불편분산: np.var(arr,ddof=1) = {np.var(arr, ddof=1):.2f}  # 표본분산 (N-1)")
print(f"표준편차: np.std(arr)     = {np.std(arr):.2f}")
print(f"최솟값:   np.min(arr)     = {np.min(arr)}")
print(f"최댓값:   np.max(arr)     = {np.max(arr)}")
print(f"범위:     np.ptp(arr)     = {np.ptp(arr)}      # max - min")

# 인덱스 반환
print(f"\n=== 인덱스 ===")
print(f"최솟값 위치: np.argmin(arr) = {np.argmin(arr)}")
print(f"최댓값 위치: np.argmax(arr) = {np.argmax(arr)}")
```

```python
# ┌─────────────────────────────────────────┐
# │  2차원 배열 축별 연산                    │
# │  axis=0: 열 방향, axis=1: 행 방향       │
# └─────────────────────────────────────────┘

mat = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])
print(f"배열:\n{mat}")

print(f"\n=== 축별 연산 ===")
print(f"전체 평균:   np.mean(mat)       = {np.mean(mat):.2f}")
print(f"열별 평균:   np.mean(mat, axis=0) = {np.mean(mat, axis=0)}")
print(f"행별 평균:   np.mean(mat, axis=1) = {np.mean(mat, axis=1)}")

print(f"\n열별 합계:   np.sum(mat, axis=0)  = {np.sum(mat, axis=0)}")
print(f"행별 합계:   np.sum(mat, axis=1)  = {np.sum(mat, axis=1)}")

print(f"\n열별 최댓값: np.max(mat, axis=0)  = {np.max(mat, axis=0)}")
print(f"행별 최솟값: np.min(mat, axis=1)  = {np.min(mat, axis=1)}")

# 백분위수
arr = np.arange(100)
print(f"\n=== 백분위수 ===")
print(f"25%:  {np.percentile(arr, 25)}")
print(f"50%:  {np.percentile(arr, 50)}  # 중앙값")
print(f"75%:  {np.percentile(arr, 75)}")
```

## 2. 누적 연산과 정렬

누적 합, 누적 곱, 정렬 등의 유용한 함수를 배웁니다.

```python
# ┌─────────────────────────────────────────┐
# │  누적 연산과 정렬                        │
# └─────────────────────────────────────────┘

arr = np.array([3, 1, 4, 1, 5, 9, 2, 6])
print(f"배열: {arr}")

# 누적 합
print(f"\n누적 합 (cumsum): {np.cumsum(arr)}")

# 누적 곱
print(f"누적 곱 (cumprod): {np.cumprod(arr)}")

# 정렬
print(f"\n오름차순 정렬: {np.sort(arr)}")
print(f"내림차순 정렬: {np.sort(arr)[::-1]}")

# 정렬된 인덱스
print(f"\n정렬 인덱스 (argsort): {np.argsort(arr)}")
print(f"→ 정렬 순서대로 원본 인덱스 반환")

# 고유값과 빈도
data = np.array([1, 2, 2, 3, 3, 3, 4, 4, 4, 4])
unique, counts = np.unique(data, return_counts=True)
print(f"\n고유값: {unique}")
print(f"빈도수: {counts}")

# 조건에 맞는 요소 찾기
arr = np.arange(10, 20)
indices = np.where(arr > 15)
print(f"\narr > 15인 인덱스: {indices[0]}")
print(f"arr > 15인 값: {arr[indices]}")
```

## 3. 수학 함수

NumPy는 다양한 수학 함수를 제공합니다.
삼각함수, 지수/로그, 반올림 등을 벡터화된 연산으로 처리합니다.

```python
# ┌─────────────────────────────────────────┐
# │  수학 함수 — 삼각함수, 지수, 로그       │
# └─────────────────────────────────────────┘

# 삼각함수
angles = np.array([0, 30, 45, 60, 90])
rad = np.deg2rad(angles)
print(f"각도: {angles}")
print(f"sin:  {np.round(np.sin(rad), 4)}")
print(f"cos:  {np.round(np.cos(rad), 4)}")
print(f"tan:  {np.round(np.tan(rad), 4)}")

# 지수와 로그
x = np.array([1, 2, 3, 4, 5])
print(f"\n=== 지수와 로그 ===")
print(f"x:       {x}")
print(f"e^x:     {np.exp(x).round(4)}")
print(f"2^x:     {np.power(2, x)}")
print(f"ln(x):   {np.log(x).round(4)}")    # 자연로그
print(f"log2(x): {np.log2(x).round(4)}")   # 밑이 2인 로그
print(f"log10(x): {np.log10(x).round(4)}") # 상용로그

# 반올림
arr = np.array([1.234, 2.567, 3.891])
print(f"\n=== 반올림 ===")
print(f"원본:        {arr}")
print(f"round(2):    {np.round(arr, 2)}")
print(f"floor:       {np.floor(arr)}")  # 내림
print(f"ceil:        {np.ceil(arr)}")   # 올림")

# where (조건부 연산)
arr = np.array([1, -2, 3, -4, 5])
print(f"\n=== where (조건부 선택) ===")
print(f"arr: {arr}")
print(f"양수면 1, 음수면 0: {np.where(arr > 0, 1, 0)}")
print(f"절댓값: {np.abs(arr)}")
```
