## 문제 6: np.where 조건 분기

배열에서 음수는 0으로, 양수는 그대로 출력하세요.

```python

```

````python
# 문제 6: 답을 작성하세요
import numpy as np
arr = np.array([-3, 5, -1, 8, 0, -7, 2])
# 음수 → 0, 양수 → 그대로
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>


## 문제 7: 정렬과 argsort

다음 점수 배열을 내림차순으로 정렬하고, 원래 인덱스도 구하세요.

```python
scores = np.array([78, 92, 65, 88, 71])
```

````python
# 문제 7: 답을 작성하세요
import numpy as np
scores = np.array([78, 92, 65, 88, 71])
# 내림차순 정렬 + 원래 인덱스
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
scores = np.array([78, 92, 65, 88, 71])
sorted_scores = np.sort(scores)[::-1]  # 내림차순
indices = np.argsort(scores)[::-1]     # 원래 인덱스
print(f'정렬: {sorted_scores}')
print(f'인덱스: {indices}')
```

## 문제 10: 파일 입출력

배열을 `.npy` 파일로 저장하고 다시 불러오세요.
저장 전과 불러온 후의 배열이 같은지 `np.array_equal`로 확인하세요.

````python
# 문제 10: 답을 작성하세요
import numpy as np
import tempfile, os
# 배열 저장 → 불러오기 → 비교
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
import tempfile, os
arr = np.arange(10) ** 2
tmpdir = tempfile.mkdtemp()
path = os.path.join(tmpdir, 'test.npy')
np.save(path, arr)
loaded = np.load(path)
print(np.array_equal(arr, loaded))  # True
os.remove(path)
os.rmdir(tmpdir)
```

</details>
