# 📘 선형대수와 난수

NumPy의 `linalg` 모듈은 행렬 연산과 선형대수를 제공하며,
`random` 모듈은 다양한 난수 생성 기능을 제공합니다.

**학습 목표:**
- 행렬 곱과 전치
- 역행렬, 행렬식, 고윳값
- 연립방정식 풀이
- 난수 생성과 시드 설정

## 1. 행렬 연산

NumPy에서 행렬 곱은 `@` 연산자 또는 `np.dot()`을 사용합니다.
요소별 곱셈(`*`)과 행렬 곱(`@`)을 구분하는 것이 중요합니다.

```python
# ┌─────────────────────────────────────────┐
# │  행렬 곱과 요소별 곱셈                   │
# │  * : 요소별 곱셈 (Hadamard 곱)          │
# │  @ 또는 np.dot() : 행렬 곱셈            │
# └─────────────────────────────────────────┘

A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print(f"A:\n{A}")
print(f"\nB:\n{B}")

# 요소별 곱셈 (Hadamard 곱)
print(f"\nA * B (요소별 곱셈):")
print(A * B)

# 행렬 곱셈
print(f"\nA @ B (행렬 곱셈):")
print(A @ B)
print(f"\nnp.dot(A, B):")
print(np.dot(A, B))

# 전치 행렬
print(f"\nA의 전치 (A.T):")
print(A.T)
```

## 2. 역행렬, 행렬식, 고윳값

선형대수의 핵심 연산들입니다.
`np.linalg` 모듈에서 제공합니다.

```python
# ┌─────────────────────────────────────────┐
# │  역행렬, 행렬식, 고윳값                  │
# │  np.linalg 모듈 사용                    │
# └─────────────────────────────────────────┘

A = np.array([[4, 7],
              [2, 6]])
print(f"A:\n{A}")

# 역행렬
A_inv = np.linalg.inv(A)
print(f"\n역행렬 A⁻¹:\n{A_inv}")
print(f"\nA @ A⁻¹ (단위행렬 확인):")
print(np.round(A @ A_inv, 10))

# 행렬식
det = np.linalg.det(A)
print(f"\n행렬식 det(A): {det:.2f}")

# 고윳값과 고유벡터
eigenvalues, eigenvectors = np.linalg.eig(A)
print(f"\n고윳값: {eigenvalues}")
print(f"고유벡터:\n{eigenvectors}")

# 연립방정식 풀이: 4x + 7y = 1, 2x + 6y = 3
b = np.array([1, 3])
x = np.linalg.solve(A, b)
print(f"\n연립방정식 해: x={x[0]:.2f}, y={x[1]:.2f}")
print(f"검증: A @ x = {A @ x}")
```

## 3. 난수 생성

NumPy의 난수 생성은 `np.random.default_rng()`를 사용하는 최신 방식을 권장합니다.
재현 가능한 결과를 위해 시드(seed)를 설정합니다.

```python
# ┌─────────────────────────────────────────┐
# │  난수 생성 — 다양한 분포                  │
# │  np.random.default_rng() 사용 (권장)    │
# └─────────────────────────────────────────┘

# 시드 설정으로 재현 가능한 난수
rng = np.random.default_rng(42)

# 균등 분포 난수 (0~1)
print("=== 균등 분포 (0~1) ===")
print(f"random(5):     {rng.random(5).round(4)}")

# 정수 난수
print(f"\n=== 정수 난수 ===")
print(f"integers(1,10,5): {rng.integers(1, 10, size=5)}")

# 정규분포 난수
print(f"\n=== 정규분포 (평균=0, 표준편차=1) ===")
normal = rng.normal(loc=0, scale=1, size=5)
print(f"normal(5): {normal.round(4)}")

# 지정된 평균과 표준편차
normal_custom = rng.normal(loc=100, scale=15, size=5)
print(f"normal(μ=100, σ=15): {normal_custom.round(2)}")

# 균등분포 (지정 범위)
print(f"\n=== 균등분포 (10~20) ===")
uniform = rng.uniform(10, 20, size=5)
print(f"uniform(10,20,5): {uniform.round(4)}")

# 선택 (복원/비복원)
print(f"\n=== 임의 선택 ===")
items = np.array(['사과', '배', '포도', '귤', '딸기'])
print(f"복원 추출: {rng.choice(items, size=5)}")
print(f"비복원 추출: {rng.choice(items, size=3, replace=False)}")
```

```python
# ┌─────────────────────────────────────────┐
# │  난수 응용 — 시뮬레이션                   │
# │  주사위 100번 던지기 시뮬레이션           │
# └─────────────────────────────────────────┘

rng = np.random.default_rng(42)

# 주사위 10,000번 던지기
dice = rng.integers(1, 7, size=10000)

# 각 눈의 빈도
unique, counts = np.unique(dice, return_counts=True)
print("=== 주사위 10,000번 시뮬레이션 ===")
for face, count in zip(unique, counts):
    print(f"  {face}: {count:5d}회 ({count/100:.1f}%)")

# 평균과 표준편차
print(f"\n평균: {np.mean(dice):.4f} (이론값: 3.5)")
print(f"표준편차: {np.std(dice):.4f}")

# 정규분포 시뮬레이션 (중심극한정리 확인)
sample_means = np.array([rng.normal(loc=100, scale=15, size=30).mean()
                        for _ in range(1000)])
print(f"\n=== 중심극한정리 시뮬레이션 ===")
print(f"모평균: 100")
print(f"표본평균들의 평균: {sample_means.mean():.2f}")
print(f"표본평균들의 표준편차: {sample_means.std():.2f}")
print(f"이론적 SE: {15/np.sqrt(30):.2f}")
```
