````python
# 문제 5: 답을 작성하세요
import numpy as np
rng = np.random.default_rng(0)
data = rng.random(100)
# cumsum으로 이동 평균 (윈도우=5)
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
rng = np.random.default_rng(0)
data = rng.random(100)
w = 5
cs = np.cumsum(data)
cs[w:] = cs[w:] - cs[:-w]
ma = cs[w-1:] / w
print(f'이동평균 처음 5개: {ma[:5].round(3)}')
```

## 문제 8: 불리언 마스킹 — 다중 조건 집계

10,000개의 난수 점수(0~100) 배열에서:
1. 90점 이상의 비율
2. 60점 미만의 개수
3. 70~80점 사이의 평균

을 구하세요.

````python
# 문제 8: 답을 작성하세요
import numpy as np
rng = np.random.default_rng(1)
scores = rng.integers(0, 101, 10_000)
# 1) 90점 이상 비율
# 2) 60점 미만 개수
# 3) 70~80점 평균
````

## 문제 9: np.unique — 빈도수 분석

난수 정수 배열에서 각 값의 빈도수를 구하고, 빈도수 내림차순으로 정렬하세요.
`return_inverse=True`를 사용해 원본 배열의 각 원소가 몇 번째 빈도 그룹에 속하는지도 확인하세요.

<details>

```python
import numpy as np
rng = np.random.default_rng(3)
data = rng.integers(1, 6, 20)
vals, counts, inv = np.unique(data, return_counts=True, return_inverse=True)
order = np.argsort(-counts)
```

## 문제 10: 종합 — 정규화와 상관관계

100명의 수학/영어 점수 난수 데이터에 대해:
1. Z-score 표준화
2. 두 과목의 피어슨 상관계수
3. 수학 점수 상위 10%의 영어 평균

을 구하세요.

````python
# 문제 10: 답을 작성하세요
import numpy as np
rng = np.random.default_rng(99)
math = rng.integers(40, 100, 100).astype(float)
english = rng.integers(40, 100, 100).astype(float)
# 1) Z-score 표준화
# 2) 피어슨 상관계수
# 3) 수학 상위 10%의 영어 평균
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
rng = np.random.default_rng(99)
math = rng.integers(40, 100, 100).astype(float)
english = rng.integers(40, 100, 100).astype(float)
# 1) Z-score
math_z = (math - math.mean()) / math.std(ddof=1)
eng_z = (english - english.mean()) / english.std(ddof=1)
print(f'수학 Z-score: 평균={math_z.mean():.4f}, 표준편차={math_z.std(ddof=1):.4f}')
# 2) 상관계수
corr = np.corrcoef(math, english)[0, 1]
print(f'피어슨 상관계수: {corr:.4f}')
# 3) 상위 10%
top10 = math >= np.percentile(math, 90)
print(f'수학 상위 10%의 영어 평균: {english[top10].mean():.1f}점')
```

</details>
