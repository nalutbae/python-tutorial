# 📘 람다와 고차 함수

**람다(lambda)**는 이름 없는 작은 함수이고, **고차 함수**는 함수를 인수로 받거나 반환하는 함수입니다.

**학습 목표:**
- 람다 표현식의 문법과 용도
- map(), filter(), reduce() 활용
- 정렬 기준으로 람다 사용
- 함수를 인수로 전달하고 반환하기

## 1. 람다 표현식

`lambda 매개변수: 표현식` 형태의 이름 없는 함수입니다.
한 줄로 표현 가능한 간단한 함수에 적합합니다.

```python
# ┌─────────────────────────────────────────┐
# │  람다 표현식 문법                           │
# │  lambda 매개변수: 표현식                    │
# │  def와 같은 기능이지만 이름이 없음            │
# │  복잡한 로직은 def로 함수를 정의하세요         │
# └─────────────────────────────────────────┘

# 기본 람다
square = lambda x: x ** 2
print(f"square(5) = {square(5)}")    # 25
# 여러 매개변수 람다
add = lambda a, b: a + b
print(f"add(3, 5) = {add(3, 5)}")   # 8
```

```python
# 기본값이 있는 람다
greet = lambda name, greeting="Hello": f"{greeting}, {name}!"
print(greet("Alice"))               # Hello, Alice!
print(greet("Bob", "Hi"))           # Hi, Bob!
```

```python
# 조건 표현식이 있는 람다
classify = lambda x: "양수" if x > 0 else "음수" if x < 0 else "영"
print(f"classify(5) = {classify(5)}")    # 양수
print(f"classify(-3) = {classify(-3)}")  # 음수
print(f"classify(0) = {classify(0)}")    # 영
# ┌───────────────────────────────────────────────┐
# │  람다 vs def 비교                                │
# │  lambda x: x ** 2      ← 한 줄, 이름 없음        │
# │  def square(x): return x ** 2  ← 여러 줄, 이름 있음│
# │  복잡한 로직은 def를 사용하세요                    │
# └───────────────────────────────────────────────┘

# 람다는 즉시 사용하고 버릴 때 유용
print(f"즉시 호출: {(lambda x: x * 2)(7)}")  # 14
```

## 2. map, filter, reduce

이들은 함수를 인수로 받는 대표적인 고차 함수입니다.

```python
# ┌─────────────────────────────────────────┐
# │  map(함수, 반복가능객체)                   │
# │  각 요소에 함수를 적용하여 새 이터레이터 반환 │
# │  list()로 변환하여 결과 확인                 │
# └─────────────────────────────────────────┘

numbers = [1, 2, 3, 4, 5]
# map + lambda: 각 요소 제곱
squared = list(map(lambda x: x ** 2, numbers))
print(f"제곱: {squared}")              # [1, 4, 9, 16, 25]
```

```python
# map + named function
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

temps_c = [0, 10, 20, 30, 40]
temps_f = list(map(celsius_to_fahrenheit, temps_c))
print(f"섭씨 → 화씨: {temps_f}")      # [32.0, 50.0, 68.0, ...]
```

```python
# map with multiple iterables
nums1 = [1, 2, 3]
nums2 = [10, 20, 30]
combined = list(map(lambda a, b: a + b, nums1, nums2))
print(f"요소별 합: {combined}")          # [11, 22, 33]
# ┌─────────────────────────────────────────┐
# │  filter(함수, 반복가능객체)               │
# │  함수가 True인 요소만 걸러내어 반환          │
# └─────────────────────────────────────────┘

# filter + lambda: 짝수만 걸러내기
evens = list(filter(lambda x: x % 2 == 0, range(1, 21)))
print(f"짝수: {evens}")
```

```python
# 긴 문자열만 걸러내기
words = ["apple", "banana", "cherry", "date", "elderberry"]
long_words = list(filter(lambda w: len(w) > 5, words))
print(f"5자 초과: {long_words}")       # ['banana', 'cherry', 'elderberry']
# ┌─────────────────────────────────────────┐
# │  reduce(함수, 반복가능객체)                │
# │  누적 연산을 수행하여 단일 값을 반환          │
# │  functools 모듈에서 가져와야 함             │
# └─────────────────────────────────────────┘

from functools import reduce
# 누적 곱
product = reduce(lambda a, b: a * b, range(1, 6))
print(f"5! = {product}")              # 120
```

```python
# 문자열 결합
words = ["Hello", "World", "Python"]
sentence = reduce(lambda a, b: a + " " + b, words)
print(f"결합: {sentence}")            # Hello World Python
# 초기값이 있는 reduce
total = reduce(lambda a, b: a + b, [1, 2, 3], 10)
print(f"초기값 10 + 합: {total}")      # 16
```

## 3. 정렬 기준과 고차 함수

`sorted()`, `max()`, `min()`의 `key` 매개변수에 람다를 전달하여 정렬 기준을 지정합니다.

```python
# sorted() + lambda: 정렬 기준 지정
words = ["banana", "apple", "cherry", "date"]
# 길이순 정렬
by_length = sorted(words, key=lambda w: len(w))
print(f"길이순: {by_length}")        # ['date', 'apple', 'banana', 'cherry']
# 마지막 글자순 정렬
by_last_char = sorted(words, key=lambda w: w[-1])
print(f"마지막 글자순: {by_last_char}")
# 내림차순
by_length_desc = sorted(words, key=lambda w: len(w), reverse=True)
print(f"길이 내림차순: {by_length_desc}")
```

```python
# 딕셔너리 리스트 정렬
students = [
    {"name": "Alice", "age": 25, "grade": 88},
    {"name": "Bob", "age": 22, "grade": 95},
    {"name": "Charlie", "age": 23, "grade": 82},
]

by_age = sorted(students, key=lambda s: s["age"])
print(f"나이순: {[s['name'] for s in by_age]}")

by_grade = sorted(students, key=lambda s: s["grade"], reverse=True)
print(f"성적순: {[s['name'] for s in by_grade]}")
```

```python
# ┌─────────────────────────────────────────┐
# │  함수를 인수로 전달하기                     │
# │  함수는 "일급 객체"입니다                   │
# │  변수에 할당, 인수로 전달, 반환 가능        │
# └─────────────────────────────────────────┘

# 함수를 변수에 할당
def greet(name):
    return f"Hello, {name}!"

say_hello = greet    # 함수를 변수에 할당
print(say_hello("Alice"))    # Hello, Alice!
```

```python
# 함수를 인수로 전달
def apply(func, value):
    return func(value)

print(f"apply(str.upper, 'hello') = {apply(str.upper, 'hello')}")  # HELLO
print(f"apply(len, 'hello') = {apply(len, 'hello')}")              # 5
```

```python
# 함수를 반환하는 함수 (클로저 기초)
def make_multiplier(factor):
    def multiply(x):
        return x * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)
print(f"double(5) = {double(5)}")   # 10
print(f"triple(5) = {triple(5)}")    # 15
```
