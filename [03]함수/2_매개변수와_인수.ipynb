# 📘 매개변수와 인수

파이썬 함수는 다양한 방식으로 **인수를 전달**할 수 있습니다.
위치 인수, 키워드 인수, 기본값, 가변 인수 등을 배웁니다.

**학습 목표:**

- *args와 **kwargs (가변 인수)
- 위치 전용/키워드 전용 매개변수

## 1. 위치 인수와 키워드 인수

함수 호출 시 **위치** 또는 **키워드**로 인수를 전달합니다.

```python
# ┌───────────────────────────────────────────────┐
# │  인수 전달 방식                                  │
# │  위치 인수: func(값1, 값2) — 순서대로 전달         │
# │  키워드 인수: func(키=값) — 이름으로 전달           │
# │  혼합: func(값1, 키=값2) — 위치 먼저, 키워드 나중   │
# └───────────────────────────────────────────────┘

# 기본 매개변수값
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Alice"))                    # Hello, Alice!
print(greet("Alice", greeting="Hi"))    # Hi, Alice!
print(greet(name="Bob", greeting="Hey")) # Hey, Bob!
```

```python
# 다양한 기본값
def power(base, exponent=2):
    return base ** exponent

print(f"power(3) = {power(3)}")           # 9 (기본값 2)
print(f"power(3, 3) = {power(3, 3)}")     # 27
print(f"power(2, exponent=10) = {power(2, exponent=10)}")  # 1024
```

```python
# ┌────────────────────────────────────────────────────┐
# │  변경 가능한(mutable) 기본값 주의점!                  │
# │  리스트, 딕셔너리 등은 기본값으로 사용하면 안 됩니다    │
# │  여러 호출에서 같은 객체가 공유되는 버그 발생          │
# │  해결: None을 기본값으로 하고 함수 내부에서 생성        │
# └────────────────────────────────────────────────────┘

# 잘못된 방식 (mutable default)
def append_wrong(item, lst=[]):
    lst.append(item)
    return lst

print(append_wrong(1))  # [1]
print(append_wrong(2))  # [1, 2] — 같은 리스트 공유!
```

```python
# 올바른 방식 (None을 기본값으로)
def append_correct(item, lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst

print(append_correct(1))  # [1]
print(append_correct(2))  # [2] — 매번 새 리스트
```

## 2. *args와 **kwargs — 가변 인수

`*args`는 임의의 위치 인수를 튜플로, `**kwargs`는 임의의 키워드 인수를 딕셔너리로 받습니다.

```python
# ┌───────────────────────────────────────────────┐
# │  *args: 임의의 위치 인수 → 튜플로 수집            │
# │  **kwargs: 임의의 키워드 인수 → 딕셔너리로 수집   │
# │  순서: 일반 → *args → **kwargs                    │
# └───────────────────────────────────────────────┘

# *args: 임의의 개수 인수
def sum_all(*args):
    print(f"args 타입: {type(args)}")   # <class 'tuple'>
    print(f"args 값: {args}")
    return sum(args)

print(f"합계: {sum_all(1, 2, 3)}")     # (1, 2, 3) → 6
print(f"합계: {sum_all(10, 20)}")       # (10, 20) → 30
```

```python
# 일반 매개변수 + *args
def show_info(name, *hobbies):
    print(f"이름: {name}")
    print(f"취미: {hobbies}")

show_info("김파이", "독서", "코딩", "음악")
```

```python
# **kwargs: 임의의 키워드 인수
def show_profile(**kwargs):
    print(f"kwargs 타입: {type(kwargs)}")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

show_profile(name="Alice", age=30, city="Seoul")
```

```python
# 모두 함께 사용: 일반 → *args → **kwargs
def func(a, b, *args, **kwargs):
    print(f"a={a}, b={b}")
    print(f"args={args}")
    print(f"kwargs={kwargs}")

func(1, 2, 3, 4, x=10, y=20)
# a=1, b=2, args=(3, 4), kwargs={'x': 10, 'y': 20}
```

## 3. 위치 전용과 키워드 전용 매개변수

Python 3.8+에서는 매개변수의 전달 방식을 강제할 수 있습니다.

```python
# ┌───────────────────────────────────────────────────┐
# │  매개변수 전달 방식 강제 (Python 3.8+)                │
# │  / 앞: 위치 전용 (키워드로 전달 불가)                  │
# │  * 뒤: 키워드 전용 (위치로 전달 불가)                  │
# │  순서: 위치전용 / 일반 / 키워드전용                     │
# └───────────────────────────────────────────────────┘

# 위치 전용 매개변수 (/)
def greet_pos(name, /, greeting="Hello"):
    """name은 위치로만 전달 가능"""
    return f"{greeting}, {name}!"

print(greet_pos("Alice"))                    # OK
print(greet_pos("Alice", greeting="Hi"))     # OK
```

```python
# greet_pos(name="Alice")                   # TypeError! 위치 전용

# 키워드 전용 매개변수 (*)
def greet_kw(*, name, greeting="Hello"):
    """name은 키워드로만 전달 가능"""
    return f"{greeting}, {name}!"

print(greet_kw(name="Alice"))               # OK
print(greet_kw(name="Alice", greeting="Hi"))  # OK
```

```python
# greet_kw("Alice")                          # TypeError! 키워드 전용

# 모두 함께: 위치전용 / 일반 / 키워드전용
def full_func(pos_only, /, normal, *, kw_only):
    return f"pos_only={pos_only}, normal={normal}, kw_only={kw_only}"

print(full_func(1, 2, kw_only=3))           # OK
print(full_func(1, normal=2, kw_only=3))     # OK
```

```python
# full_func(pos_only=1, 2, 3)               # TypeError!

# 언패킹으로 인수 전달
def introduce(name, age, city):
    return f"{name}({age}), {city}"

info = ("Alice", 30, "Seoul")
print(introduce(*info))                       # 튜플 언패킹

details = {"name": "Bob", "age": 25, "city": "Busan"}
print(introduce(**details))                   # 딕셔너리 언패킹
```
