# 📘 내장 함수와 고차 함수

파이썬은 **유용한 내장 함수**를 많이 제공하며,
함수를 인수로 받거나 반환하는 **고차 함수**로 강력한 데이터 처리가 가능합니다.

**학습 목표:**
- 자주 쓰는 내장 함수 (타입 변환, 수학, 이터레이블)
- all/any, zip, enumerate 실전 활용
- sorted, min, max의 key 매개변수
- functools.partial로 부분 적용

## 1. 타입 변환과 수학 내장 함수

`int()`, `float()`, `str()`, `bool()` 등으로 자료형을 변환하고,
`abs()`, `round()`, `min()`, `max()` 등으로 수학 연산을 수행합니다.

```python
# ┌─────────────────────────────────────────┐
# │  타입 변환 함수                              │
# │  int(), float(), str(), bool(), list()...  │
# └─────────────────────────────────────────┘

# 타입 변환
print(f"int('42') = {int('42')}")          # 42
print(f"int(3.7) = {int(3.7)}")           # 3 (소수점 버림)
print(f"int('0xff', 16) = {int('0xff', 16)}")  # 255 (16진수)
print(f"float('3.14') = {float('3.14')}")    # 3.14
print(f"str(42) = {str(42)!r}")             # '42'
print(f"bool(0) = {bool(0)}")              # False
print(f"bool('hello') = {bool('hello')}")    # True
```

```python
# 컬렉션 변환
print(f"\nlist('hello') = {list('hello')}")
print(f"tuple([1,2,3]) = {tuple([1,2,3])}")
print(f"set([1,2,2,3]) = {set([1,2,2,3])}")
print(f"dict(a=1, b=2) = {dict(a=1, b=2)}")
```

```python
# ┌─────────────────────────────────────────┐
# │  수학 내장 함수                              │
# │  abs(): 절댓값                              │
# │  round(): 반올림                            │
# │  min/max(): 최솟값/최댓값                   │
# │  sum(): 합계                                │
# │  pow(): 거듭제곱                            │
# │  divmod(): 몫과 나머지                       │
# └─────────────────────────────────────────┘

print(f"\nabs(-5) = {abs(-5)}")              # 5
print(f"round(3.14159, 2) = {round(3.14159, 2)}")  # 3.14
print(f"min(1,2,3) = {min(1,2,3)}")        # 1
print(f"max(1,2,3) = {max(1,2,3)}")        # 3
print(f"sum([1,2,3,4,5]) = {sum([1,2,3,4,5])}")  # 15
print(f"pow(2, 3) = {pow(2, 3)}")           # 8
print(f"divmod(17, 5) = {divmod(17, 5)}")    # (3, 2)
```

## 2. 이터러블 함수 — enumerate, zip, all, any

이 함수들은 반복 가능한 객체를 다루는 데 필수적입니다.

```python
# enumerate(): 인덱스와 값을 함께 반환
fruits = ["apple", "banana", "cherry"]
print("=== enumerate ===")
for index, fruit in enumerate(fruits):
    print(f"  {index}: {fruit}")

for index, fruit in enumerate(fruits, start=1):
    print(f"  {index}번: {fruit}")
```

```python
# zip(): 여러 시퀀스를 병렬로 묶기
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
cities = ["Seoul", "Busan", "Jeju"]

print("\n=== zip ===")
for name, age, city in zip(names, ages, cities):
    print(f"  {name}({age}), {city}")
# zip으로 딕셔너리 만들기
person = dict(zip(["name", "age", "city"], ["Alice", 30, "Seoul"]))
print(f"zip으로 딕셔너리: {person}")
```

```python
# all(): 모든 요소가 참이면 True
print(f"\nall([True, True, True]) = {all([True, True, True])}")    # True
print(f"all([True, False, True]) = {all([True, False, True])}")    # False
print(f"all([1, 2, 3]) = {all([1, 2, 3])}")                     # True
print(f"all([1, 0, 3]) = {all([1, 0, 3])}")                     # False
```

```python
# any(): 하나라도 참이면 True
print(f"any([False, False, True]) = {any([False, False, True])}")  # True
print(f"any([False, False, False]) = {any([False, False, False])}") # False
print(f"any([0, 0, 1]) = {any([0, 0, 1])}")                      # True
```

```python
# 실용 예시: 모든 값이 양수인지 확인
numbers = [1, 2, 3, 4, 5]
print(f"\n모두 양수? {all(n > 0 for n in numbers)}")     # True
print(f"음수 있음? {any(n < 0 for n in numbers)}")         # False
```

## 3. sorted, min, max의 key 매개변수와 functools.partial

`key` 매개변수에 함수를 전달하면 정렬/최솟값/최댓값의 기준을 커스터마이징할 수 있습니다.

```python
# sorted() + key
words = ["banana", "apple", "cherry", "date"]
print(f"길이순: {sorted(words, key=len)}")
print(f"마지막 글자순: {sorted(words, key=lambda w: w[-1])}")
print(f"내림차순: {sorted(words, key=lambda w: len(w), reverse=True)}")
```

```python
# 딕셔너리 리스트 정렬
students = [
    {"name": "Alice", "age": 25, "grade": 88},
    {"name": "Bob", "age": 22, "grade": 95},
    {"name": "Charlie", "age": 23, "grade": 82},
]
by_age = sorted(students, key=lambda s: s["age"])
by_grade = sorted(students, key=lambda s: s["grade"], reverse=True)
print(f"\n나이순: {[s['name'] for s in by_age]}")
print(f"성적순: {[s['name'] for s in by_grade]}")
```

```python
# min/max + key
print(f"\n가장 짧은 단어: {min(words, key=len)}")
print(f"가장 긴 단어: {max(words, key=len)}")
print(f"성적 최고: {max(students, key=lambda s: s['grade'])}")
```

```python
# ┌─────────────────────────────────────────┐
# │  functools.partial: 함수의 일부 인수 고정   │
# │  인수를 부분적으로 적용한 새 함수 생성        │
# └─────────────────────────────────────────┘

from functools import partial

def power(base, exponent):
    return base ** exponent

square = partial(power, exponent=2)
cube = partial(power, exponent=3)

print(f"\nsquare(5) = {square(5)}")    # 25
print(f"cube(3) = {cube(3)}")        # 27
```

```python
# 실용 예시: 기본값이 있는 함수 만들기
def greet(greeting, name):
    return f"{greeting}, {name}!"

hello = partial(greet, greeting="Hello")
hi = partial(greet, greeting="Hi")

print(hello("Alice"))    # Hello, Alice!
print(hi("Bob"))         # Hi, Bob!
```

```python
# reduce: 누적 연산
from functools import reduce
numbers = [1, 2, 3, 4, 5]
product = reduce(lambda a, b: a * b, numbers)
print(f"\n리스트 곱: {product}")    # 120
```
