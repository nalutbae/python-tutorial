## 정답

아래 셀을 실행하면 모든 문제의 정답이 출력됩니다.

````python
# ===== 문제 1 정답 =====
def make_multiplier(n):
    if n <= 0:
        raise ValueError('n must be positive')
    def is_multiple(x):
        return x % n == 0
    return is_multiple

# ===== 문제 2 정답 =====
import functools
import time

def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f"{func.__name__} 실행 시간: {elapsed:.4f}초")
        return result
    return wrapper

# ===== 문제 3 정답 =====
def filter_even(nums):
    for n in nums:
        if n % 2 == 0:
            yield n

def square(nums):
    for n in nums:
        yield n * n

def take(nums, n):
    count = 0
    for item in nums:
        if count >= n:
            break
        yield item
        count += 1

# ===== 문제 4 정답 =====
students = [
    {'name': 'Kim', 'grade': 2, 'score': 85},
    {'name': 'Lee', 'grade': 1, 'score': 92},
    {'name': 'Park', 'grade': 2, 'score': 78},
    {'name': 'Choi', 'grade': 1, 'score': 88},
]
result = sorted(students, key=lambda s: (-s['grade'], s['score']))

# ===== 문제 5 정답 =====
from functools import lru_cache

def fib_naive(n):
    fib_naive.call_count += 1
    if n <= 1:
        return n
    return fib_naive(n - 1) + fib_naive(n - 2)
fib_naive.call_count = 0

@lru_cache(maxsize=None)
def fib_cached(n):
    fib_cached.call_count += 1
    if n <= 1:
        return n
    return fib_cached(n - 1) + fib_cached(n - 2)
fib_cached.call_count = 0

# ===== 문제 6 정답 =====
class FileContext:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
        self.file = None

    def __enter__(self):
        self.file = open(self.filename, self.mode)
        print(f"{self.filename} 열기")
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        print(f"{self.filename} 닫기")
        self.file.close()

# ===== 문제 7 정답 =====
def flatten(items):
    for item in items:
        if isinstance(item, list):
            yield from flatten(item)
        elif isinstance(item, int):
            yield item

# ===== 문제 8 정답 =====
from functools import partial

def power(base, exp):
    return base ** exp

square = partial(power, exp=2)
cube = partial(power, exp=3)

# ===== 문제 9 정답 =====
def uppercase_result(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()
    return wrapper

def exclaim(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result + '!'
    return wrapper

# ===== 문제 10 정답 =====
import sys

n = 10_000_000
list_comp = [x * x for x in range(n)]
gen_expr = (x * x for x in range(n))
list_size = sys.getsizeof(list_comp)
gen_size = sys.getsizeof(gen_expr)
print(f'리스트 내포 크기: {list_size} bytes')
print(f'제너레이터 표현식 크기: {gen_size} bytes')

print('\n모든 정답 출력 완료!')
````
