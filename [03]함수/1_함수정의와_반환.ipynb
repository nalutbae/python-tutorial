# 📘 함수 정의와 반환값

함수는 **재사용 가능한 코드 블록**입니다. 입력을 받아 처리하고 결과를 반환합니다.

**학습 목표:**
- 함수 정의와 호출
- 매개변수와 반환값
- 여러 값 반환하기 (튜플, 딕셔너리)
- 함수 안에서 함수 호출하기
- docstring으로 함수 설명 작성

## 1. 함수 정의와 호출

`def` 키워드로 함수를 정의하고, 함수 이름 뒤에 괄호로 호출합니다.

```python
# ┌───────────────────────────────────────────┐
# │  함수 정의 구조                               │
# │  def 함수이름(매개변수):                       │
# │      """docstring (선택)"""                 │
# │      실행할 코드                               │
# │      return 반환값 (선택)                      │
# └───────────────────────────────────────────┘

# 매개변수 없는 함수
def greet():
    """간단한 인사 함수"""
    print("Hello, World!")

greet()  # Hello, World!
```

```python
# 매개변수 있는 함수
def greet_name(name):
    print(f"Hello, {name}!")

greet_name("Alice")  # Hello, Alice!
```

```python
# 반환값이 있는 함수
def add(a, b):
    return a + b

result = add(3, 5)
print(f"add(3, 5) = {result}")  # 8
```

```python
# ┌─────────────────────────────────────────┐
# │  return이 없는 함수는 None을 반환합니다    │
# └─────────────────────────────────────────┘

def say_hello(name):
    print(f"Hello, {name}")

value = say_hello("Bob")  # Hello, Bob
print(f"반환값: {value}")       # None
```

## 2. 여러 값 반환하기

파이썬은 튜플을 이용해 **여러 값을 동시에 반환**할 수 있습니다.

```python
# 여러 값 반환 (튜플 언패킹)
def divide_and_remainder(a, b):
    """몫과 나머지를 동시에 반환"""
    quotient = a // b
    remainder = a % b
    return quotient, remainder

q, r = divide_and_remainder(17, 5)
print(f"몫: {q}, 나머지: {r}")    # 몫: 3, 나머지: 2
```

```python
# 딕셔너리로 반환
def get_user_info():
    return {"name": "Alice", "age": 30, "city": "Seoul"}

info = get_user_info()
print(f"사용자 정보: {info}")
```

```python
# 리스트로 반환
def min_max(numbers):
    return [min(numbers), max(numbers)]

result = min_max([3, 1, 7, 2, 9])
print(f"최솟값: {result[0]}, 최댓값: {result[1]}")
```

```python
# 조건에 따른 early return
def is_even(number):
    if number % 2 != 0:
        return False
    return True

print(f"is_even(4): {is_even(4)}")   # True
print(f"is_even(7): {is_even(7)}")   # False
```

```python
# 함수 안에서 다른 함수 호출
def double(x):
    return x * 2

def square_then_double(x):
    return double(x ** 2)

print(f"square_then_double(3) = {square_then_double(3)}")  # 18
```

## 3. docstring과 타입 힌트

**docstring**은 함수의 설명서, **타입 힌트**는 매개변수와 반환값의 자료형을 명시합니다.

```python
# ┌───────────────────────────────────────────────────┐
# │  docstring 작성 규칙 (PEP 257)                        │
# │  첫 줄: 요약 (한 줄)                                  │
# │  빈 줄                                                 │
# │  상세 설명, 매개변수, 반환값, 예외 등                   │
# │  help() 함수로 docstring을 볼 수 있습니다               │
# └───────────────────────────────────────────────────┘
```

```python
def calculate_bmi(weight_kg, height_m):
    """체질량지수(BMI)를 계산합니다.
    
    Args:
        weight_kg: 체중(kg)
        height_m: 키(m)
    
    Returns:
        float: BMI 값
    """
    return weight_kg / (height_m ** 2)

bmi = calculate_bmi(70, 1.75)
print(f"BMI: {bmi:.1f}")       # 22.9
# help()로 docstring 확인
print(f"\n함수 이름: {calculate_bmi.__name__}")
print(f"설명: {calculate_bmi.__doc__[:20]}...")
```

```python
# ┌─────────────────────────────────────────┐
# │  타입 힌트 (Type Hints) — Python 3.5+  │
# │  매개변수와 반환값의 자료형을 명시           │
# │  실행에는 영향 없음 (문서화 목적)            │
# └─────────────────────────────────────────┘

def greet_typed(name: str, age: int) -> str:
    """이름과 나이를 받아 인사 문자열을 반환합니다."""
    return f"Hello, {name}! You are {age} years old."

result = greet_typed("Alice", 30)
print(f"\n{result}")
```

```python
# 타입 힌트 확인
from typing import get_type_hints
hints = get_type_hints(greet_typed)
print(f"타입 힌트: {hints}")
```
