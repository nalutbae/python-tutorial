## 문제 1: 단일 표본 t검정

어느 학교 학생 20명의 수학 점수가 다음과 같을 때, 모평균이 70점인지 검정하세요.

## 문제 2: 대응표본 t검정 — 해석

다이어트 전후 체중 데이터에 대해 대응표본 t검정을 수행하고, 효과 크기(Cohen's d)를 계산하세요.

```python
before = [82, 85, 78, 90, 76, 88, 84, 79, 92, 81]
after  = [78, 80, 75, 85, 73, 83, 79, 76, 87, 77]
```

````python
# 문제 2: 답을 작성하세요
import numpy as np
from scipy import stats
before = [82, 85, 78, 90, 76, 88, 84, 79, 92, 81]
after  = [78, 80, 75, 85, 73, 83, 79, 76, 87, 77]
# 대응표본 t검정 + Cohen's d
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
from scipy import stats
before = [82, 85, 78, 90, 76, 88, 84, 79, 92, 81]
after  = [78, 80, 75, 85, 73, 83, 79, 76, 87, 77]
t_stat, p_val = stats.ttest_rel(before, after)
diff = np.array(before) - np.array(after)
cohens_d = diff.mean() / diff.std(ddof=1)
print(f'평균 차이: {diff.mean():.2f}')
print(f't 통계량: {t_stat:.4f}, p값: {p_val:.4f}')
print(f"Cohen's d: {cohens_d:.4f}")
```

</details>

## 문제 4: 신뢰 구간 — 폭 비교

모평균 μ=50, 모표준편차 σ=10인 정규분포에서:
1. 표본 크기 n=25의 95% 신뢰 구간
2. 표본 크기 n=100의 95% 신뢰 구간

두 신뢰 구간의 폭을 비교하세요.

````python
# 문제 4: 답을 작성하세요
import numpy as np
from scipy import stats
mu, sigma = 50, 10
# n=25, n=100 신뢰구간 폭 비교
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
from scipy import stats
mu, sigma = 50, 10
for n in [25, 100]:
    se = sigma / np.sqrt(n)
    z = stats.norm.ppf(0.975)
    margin = z * se
    print(f'n={n}: [{mu-margin:.2f}, {mu+margin:.2f}], 폭={2*margin:.2f}')
# n=25: 폭=7.84, n=100: 폭=3.92 (2배 차이)
```

## 문제 9: Bonferroni 보정

3개의 독립 검정을 수행했을 때 각각의 p값이 0.02, 0.04, 0.008이라면,
Bonferroni 보정을 적용한 후 어떤 결과가 유의한지 판단하세요. (α=0.05)

````python
# 문제 9: 답을 작성하세요
# Bonferroni 보정
p_values = [0.02, 0.04, 0.008]
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
p_values = [0.02, 0.04, 0.008]
n_tests = len(p_values)
alpha = 0.05
bonferroni_alpha = alpha / n_tests
print(f'Bonferroni 보정 α: {bonferroni_alpha:.4f}')
for i, p in enumerate(p_values, 1):
    result = '유의' if p < bonferroni_alpha else '유의하지 않음'
    print(f'  검정 {i}: p={p:.3f} → {result}')
# 검정 1: p=0.02 → 유의하지 않음 (0.02 > 0.0167)
# 검정 2: p=0.04 → 유의하지 않음
# 검정 3: p=0.008 → 유의 (0.008 < 0.0167)
```

## 문제 10: 검정력(Power) 개념 — 시뮬레이션

모평균 μ=52(실제 효과 있음), 모표준편차 σ=10인 정규분포에서 표본 크기 n=30을 1000번 추출하고,
μ=50인 단일 표본 t검정을 수행하세요. 유의수준 α=0.05에서 검정력(귀무가설 기각 비율)을 추정하세요.

````python
# 문제 10: 답을 작성하세요
import numpy as np
from scipy import stats
rng = np.random.default_rng(0)
mu_real, mu_null, sigma, n = 52, 50, 10, 30
# 검정력 시뮬레이션
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
from scipy import stats
rng = np.random.default_rng(0)
mu_real, mu_null, sigma, n = 52, 50, 10, 30
reject = 0
N_sim = 1000
for _ in range(N_sim):
    sample = rng.normal(mu_real, sigma, n)
    t_stat, p_val = stats.ttest_1samp(sample, mu_null)
    if p_val < 0.05:
        reject += 1
power = reject / N_sim
print(f'검정력(추정): {power:.3f}')
```

</details>
