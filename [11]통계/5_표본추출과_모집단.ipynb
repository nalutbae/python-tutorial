# 📘 표본 추출과 모집단 — 추측통계의 기초

통계학에서 가장 중요한 개념 중 하나가 **모집단**과 **표본**의 관계입니다.

우리가 가진 데이터(표본)로부터 전체(모집단)를 추측하는 것이 **추측통계**의 목표입니다.

**학습 목표:**
- 모집단과 표본의 차이 이해
- 무작위 추출 (random sampling)의 개념
- 표본평균과 모집단 평균의 관계
- 모집단 분포와 정규분포
- 난수 시드(seed)와 재현성

## 1. 모집단과 표본이란?

**모집단**(population): 우리가 알고 싶은 전체 대상
**표본**(sample): 모집단에서 일부만 추출한 데이터

> 💡 현실에서는 모집단 전체를 조사하는 것(전수조사)이 불가능한 경우가 많습니다.
> 예: "한국 모든 물고기의 평균 길이"를 구하려면 모든 물고기를 잡을 수 없으니,
> 일부만 잡아서(표본) 전체를 추측합니다.

| 개념 | 설명 | 예시 |
|------|------|------|
| **모집단** | 관심 있는 전체 집단 | 호수의 모든 물고기 |
| **표본** | 모집단에서 추출한 일부 | 잡은 10마리 물고기 |
| **모수** | 모집단의 통계량 | 호수 물고기 전체 평균 |
| **통계량** | 표본으로 계산한 값 | 잡은 10마리의 평균 |

## 2. 작은 호수에서 표본 추출

5마리 물고기밖에 없는 "작은 호수"에서 시작해 봅시다.
이때 모집단은 [2, 3, 4, 5, 6]이며, 모집단 평균은 4입니다.

```python
# ┌─────────────────────────────────────────┐
# │  모집단과 표본 추출                         │
# │  모집단: 전체 대상 (여기서는 5마리 물고기)    │
# │  표본: 모집단에서 일부만 추출한 데이터        │
# │  np.random.choice() → 무작위 추출            │
# └─────────────────────────────────────────┘

import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# seaborn 스타일 설정
sns.set()
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 5마리 물고기가 있는 호수 (모집단)
fish_5 = np.array([2, 3, 4, 5, 6])
print(f"모집단(5마리 물고기): {fish_5}")
print(f"모집단 평균: {np.mean(fish_5)}")
print(f"모집단 크기: {len(fish_5)}")
```

### 무작위 추출 — np.random.choice()

**무작위 추출**(random sampling)은 모집단에서 공평하게 표본을 뽑는 방법입니다.
`np.random.choice()` 함수로 구현합니다.

| 매개변수 | 의미 |
|---------|------|
| `a` | 모집단 (배열) |
| `size` | 추출할 표본 크기 |
| `replace` | 복원추출(True) / 비복원추출(False) |

```python
# ┌─────────────────────────────────────────┐
# │  무작위 추출 시뮬레이션                     │
# │  np.random.choice(모집단, size=표본크기)    │
# │  replace=False → 비복원 추출 (같은 것 중복 불가)│
# │  replace=True  → 복원 추출 (중복 허용)       │
# └─────────────────────────────────────────┘

# 1마리 무작위 추출 (비복원)
sample_1 = np.random.choice(fish_5, size=1, replace=False)
print(f"1마리 추출: {sample_1}")

# 3마리 무작위 추출 (비복원)
sample_3 = np.random.choice(fish_5, size=3, replace=False)
print(f"3마리 추출: {sample_3}")

# 다시 추출하면 다른 결과
sample_3b = np.random.choice(fish_5, size=3, replace=False)
print(f"다시 3마리 추출: {sample_3b}")
print(f"→ 매번 다른 표본이 추출됨")
```

### 난수 시드 — 재현 가능한 결과

실험을 반복하면 매번 다른 결과가 나옵니다.
**난수 시드**(random seed)를 설정하면 항상 같은 결과를 얻을 수 있습니다.
이것은 연구의 **재현성**(reproducibility)을 위해 중요합니다.

```python
# ┌─────────────────────────────────────────┐
# │  난수 시드 (random seed)                    │
# │  np.random.seed(n) → 항상 같은 난수 생성    │
# │  실험의 재현성을 위해 중요                  │
# └─────────────────────────────────────────┘

# 시드를 설정하면 항상 같은 결과
np.random.seed(1)
result1 = np.random.choice(fish_5, size=3, replace=False)
print(f"시드=1, 첫 번째: {result1}")

np.random.seed(1)
result2 = np.random.choice(fish_5, size=3, replace=False)
print(f"시드=1, 두 번째: {result2}")

print(f"\n두 결과가 같은가? {np.array_equal(result1, result2)}")
print("→ 시드를 같게 하면 항상 같은 표본이 추출됨")

# ┌─────────────────────────────────────────┐
# │  표본평균 계산 시뮬레이션                   │
# │  표본을 추출 → 표본평균 계산               │
# │  표본평균은 모집단 평균의 추정값            │
# └─────────────────────────────────────────┘

np.random.seed(1)
sample = np.random.choice(fish_5, size=3, replace=False)
sample_mean = np.mean(sample)
print(f"\n표본: {sample}")
print(f"표본평균: {sample_mean:.2f}")
print(f"모집단 평균: {np.mean(fish_5):.2f}")
print(f"차이: {abs(sample_mean - np.mean(fish_5)):.2f}")
```

## 3. 큰 모집단에서 표본 추출

현실에서는 모집단이 매우 큽니다.
100,000마리 물고기가 있는 호수에서 10마리를 표본으로 추출해 봅시다.

```python
# ┌─────────────────────────────────────────┐
# │  큰 모집단에서 표본 추출                    │
# │  100,000마리 물고기 데이터 읽기             │
# │  모집단의 크기가 클수록 표본의 의미가 커짐  │
# └─────────────────────────────────────────┘

# 100,000마리 물고기 길이 데이터 (모집단)
fish_100000 = pd.read_csv("fish_length_100000.csv")["length"]
print(f"모집단 크기: {len(fish_100000):,}")
print(f"처음 5개: {fish_100000.head().values}")
print(f"모집단 평균: {np.mean(fish_100000):.4f}")
print(f"모집단 표준편차: {np.std(fish_100000, ddof=0):.4f}")
```

```python
# ┌─────────────────────────────────────────┐
# │  표본 추출 시뮬레이션                       │
# │  모집단에서 10마리 무작위 추출               │
# │  표본평균 ≈ 모집단 평균 (근사)             │
# └─────────────────────────────────────────┘

# 10마리 무작위 추출 (비복원)
np.random.seed(42)
sampling_result = np.random.choice(fish_100000, size=10, replace=False)
print(f"표본 (10마리): {np.round(sampling_result, 2)}")
print(f"표본평균: {np.mean(sampling_result):.4f}")
print(f"모집단 평균: {np.mean(fish_100000):.4f}")
print(f"\n→ 표본평균이 모집단 평균에 근사함")
```

## 4. 모집단 분포

**모집단 분포**(population distribution)는 모집단의 데이터가 어떻게 퍼져 있는지를 나타냅니다.
히스토그램으로 시각화하면 분포의 모양을 볼 수 있습니다.

> 💡 모집단 분포의 모양(정규분포, 치우친 분포 등)이
> 표본 추출 결과에 영향을 미칩니다.

```python
# ┌─────────────────────────────────────────┐
# │  모집단 분포의 시각화                       │
# │  히스토그램 → 데이터의 분포 모양 확인        │
# │  모집단 평균, 분산, 표준편차 계산            │
# └─────────────────────────────────────────┘

# 모집단 통계량
pop_mean = np.mean(fish_100000)
pop_std = np.std(fish_100000, ddof=0)
pop_var = np.var(fish_100000, ddof=0)

print(f"=== 모집단 통계량 ===")
print(f"모집단 평균:    {pop_mean:.4f}")
print(f"모집단 분산:    {pop_var:.4f}")
print(f"모집단 표준편차: {pop_std:.4f}")

# 모집단 분포 시각화
plt.figure(figsize=(8, 5))
sns.histplot(fish_100000, bins=50, color='steelblue', edgecolor='white')
plt.axvline(pop_mean, color='red', linestyle='--', label=f'평균={pop_mean:.2f}')
plt.title("모집단 분포 (100,000마리 물고기)")
plt.xlabel("길이 (cm)")
plt.ylabel("빈도")
plt.legend()
plt.tight_layout()
plt.savefig("population_dist.png", dpi=100)
plt.show()
print("모집단 분포 그래프 저장 완료")
```

## 5. 정규분포

**정규분포**(normal distribution)는 통계학에서 가장 중요한 분포입니다.
종 모양의 대칭 곡선이며, 자연계의 많은 데이터가 정규분포를 따릅니다.

정규분포는 두 개의 모수로 결정됩니다:
- **평균** μ: 분포의 중심
- **표준편차** σ: 분포의 퍼짐 정도

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$

> 💡 정규분포의 약 68%가 μ±σ, 약 95%가 μ±2σ 안에 들어갑니다.

```python
# ┌─────────────────────────────────────────┐
# │  정규분포의 확률밀도함수                    │
# │  stats.norm.pdf(x, loc=평균, scale=표준편차)│
# │  정규분포: 평균과 표준편차로 결정            │
# │  종 모양의 대칭 곡선                        │
# └─────────────────────────────────────────┘

# 정규분포 확률밀도함수
x = np.arange(start=1, stop=7.1, step=0.1)

# 평균=4, 표준편차=0.8인 정규분포
pdf_values = stats.norm.pdf(x=x, loc=4, scale=0.8)

print(f"x 범위: {x[0]:.1f} ~ {x[-1]:.1f}")
print(f"정규분호 평균(μ): 4")
print(f"정규분포 표준편차(σ): 0.8")
print(f"\n확률밀도의 최댓값: {np.max(pdf_values):.4f} (평균에서)")

# 정규분포 곡선 그리기
plt.figure(figsize=(8, 5))
plt.plot(x, pdf_values, color='black', linewidth=2)
plt.title("정규분포 확률밀도함수 (μ=4, σ=0.8)")
plt.xlabel("x")
plt.ylabel("확률밀도")
plt.fill_between(x, pdf_values, alpha=0.2, color='steelblue')
plt.tight_layout()
plt.savefig("normal_pdf.png", dpi=100)
plt.show()
print("정규분포 곡선 저장 완료")
```

## 6. 모집단 분포와 정규분포 비교

모집단 히스토그램 위에 정규분포 곡선을 겹쳐 그리면,
실제 데이터가 정규분포에 얼마나 가까운지 확인할 수 있습니다.

```python
# ┌─────────────────────────────────────────┐
# │  모집단 분포와 정규분포 비교                │
# │  히스토그램 + 정규분포 곡선 겹치기          │
# │  데이터가 정규분포를 따르는지 확인          │
# └─────────────────────────────────────────┘

# 모집단 히스토그램 + 정규분포 곡선
plt.figure(figsize=(8, 5))
sns.histplot(fish_100000, bins=50, stat='density',
             color='steelblue', edgecolor='white', alpha=0.6)

# 정규분포 곡선 겹치기
x_range = np.linspace(fish_100000.min(), fish_100000.max(), 200)
plt.plot(x_range, stats.norm.pdf(x_range, loc=pop_mean, scale=pop_std),
         color='red', linewidth=2, label='정규분포 곡선')

plt.title("모집단 분포 vs 정규분포")
plt.xlabel("길이 (cm)")
plt.ylabel("밀도")
plt.legend()
plt.tight_layout()
plt.savefig("population_vs_normal.png", dpi=100)
plt.show()
print("모집단 분포 vs 정규분포 비교 그래프 저장 완료")
```

## 7. 정규분포에서 표본 추출

정규분포를 따르는 모집단에서 직접 표본을 추출할 수도 있습니다.
`stats.norm.rvs()`를 사용하면 정규분포에서 난수를 생성합니다.

| 함수 | 의미 |
|------|------|
| `stats.norm.pdf(x, loc, scale)` | 확률밀도함수 (PDF) |
| `stats.norm.rvs(loc, scale, size)` | 난수 생성 (표본 추출) |
| `stats.norm.cdf(x, loc, scale)` | 누적분포함수 (CDF) |

```python
# ┌─────────────────────────────────────────┐
# │  정규분포에서 표본 추출                     │
# │  stats.norm.rvs(loc=평균, scale=표준편차)   │
# │  실제 모집단 없이도 정규분포 가정으로 추출   │
# └─────────────────────────────────────────┘

# 정규분포 N(4, 0.8²)에서 10개 표본 추출
np.random.seed(1)
sampling_norm = stats.norm.rvs(loc=4, scale=0.8, size=10)
print(f"정규분포에서 추출한 표본:")
print(f"{np.round(sampling_norm, 4)}")
print(f"\n표본평균: {np.mean(sampling_norm):.4f}")
print(f"모집단 평균: 4.0000")
print(f"표본표준편차: {np.std(sampling_norm, ddof=1):.4f}")
print(f"모집단 표준편차: 0.8000")

# ┌─────────────────────────────────────────┐
# │  표본 크기에 따른 표본평균의 변화           │
# │  표본 크기가 클수록 표본평균이 모평균에 수렴│
# └─────────────────────────────────────────┘

print("\n=== 표본 크기에 따른 표본평균 비교 ===")
for n in [10, 100, 1000, 10000]:
    np.random.seed(42)
    sample = stats.norm.rvs(loc=4, scale=0.8, size=n)
    print(f"표본 크기={n:5d}: 표본평균={np.mean(sample):.4f}")

print(f"\n모집단 평균: 4.0000")
print("→ 표본 크기가 클수록 표본평균이 모집단 평균에 가까워짐")
```
