## 문제 3: 공분산과 상관계수

두 변수 X, Y의 공분산과 피어슨 상관계수를 구하세요.

```python
X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]
```

````python
# 문제 3: 답을 작성하세요
import numpy as np
X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]
# 공분산, 상관계수
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]
cov = np.cov(X, Y, ddof=1)[0, 1]
corr = np.corrcoef(X, Y)[0, 1]
print(f'공분산: {cov:.2f}')
print(f'상관계수: {corr:.4f}')
```

## 문제 4: 정규분포 확률

평균 100, 표준편차 15인 정규분포에서:
1. P(X < 85)
2. P(85 ≤ X ≤ 115)

을 구하세요.

````python
# 문제 4: 답을 작성하세요
from scipy import stats
# 정규분포 확률 계산
````

<summary>✅ 정답 보기 (클릭)</summary>


## 문제 6: 신뢰 구간

표본 30개의 평균이 72, 표준편차가 8일 때 95% 신뢰 구간을 구하세요.

````python
# 문제 6: 답을 작성하세요
from scipy import stats
import numpy as np
# 95% 신뢰 구간
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from scipy import stats
import numpy as np
x_bar, s, n = 72, 8, 30
t = stats.t.ppf(0.975, df=n-1)
margin = t * s / np.sqrt(n)
print(f'95% 신뢰구간: [{x_bar-margin:.2f}, {x_bar+margin:.2f}]')
```

</details>

## 문제 8: 카이제곱 검정

관측 빈도가 `[30, 25, 20, 15, 10]`일 때, 동일한 비율인지 검정하세요.

````python
# 문제 8: 답을 작성하세요
from scipy import stats
observed = [30, 25, 20, 15, 10]
# 카이제곱 적합도 검정
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from scipy import stats
observed = [30, 25, 20, 15, 10]
expected = [20] * 5

print(f'카이제곱 통계량: {chi2:.4f}')
print(f'p값: {p_val:.4f}')
```

## 문제 9: 상관관계 시각화

두 변수의 산점도와 회귀선을 그리세요.

```python
rng = np.random.default_rng(1)
x = rng.normal(50, 10, 50)
y = 0.8 * x + rng.normal(0, 5, 50)
```

````python
# 문제 9: 답을 작성하세요
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(1)
x = rng.normal(50, 10, 50)
y = 0.8 * x + rng.normal(0, 5, 50)
# 산점도 + 회귀선
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(1)
x = rng.normal(50, 10, 50)
y = 0.8 * x + rng.normal(0, 5, 50)
plt.scatter(x, y, alpha=0.7)
m, b = np.polyfit(x, y, 1)
plt.plot(x, m*x+b, 'r-', label=f'y={m:.2f}x+{b:.1f}')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.title('산점도와 회귀선')
plt.show()
```

</details>

## 문제 10: 데이터 시각화 종합

50개의 난수 데이터에 대해 히스토그램과 박스플롯을 나란히 그리세요.

<details>

