# 📘 공분산과 상관계수

두 변수 사이의 관계를 수치로 나타내는 대표적인 방법이 **공분산**과 **상관계수**입니다.

- **공분산**(Covariance): 두 변수가 함께 변하는 정도
- **상관계수**(Correlation): 공분산을 -1~1 사이로 표준화한 값

이 노트북에서는 공분산과 피어슨 상관계수를 직접 계산해 봅니다.

**학습 목표:**
- 공분산의 개념과 계산
- 표본공분산과 불편공분산의 차이
- 피어슨 상관계수의 의미와 계산
- 분산-공분산 행렬과 상관행렬

## 1. 공분산이란?

**공분산**은 두 변수가 어떻게 함께 변하는지를 나타냅니다.

- **양의 공분산**: 한 변수가 증가할 때 다른 변수도 증가 (같은 방향)
- **음의 공분산**: 한 변수가 증가할 때 다른 변수는 감소 (반대 방향)
- **0에 가까움**: 두 변수 사이에 선형 관계가 없음

$$\text{Cov}(X,Y) = \frac{1}{N}\sum_{i=1}^{N}(x_i - \bar{x})(y_i - \bar{y})$$

> 💡 공분산은 분산의 "두 변수 버전"입니다.
> 분산이 한 변수의 퍼짐이라면, 공분산은 두 변수가 **함께** 퍼지는 정도입니다.

````python
# ┌─────────────────────────────────────────┐
# │  공분산 계산을 위한 데이터 준비              │
# │  x: 키(cm), y: 몸무게(kg) 같은 관계        │
# │  두 변수가 같이 증가하면 양의 공분산         │
# └─────────────────────────────────────────┘

import numpy as np
import pandas as pd

# CSV 파일에서 공분산 데이터 읽기
cov_data = pd.read_csv("cov.csv")
print("공분산 데이터 (x=키, y=몸무게):")
print(cov_data)

x = cov_data["x"]
y = cov_data["y"]
N = len(cov_data)

# 각 변수의 평균
mu_x = np.mean(x)
mu_y = np.mean(y)
print(f"\nx의 평균: {mu_x:.2f}")
print(f"y의 평균: {mu_y:.2f}")
print(f"샘플 크기 N: {N}")
````

````python
# ┌─────────────────────────────────────────┐
# │  공분산의 직접 계산                        │
# │  1. 각 변수의 편차 계산                     │
# │  2. 편차의 곱을 합산                         │
# │  3. N(또는 N-1)로 나눔                     │
# └─────────────────────────────────────────┘

# Step 1: 편차 계산
dev_x = x - mu_x
dev_y = y - mu_y
print(f"x의 편차: {dev_x.values}")
print(f"y의 편차: {dev_y.values}")

# Step 2: 편차의 곱
product = dev_x * dev_y
print(f"\n편차의 곱 (x편차 × y편차): {product.values}")

# Step 3: 표본공분산 (N으로 나눔)
cov_sample = np.sum(product) / N
print(f"\n표본공분산 (N={N}으로 나눔): {cov_sample:.4f}")

# 불편공분산 (N-1로 나눔)
cov_unbiased = np.sum(product) / (N - 1)
print(f"불편공분산 (N-1={N-1}로 나눔): {cov_unbiased:.4f}")

print(f"\n→ 공분산 > 0: x가 증가할 때 y도 증가하는 경향 (양의 상관)")
````

## 2. 분산-공분산 행렬

**분산-공분산 행렬**(covariance matrix)은 모든 변수 쌍의 공분산을 행렬로 정리합니다.

$$\begin{pmatrix} \text{Var}(X) & \text{Cov}(X,Y) \\ \text{Cov}(Y,X) & \text{Var}(Y) \end{pmatrix}$$

- 대각선: 각 변수의 **분산**
- 비대각선: 두 변수 간의 **공분산**

````python
# ┌─────────────────────────────────────────┐
# │  np.cov() — 분산-공분산 행렬                │
# │  ddof=0 → 표본공분산 (N으로 나눔)           │
# │  ddof=1 → 불편공분산 (N-1로 나눔) ← 기본값  │
# │  대각선: 분산, 비대각선: 공분산              │
# └─────────────────────────────────────────┘

# 표본공분산 행렬 (ddof=0)
cov_matrix_sample = np.cov(x, y, ddof=0)
print("표본공분산 행렬 (ddof=0):")
print(cov_matrix_sample)

# 불편공분산 행렬 (ddof=1, 기본값)
cov_matrix_unbiased = np.cov(x, y, ddof=1)
print("\n불편공분산 행렬 (ddof=1):")
print(cov_matrix_sample)

# 행렬 읽는 법
print("\n=== 행렬 해석 ===")
print(f"Var(X)  = {cov_matrix_unbiased[0,0]:.4f}  ← x의 분산")
print(f"Cov(X,Y)= {cov_matrix_unbiased[0,1]:.4f}  ← x와 y의 공분산")
print(f"Cov(Y,X)= {cov_matrix_unbiased[1,0]:.4f}  ← y와 x의 공분산 (대칭)")
print(f"Var(Y)  = {cov_matrix_unbiased[1,1]:.4f}  ← y의 분산")

# 직접 계산과 비교
print(f"\n직접 계산 불편공분산: {cov_unbiased:.4f}")
print(f"np.cov 결과:           {cov_matrix_unbiased[0,1]:.4f}")
````

## 3. 피어슨 상관계수

**상관계수**는 공분산을 -1과 1 사이로 표준화한 값입니다.
두 변수의 **선형 관계의 강도**를 나타냅니다.

$$r = \frac{\text{Cov}(X,Y)}{\sigma_X \cdot \sigma_Y}$$

| 상관계수 | 의미 |
|---------|------|
| r ≈ +1 | 강한 양의 상관 (같이 증가) |
| r ≈ 0 | 상관 없음 |
| r ≈ -1 | 강한 음의 상관 (하나 증가, 하나 감소) |

> 💡 상관계수는 **방향**뿐만 아니라 **강도**도 알려줍니다.
> |r|이 0.7 이상이면 강한 상관, 0.3 이하면 약한 상관으로 해석합니다.

````python
# ┌─────────────────────────────────────────┐
# │  피어슨 상관계수 직접 계산                   │
# │  r = Cov(X,Y) / (σx × σy)                 │
# │  -1 ≤ r ≤ 1                                │
# │  r ≈ 1: 강한 양의 상관                      │
# │  r ≈ 0: 상관 없음                           │
# │  r ≈ -1: 강한 음의 상관                     │
# └─────────────────────────────────────────┘

# 불편분산과 불편공분산
var_x = np.var(x, ddof=1)
var_y = np.var(y, ddof=1)
cov_xy = np.cov(x, y, ddof=1)[0, 1]

# 피어슨 상관계수
rho = cov_xy / np.sqrt(var_x * var_y)
print(f"x의 불편분산: {var_x:.4f}")
print(f"y의 불편분산: {var_y:.4f}")
print(f"x와 y의 공분산: {cov_xy:.4f}")
print(f"\n피어슨 상관계수 (직접 계산): {rho:.4f}")

# np.corrcoef()로 계산
corr_matrix = np.corrcoef(x, y)
print(f"\nnp.corrcoef 결과:")
print(corr_matrix)
print(f"\n상관계수 r = {corr_matrix[0,1]:.4f}")

# 해석
if abs(rho) >= 0.7:
    strength = "강한"
elif abs(rho) >= 0.3:
    strength = "중간 정도의"
else:
    strength = "약한"

direction = "양의" if rho > 0 else "음의"
print(f"\n해석: {strength} {direction} 상관관계")
````

````python
# ┌─────────────────────────────────────────┐
# │  pandas로 상관계수 계산                     │
# │  .corr() → 피어슨 상관계수                  │
# │  .cov()  → 공분산 행렬                      │
# └─────────────────────────────────────────┘

# DataFrame의 상관계수
print("상관계수 행렬 (corr):")
print(cov_data.corr())

print("\n공분산 행렬 (cov):")
print(cov_data.cov())

# ┌─────────────────────────────────────────┐
# │  상관계수의 범위와 의미                     │
# │  r > 0.7  : 강한 양의 상관                  │
# │  0.3 < r < 0.7: 중간 양의 상관              │
# │  r < 0.3  : 약한 상관                       │
# │  음수면 방향만 반대                           │
# └─────────────────────────────────────────┘

# 다양한 상관관계 시뮬레이션
np.random.seed(42)
n = 100

# 강한 양의 상관
x_strong = np.random.randn(n)
y_strong = x_strong + np.random.randn(n) * 0.3
r_strong = np.corrcoef(x_strong, y_strong)[0, 1]

# 약한 상관
x_weak = np.random.randn(n)
y_weak = np.random.randn(n)
r_weak = np.corrcoef(x_weak, y_weak)[0, 1]

# 음의 상관
x_neg = np.random.randn(n)
y_neg = -x_neg + np.random.randn(n) * 0.5
r_neg = np.corrcoef(x_neg, y_neg)[0, 1]

print(f"\n상관계수 시뮬레이션:")
print(f"  강한 양의 상관: r = {r_strong:.4f}")
print(f"  약한 상관:      r = {r_weak:.4f}")
print(f"  음의 상관:      r = {r_neg:.4f}")
````
