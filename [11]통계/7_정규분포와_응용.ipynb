# 📘 정규분포와 응용

**정규분포**(Normal Distribution)는 통계학에서 가장 중요한 분포입니다.
많은 자연현상의 데이터가 정규분포를 따르며,
통계적 추론과 검정의 기본이 됩니다.

이 노트북에서는 정규분포의 확률밀도, 누적분포함수, 퍼센트 포인트 등을 다룹니다.

**학습 목표:**
- 확률 밀도함수(PDF)와 그래프
- 누적분포함수(CDF)와 백분위수
- 퍼센트 포인트(ppf)와 신뢰구간
- t값과 t분포

## 1. 확률 밀도함수(PDF)

**확률 밀도함수**(PDF)는 특정 값에서 확률 밀도가 얼마인지를 알려줍니다.

정규분포의 확률 밀도함수:

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

> 💡 확률 밀도는 "확률"이 아닙니다. 그래프 아래 면적의 합이 1이 되는 것이 확률입니다.

```python
# ┌─────────────────────────────────────────┐
# │  라이브러리 임포트                          │
# │  numpy → 수치 계산                          │
# │  scipy.stats → 통계 분포                    │
# │  matplotlib, seaborn → 그래프               │
# └─────────────────────────────────────────┘

import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 모집단: 정규분포 N(평균=4, 표준편차=0.8)
population = stats.norm(loc=4, scale=0.8)
print("라이브러리 임포트 완료!")
```

```python
# ┌─────────────────────────────────────────┐
# │  확률 밀도함수 직접 계산                │
# │  그래프 아래 면적의 합 = 1 (확률)     │
# │  수식: f(x) = 1/(σ√(2π)) · exp(-(x-μ)²/(2σ²))│
# └─────────────────────────────────────────┘

# 정규분포 검증: x=3, μ=4, σ=0.8
x = 3
mu = 4
sigma = 0.8

# 직접 계산
pdf_manual = 1 / (np.sqrt(2 * np.pi * sigma**2)) * \
    np.exp(-((x - mu)**2) / (2 * sigma**2))
print(f"직접 계산 확률 밀도 (x={x}): {pdf_manual:.6f}")

# scipy.stats로 계산
pdf_scipy = stats.norm.pdf(x=3, loc=4, scale=0.8)
print(f"scipy.stats 확률 밀도 (x={x}): {pdf_scipy:.6f}")

# 결과 일치 확인
print(f"\n두 결과가 일치하는가? {np.isclose(pdf_manual, pdf_scipy)}")
```

```python
# ┌─────────────────────────────────────────┐
# │  확률 밀도함수 그래프                   │
# │  stats.norm.pdf() → 모든 x값에 대한 확률 밀도  │
# │  종 모양의 대칭 곡선                    │
# └─────────────────────────────────────────┘

# 확률 밀도 곡선 그리기
x_plot = np.arange(start=1, stop=7.1, step=0.1)

plt.figure(figsize=(8, 5))
plt.plot(x_plot, stats.norm.pdf(x=x_plot, loc=4, scale=0.8),
         color='black', linewidth=2)
plt.title('정규분포 확률 밀도함수 (μ=4, σ=0.8)')
plt.xlabel('x')
plt.ylabel('확률 밀도')
plt.fill_between(x_plot, stats.norm.pdf(x=x_plot, loc=4, scale=0.8),
                 alpha=0.2, color='steelblue')
plt.tight_layout()
plt.savefig('normal_pdf.png', dpi=100)
plt.show()
print('확률 밀도함수 그래프 저장 완료')
```

## 2. 누적분포함수(CDF) — 확률 계산

**누적분포함수**(CDF)는 확률 변수가 특정 값 이하가 될 확률을 계산합니다.

$$P(X \leq x) = \text{CDF}(x)$$

> 💡 예: CDF(3) = P(X ≤ 3)는 값이 3 이하일 확률입니다.
> 이는 그래프에서 x=3 왼쪽 면적이 됩니다.

```python
# ┌─────────────────────────────────────────┐
# │  누적분포함수(CDF) — 특정 값 이하의 확률    │
# │  stats.norm.cdf(x, loc, scale)                │
# │  P(X ≤ x)를 계산                              │
# └─────────────────────────────────────────┘

# 시뮬레이션: 모집단에서 100,000개 표본 추출
np.random.seed(1)
simulated_sample = population.rvs(size=100000)

# x=3 이하의 비율 (시뮬레이션)
prop_sim = np.sum(simulated_sample <= 3) / len(simulated_sample)
print(f"시뮬레이션: P(X ≤ 3) = {prop_sim:.4f}")

# CDF로 계산 (이론값)
cdf_3 = stats.norm.cdf(x=3, loc=4, scale=0.8)
print(f"CDF:       P(X ≤ 3) = {cdf_3:.4f}")
print(f"\n→ 시뮬레이션과 CDF가 거의 일치함")
```

```python
# ┌─────────────────────────────────────────┐
# │  CDF의 의미와 특징                         │
# │  CDF(3) = P(X ≤ 3) → 3 이하의 확률      │
# │  CDF(4) = P(X ≤ 4) → 4 이하의 확률 = 0.5  │
# │  (4는 평균이므로 정확히 50%)            │
# └─────────────────────────────────────────┘

# x=4(평균)에서의 CDF
cdf_4 = stats.norm.cdf(x=4, loc=4, scale=0.8)
print(f"CDF(3): P(X ≤ 3) = {cdf_3:.4f}")
print(f"CDF(4): P(X ≤ 4) = {cdf_4:.4f}")
print(f"\n→ 평균(4)에서 CDF가 0.5인 것은 평균을 기준으로 정확히 50%에 데이터가 분포함")
```

## 3. 퍼센트 포인트(ppf) — 확률에서 값 찾기

**퍼센트 포인트**(ppf, percent point function)는 CDF의 역연산입니다.
"P(X ≤ x) = q가 될 때 x는?"이라는 질문에 답합니다.

> 💡 응용: "95% 신뢰구간은?"이런 질문에 답할 수 있습니다.
> 상위 2.5% 지점과 하위 2.5% 지점을 ppf로 구하면 95% 구간이 됩니다.

```python
# ┌─────────────────────────────────────────┐
# │  퍼센트 포인트(ppf) — CDF의 역          │
# │  stats.norm.ppf(q, loc, scale)                │
# │  "P(X ≤ x) = q가 될 때 x는?"에 답        │
# └─────────────────────────────────────────┘

# 하위 2.5% 지점
lower = stats.norm.ppf(loc=4, scale=0.8, q=0.025)
print(f"하위 2.5% 지점: {lower:.4f}")

# 상위 2.5% 지점
upper = stats.norm.ppf(loc=4, scale=0.8, q=0.975)
print(f"상위 2.5% 지점: {upper:.4f}")

print(f"\n95% 신뢰구간: [{lower:.4f}, {upper:.4f}]")
print(f"→ 물고기 길이 [{lower:.2f}, {upper:.2f}]cm 사이에 95%가 포함될 확률")

# ppf는 cdf의 역연산
# CDF(3) 값을 ppf에 넣으면 다시 3이 나와야 함
original_value = 3
cdf_value = stats.norm.cdf(loc=4, scale=0.8, x=original_value)
ppf_value = stats.norm.ppf(loc=4, scale=0.8, q=cdf_value)
print(f"\n=== CDF와 ppf는 역연산 관계 ===")
print(f"원래 값: {original_value}")
print(f"CDF({original_value}) = {cdf_value:.6f}")
print(f"ppf({cdf_value:.6f}) = {ppf_value:.4f}")
print(f"→ ppf는 cdf의 역연산임을 확인")

# 중앙값 (50%)
median = stats.norm.ppf(loc=4, scale=0.8, q=0.5)
print(f"\nppf(0.5) = {median:.4f}  ← 평균과 일치")
```

## 4. t값과 t분포

**t값**(t-value)은 표본평균을 표준오차로 표준화한 값입니다:

$$t = \frac{\bar{x} - \mu}{SE} = \frac{\bar{x} - \mu}{s / \sqrt{n}}$$

모분산이 아닌 표본분산을 사용하기 때문에, 표본 크기가 작으면 정규분포에서 벗어납니다.

> 💡 표본 크기가 큰 경우 t분포는 정규분포에 가까워지지만,
> 표본 크기가 작을수록 꼬리가 더 두껍습니다.

```python
# ┌─────────────────────────────────────────┐
# │  t값의 표본 분포 시뮬레이션               │
# │  t = (표본평균 - 모평균) / (표준오차)  │
# │  표본 크기가 작으면 t분포가 정규분포보다 더 꼬리가 두꺼움  │
# └─────────────────────────────────────────┘

# t값 시뮬레이션
np.random.seed(1)
n_trial = 10000
t_value_array = np.zeros(n_trial)

norm_dist = stats.norm(loc=4, scale=0.8)

for i in range(n_trial):
    sample = norm_dist.rvs(size=10)
    sample_mean = np.mean(sample)
    sample_se = np.std(sample, ddof=1) / np.sqrt(10)
    t_value_array[i] = (sample_mean - 4) / sample_se

print(f"t값 시뮬레이션: {n_trial:,}회")
print(f"t값 평균: {np.mean(t_value_array):.4f} (이론값: 0)")
print(f"t값 표준편차: {np.std(t_value_array, ddof=1):.4f}")
```

```python
# ┌─────────────────────────────────────────┐
# │  t값 분포 vs 정규분포 비교                │
# │  t분포는 정규분포보다 꼬리가 더 두꺼움    │
# │  표본 크기가 커질수록 정규분포에 가까워짐  │
# └─────────────────────────────────────────┘

x = np.arange(start=-8, stop=8.1, step=0.1)

plt.figure(figsize=(8, 5))
# 정규분포 (점선)
plt.plot(x, stats.norm.pdf(x=x), color='black', linestyle='dotted',
         linewidth=2, label='정규분포')
# t분포 df=9 (실선)
plt.plot(x, stats.t.pdf(x=x, df=9), color='steelblue',
         linewidth=2, label='t분포 (df=9)')
plt.title('정규분포 vs t분포 (df=9)')
plt.xlabel('x')
plt.ylabel('확률 밀도')
plt.legend()
plt.tight_layout()
plt.savefig('t_vs_normal.png', dpi=100)
plt.show()

print("t분포의 특징:")
print("  • 정규분포보다 꼬리가 더 넓음 (두껍습니다)")
print("  • 표본 크기가 커질수록 정규분포에 가까워짐")
print("  • df = n - 1 (자유도)")
```

```python
# ┌─────────────────────────────────────────┐
# │  t값 히스토그램 vs t분포 곡선          │
# │  시뮬레이션 t값 분포가 t분포에 가까움  │
# └─────────────────────────────────────────┘

plt.figure(figsize=(8, 5))
# t값 히스토그램
sns.histplot(t_value_array, bins=30, color='steelblue',
             edgecolor='white', stat='density', alpha=0.6)
# t분포 곡선
plt.plot(x, stats.t.pdf(x=x, df=9), color='red',
         linewidth=2, label='t분포 (df=9)')
# 정규분포 곡선 (참고)
plt.plot(x, stats.norm.pdf(x=x), color='black',
         linestyle='dotted', linewidth=1.5, label='정규분포')
plt.title('t값 시뮬레이션 vs t분포')
plt.xlabel('t값')
plt.ylabel('밀도')
plt.legend()
plt.tight_layout()
plt.savefig('t_distribution_sim.png', dpi=100)
plt.show()

print(f"t값 히스토그램과 t분포 곡선 비교 완료")
print(f"\n→ 시뮬레이션 t값이 t분포에 매우 가까워 있음을 확인")
```
