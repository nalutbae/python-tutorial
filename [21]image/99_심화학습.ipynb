## 문제 1: PIL ↔ OpenCV 변환

Pillow 이미지를 만들고 OpenCV 형식(BGR NumPy 배열)으로 변환한 뒤,
다시 Pillow 이미지로 되돌리세요. 변환 과정에서 색상이 유지되는지 확인하세요.

힌트: PIL은 RGB, OpenCV는 BGR 순서입니다.

````python
# 문제 1: 답을 작성하세요
from PIL import Image
import numpy as np
import cv2
# PIL → NumPy(BGR) → PIL 변환
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from PIL import Image
import numpy as np
import cv2
# PIL 이미지 생성
pil_img = Image.new('RGB', (100, 100), (255, 128, 0))
# PIL → OpenCV (RGB → BGR)
cv_img = cv2.cvtColor(np.array(pil_img), cv2.COLOR_RGB2BGR)
# OpenCV → PIL (BGR → RGB)
pil_back = Image.fromarray(cv2.cvtColor(cv_img, cv2.COLOR_BGR2RGB))
print(f'원본 픽셀: {pil_img.getpixel((0, 0))}')
print(f'변환 후 픽셀: {pil_back.getpixel((0, 0))}')
```

## 문제 2: OpenCV — HSV 색상 검출 + 도형 그리기

200×200 BGR 이미지를 만들고, 빨간 원과 파란 사각형을 그린 후,
HSV 변환 후 빨간색 영역만 마스크로 추출하세요.

힌트: 빨간색은 H=0~10과 H=170~180 두 범위를 OR 결합해야 합니다.

````python
# 문제 2: 답을 작성하세요
import cv2
import numpy as np
# 빨간 원 + 파란 사각형 → HSV → 빨간색 마스크
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import cv2
import numpy as np
img = np.zeros((200, 200, 3), dtype=np.uint8)
cv2.circle(img, (50, 100), 40, (0, 0, 255), -1)  # 빨간 원
cv2.rectangle(img, (110, 50), (180, 150), (255, 0, 0), -1)  # 파란 사각형
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
mask1 = cv2.inRange(hsv, np.array([0, 100, 100]), np.array([10, 255, 255]))
mask2 = cv2.inRange(hsv, np.array([170, 100, 100]), np.array([180, 255, 255]))
red_mask = mask1 | mask2
result = cv2.bitwise_and(img, img, mask=red_mask)
print(f'빨간색 픽셀 수: {red_mask.sum() // 255}')
```

각 결과에서 전경(흰색) 픽셀의 비율을 계산하세요.

````python
# 문제 5: 답을 작성하세요
from skimage import filters
import numpy as np
# Otsu vs adaptive 이진화 비교
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from skimage import filters
import numpy as np
rng = np.random.default_rng(42)
img = rng.integers(50, 200, (100, 100)).astype(np.uint8)
# Otsu 이진화
thresh = filters.threshold_otsu(img)
binary_otsu = img > thresh
# 적응형 이진화
thresh_adapt = filters.threshold_local(img, block_size=21, method='gaussian')
binary_adapt = img > thresh_adapt
print(f'Otsu 임계값: {thresh:.1f}, 전경 비율: {binary_otsu.mean():.2%}')
print(f'적응형 전경 비율: {binary_adapt.mean():.2%}')
```

## 문제 6: scikit-image — 세분화 + regionprops

`skimage.data.coins()` 이미지에서 Otsu 이진화 후 라벨링을 수행하고,
면적이 500 이상인 객체의 개수와 평균 면적을 구하세요.

````python
# 문제 6: 답을 작성하세요
from skimage import data, filters, measure
import numpy as np
# 라벨링 + regionprops
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from skimage import data, filters, measure
import numpy as np
img = data.coins()
thresh = filters.threshold_otsu(img)
binary = img > thresh
labels = measure.label(binary)
props = measure.regionprops(labels)
large = [p for p in props if p.area >= 500]
print(f'총 객체: {len(props)}, 면적 500+ 객체: {len(large)}')
if large:
    print(f'평균 면적: {np.mean([p.area for p in large]):.0f}')
```

````python
import numpy as np
````

```python
# NumPy 히스토그램 평활화
hist, _ = np.histogram(np.array(img), 256, (0, 256))
cdf = hist.cumsum()
cdf_m = (cdf - cdf.min()) * 255 // (cdf.max() - cdf.min())
equalized_np = cdf_m[np.array(img)]
# Pillow equalize
equalized_pil = ImageOps.equalize(img)
print(f'원본 픽셀 범위: {np.array(img).min()}-{np.array(img).max()}')
```

## 문제 9: OpenCV — 모폴로지 연산 비교

5×5 네모 패턴에 노이즈(작은 점)를 추가한 이미지에서:
1. 침식(Erosion)으로 노이즈 제거
2. 팽창(Dilation)으로 원래 크기 복원
3. 열기(Opening) = 침식 + 팽창을 한 번에 수행

각 결과의 흰색 픽셀 수를 비교하세요.

````python
# 문제 9: 답을 작성하세요
import cv2
import numpy as np
# 모폴로지 연산 비교
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import cv2
import numpy as np
img = np.zeros((200, 200), dtype=np.uint8)
cv2.rectangle(img, (70, 70), (130, 130), 255, -1)
# 노이즈 추가
rng = np.random.default_rng(0)
noise = rng.random(img.shape) > 0.97
img[noise] = 255
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))
eroded = cv2.erode(img, kernel, iterations=1)
dilated = cv2.dilate(eroded, kernel, iterations=1)
opened = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
print(f'노이즈 포함: {np.sum(img==255):,} 픽셀')
print(f'침식 후: {np.sum(eroded==255):,} 픽셀')
print(f'침식+팽창: {np.sum(dilated==255):,} 픽셀')
print(f'열기(Opening): {np.sum(opened==255):,} 픽셀')
```

## 문제 10: 종합 — 처리 파이프라인

다음 파이프라인을 구현하세요:
1. skimage에서 `data.astronaut()` 이미지 로드
2. NumPy 배열을 OpenCV 형식(BGR)으로 변환
3. 가우시안 블러 적용 (`cv2.GaussianBlur`)
4. HSV로 변환 후 피부색 영역 검출
5. 마스크를 Pillow 이미지로 변환하여 `putalpha()`로 투명도 적용

최종 결과의 크기와 모드를 출력하세요.

````python
# 문제 10: 답을 작성하세요
from skimage import data
import cv2
import numpy as np
from PIL import Image
# skimage → OpenCV → 블러 → 피부색 → Pillow 파이프라인
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from skimage import data
import cv2
import numpy as np
from PIL import Image
# 1) skimage 이미지 로드
img_rgb = data.astronaut()  # RGB
# 2) OpenCV BGR 변환
img_bgr = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)
# 3) 가우시안 블러
blurred = cv2.GaussianBlur(img_bgr, (15, 15), 0)
# 4) HSV 변환 + 피부색 검출
hsv = cv2.cvtColor(blurred, cv2.COLOR_BGR2HSV)
lower_skin = np.array([0, 30, 60])
upper_skin = np.array([25, 255, 255])
mask = cv2.inRange(hsv, lower_skin, upper_skin)
# 5) Pillow로 변환 + 투명도
mask_pil = Image.fromarray(mask)
result = Image.fromarray(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB))
result.putalpha(mask_pil)
print(f'결과 크기: {result.size}, 모드: {result.mode}')
```

</details>
