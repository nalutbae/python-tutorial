## 문제 2: OpenCV — BGR과 RGB 변환

파란 배경 이미지를 만들고, BGR을 RGB로 변환하여 색상이 바뀌는지 확인하세요.
matplotlib으로 두 이미지를 나란히 표시하세요.

````python
# 문제 2: 답을 작성하세요
import cv2
import numpy as np
import matplotlib.pyplot as plt
# BGR 파란 배경 → RGB 변환 비교
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
bgr = np.zeros((100, 100, 3), dtype=np.uint8)
bgr[:, :] = [255, 0, 0]  # BGR: 파랑
rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(bgr); axes[0].set_title('BGR (matplotlib)')
axes[1].imshow(rgb); axes[1].set_title('RGB 변환')
for ax in axes: ax.axis('off')
plt.show()
```

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = np.zeros((200, 200), dtype=np.uint8)

cv2.circle(img, (100, 100), 50, 0, -1)
```

## 문제 9: OpenCV — 회전과 대칭

테스트 이미지를 45도 회전하고, 좌우 대칭을 시키세요.
`cv2.getRotationMatrix2D`와 `cv2.flip`을 사용하세요.

````python
# 문제 9: 답을 작성하세요
import cv2
import numpy as np
import matplotlib.pyplot as plt
# 회전 + 좌우 대칭
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = np.zeros((200, 200, 3), dtype=np.uint8)
cv2.rectangle(img, (50, 50), (150, 150), (0, 255, 0), -1)
# 45도 회전
h, w = img.shape[:2]
M = cv2.getRotationMatrix2D((w//2, h//2), 45, 1.0)
rotated = cv2.warpAffine(img, M, (w, h))
# 좌우 대칭
flipped = cv2.flip(img, 1)
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, im, t in zip(axes, [img, rotated, flipped], ['원본', '45도 회전', '좌우 대칭']):
    ax.imshow(im[..., ::-1]); ax.set_title(t); ax.axis('off')
plt.show()
```

</details>

````python
# 문제 10: 답을 작성하세요
from PIL import Image
import cv2
import numpy as np
# PIL → NumPy → OpenCV 블러 → PIL 변환
````

<details>
<summary>✅ 정답 보기 (클릭)</summary>

```python
from PIL import Image
import cv2
import numpy as np
# PIL 이미지 생성
pil_img = Image.new('RGB', (200, 200), (255, 128, 0))
# PIL → NumPy (RGB)
arr = np.array(pil_img)
# OpenCV 블러 (RGB→BGR 변환 필요)
bgr = cv2.cvtColor(arr, cv2.COLOR_RGB2BGR)
blurred_bgr = cv2.GaussianBlur(bgr, (15, 15), 0)
# BGR → RGB → PIL
blurred_rgb = cv2.cvtColor(blurred_bgr, cv2.COLOR_BGR2RGB)
result = Image.fromarray(blurred_rgb)
print(f'원본: {pil_img.size}, 결과: {result.size}')
```

</details>
