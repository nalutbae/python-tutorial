## 2. 이미지 읽기, 보기, 저장

OpenCV에서 이미지는 **NumPy 배열**로 표현됩니다.
이미지를 읽고, 화면에 표시하고, 파일로 저장하는 기본 작업을 배웁니다.

> 💡 OpenCV는 **BGR**(파랑-초록-빨강) 순서를 사용합니다.
> matplotlib은 **RGB** 순서를 사용하므로 변환이 필요합니다.

````python
# ┌─────────────────────────────────────────┐
# │  테스트 이미지 생성                       │
# │  실제 이미지 파일 대신 NumPy로 생성       │
# └─────────────────────────────────────────┘

# 200×300 크기의 파란색 이미지 (BGR)
blue_img = np.zeros((200, 300, 3), dtype=np.uint8)
blue_img[:, :, 0] = 200  # B 채널

# 200×300 크기의 빨간색 이미지 (BGR)
red_img = np.zeros((200, 300, 3), dtype=np.uint8)
red_img[:, :, 2] = 200  # R 채널

# 200×300 크기의 초록색 이미지 (BGR)
green_img = np.zeros((200, 300, 3), dtype=np.uint8)
green_img[:, :, 1] = 200  # G 채널

# 그라데이션 이미지
gradient = np.zeros((200, 300, 3), dtype=np.uint8)
gradient[:, :, 0] = np.linspace(0, 255, 300).astype(np.uint8)  # B
gradient[:, :, 1] = np.linspace(0, 255, 300).astype(np.uint8)  # G
gradient[:, :, 2] = np.linspace(255, 0, 300).astype(np.uint8)  # R

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, img, title in zip(axes,
    [blue_img, red_img, green_img, gradient],
    ['파랑 (BGR)', '빨강 (BGR)', '초록 (BGR)', '그라데이션']):
    # BGR → RGB 변환
    ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(title)
    ax.axis('off')
plt.tight_layout()
plt.show()

# 이미지 속성 확인
print(f"이미지 shape: {blue_img.shape}  # (높이, 너비, 채널)")
print(f"dtype: {blue_img.dtype}  # uint8 (0~255)")
print(f"전체 픽셀 수: {blue_img.size}")
````

````python
# ┌─────────────────────────────────────────┐
# │  이미지 저장과 읽기                      │
# │  imwrite로 저장, imread로 읽기           │
# └─────────────────────────────────────────┘

import os

# 테스트 이미지 저장
test_dir = "image_test"
os.makedirs(test_dir, exist_ok=True)

# 이미지 저장
cv2.imwrite(os.path.join(test_dir, "blue.png"), blue_img)
cv2.imwrite(os.path.join(test_dir, "gradient.png"), gradient)

# 이미지 읽기
img_read = cv2.imread(os.path.join(test_dir, "gradient.png"))
print(f"읽은 이미지 shape: {img_read.shape}")
print(f"dtype: {img_read.dtype}")

# 읽기 옵션
img_color = cv2.imread(os.path.join(test_dir, "gradient.png"), cv2.IMREAD_COLOR)      # 컬러 (기본)
img_gray = cv2.imread(os.path.join(test_dir, "gradient.png"), cv2.IMREAD_GRAYSCALE)    # 흑백
img_unchanged = cv2.imread(os.path.join(test_dir, "gradient.png"), cv2.IMREAD_UNCHANGED) # 원본 그대로

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].imshow(cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB))
axes[0].set_title(f'컬러: {img_color.shape}')
axes[0].axis('off')
axes[1].imshow(img_gray, cmap='gray')
axes[1].set_title(f'흑백: {img_gray.shape}')
axes[1].axis('off')
axes[2].imshow(cv2.cvtColor(img_unchanged, cv2.COLOR_BGR2RGB))
axes[2].set_title(f'원본: {img_unchanged.shape}')
axes[2].axis('off')
plt.tight_layout()
plt.show()

# 정리
import shutil
shutil.rmtree(test_dir)
print("테스트 파일 정리 완료")
````

## 3. 이미지 구조 이해 — 픽셀과 채널

이미지는 **너비 × 높이 × 채널**의 3차원 NumPy 배열입니다.

| 속성 | 설명 | 예시 |
|------|------|------|
| shape | (높이, 너비, 채널) | (480, 640, 3) |
| dtype | 데이터 타입 | uint8 (0~255) |
| 채널 | BGR 순서 (OpenCV) | [B, G, R] |

> 💡 컬러 이미지는 3채널(BGR), 흑백 이미지는 1채널입니다.

````python
# ┌─────────────────────────────────────────┐
# │  픽셀 접근과 ROI (관심 영역)              │
# │  특정 픽셀 읽기/쓰기, 영역 추출          │
# └─────────────────────────────────────────┘

# 100×100 컬러 이미지 생성
img = np.zeros((100, 100, 3), dtype=np.uint8)

# 특정 픽셀 접근 (BGR 순서)
pixel = img[50, 50]  # (y, x) 순서!
print(f"픽셀 (50,50) BGR 값: {pixel}")

# 픽셀 색상 변경
img[50, 50] = [255, 0, 0]  # 파란색 점
img[30:70, 20:80] = [0, 255, 0]  # 초록 영역
img[10:30, 60:90] = [0, 0, 255]  # 빨간 영역

# ROI (Region of Interest) 추출
roi = img[30:70, 20:80].copy()
print(f"\nROI shape: {roi.shape}")

# 흑백 이미지
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(f"흑백 shape: {gray.shape}  # 채널 없음!")
print(f"흑백 픽셀 (50,50): {gray[50, 50]}")

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title('컬러 이미지')
axes[0].axis('off')
axes[1].imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
axes[1].set_title('ROI 추출')
axes[1].axis('off')
axes[2].imshow(gray, cmap='gray')
axes[2].set_title('흑백 변환')
axes[2].axis('off')
plt.tight_layout()
plt.show()
````

## 4. 색상 공간 변환

OpenCV는 다양한 색상 공간을 지원합니다.
가장 많이 쓰이는 것은 **BGR ↔ RGB ↔ HSV** 변환입니다.

| 변환 | 코드 | 용도 |
|------|------|------|
| BGR → RGB | `COLOR_BGR2RGB` | matplotlib 표시 |
| BGR → HSV | `COLOR_BGR2HSV` | 색상 기반 분리 |
| BGR → GRAY | `COLOR_BGR2GRAY` | 흑백 변환 |
| BGR → LAB | `COLOR_BGR2LAB` | 지각적 색상 |

````python
# ┌─────────────────────────────────────────┐
# │  색상 공간 변환                          │
# │  BGR ↔ RGB ↔ HSV 변환 실습              │
# └─────────────────────────────────────────┘

# 컬러풀한 테스트 이미지 생성
h, w = 200, 300
img_bgr = np.zeros((h, w, 3), dtype=np.uint8)

# 무지개 색상 채우기
for i in range(w):
    hue = int(180 * i / w)
    img_bgr[:, i] = [hue * 255 // 180, 200, 255]  # HSV → BGR 근사

# HSV를 정확히 사용
img_hsv = np.zeros((h, w, 3), dtype=np.uint8)
for i in range(w):
    img_hsv[:, i] = [int(180 * i / w), 255, 255]
img_bgr_from_hsv = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2BGR)

# 다양한 색상 공간 변환
img_rgb = cv2.cvtColor(img_bgr_from_hsv, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_bgr_from_hsv, cv2.COLOR_BGR2GRAY)
img_lab = cv2.cvtColor(img_bgr_from_hsv, cv2.COLOR_BGR2LAB)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes[0, 0].imshow(img_rgb)
axes[0, 0].set_title('RGB (matplotlib 표시용)')
axes[0, 0].axis('off')
axes[0, 1].imshow(img_gray, cmap='gray')
axes[0, 1].set_title('흑백 (GRAY)')
axes[0, 1].axis('off')
axes[1, 0].imshow(img_hsv[:, :, 0], cmap='hsv')
axes[1, 0].set_title('HSV — Hue 채널')
axes[1, 0].axis('off')
axes[1, 1].imshow(cv2.cvtColor(img_lab, cv2.COLOR_LAB2RGB))
axes[1, 1].set_title('LAB')
axes[1, 1].axis('off')
plt.tight_layout()
plt.show()

print("색상 공간 변환 완료!")
print(f"BGR shape: {img_bgr_from_hsv.shape}")
print(f"Gray shape: {img_gray.shape}  # 채널 없음")
````
