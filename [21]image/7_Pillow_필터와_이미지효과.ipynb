# 📘 Pillow 필터와 이미지 효과

Pillow는 다양한 내장 필터와 이미지 효과를 제공합니다.
사진 보정, 예술 효과, 색상 조정 등을 쉽게 적용할 수 있습니다.

**학습 목표:**
- 내장 필터 (BLUR, SHARPEN, EMBOSS 등)
- 이미지 향상 (밝기, 대비, 색상, 선명도)
- 이미지 합성과 마스킹
- 한글 텍스트 렌더링

## 1. 내장 필터

Pillow의 `ImageFilter` 모듈은 다양한 필터를 제공합니다.

| 필터 | 효과 |
|------|------|
| BLUR | 흐리게 |
| SHARPEN | 선명하게 |
| EMBOSS | 양각 효과 |
| FIND_EDGES | 엣지 검출 |
| CONTOUR | 윤곽선 |
| SMOOTH | 부드럽게 |
| EDGE_ENHANCE | 엣지 강조 |

```python
# ┌─────────────────────────────────────────┐
# │  Pillow 내장 필터                       │
# │  ImageFilter 모듈의 다양한 필터 적용     │
# └─────────────────────────────────────────┘

from PIL import Image, ImageDraw, ImageFilter
import numpy as np
import matplotlib.pyplot as plt

# 테스트 이미지 생성
img = Image.new('RGB', (300, 300), (200, 220, 240))
draw = ImageDraw.Draw(img)
draw.rectangle([20, 20, 130, 130], fill=(255, 150, 50), outline='black', width=2)
draw.ellipse([170, 20, 280, 130], fill=(50, 200, 100), outline='black', width=2)
draw.rectangle([80, 170, 220, 280], fill=(100, 100, 255), outline='black', width=2)

# 필터 적용
filters = {
    '원본': img,
    'BLUR': img.filter(ImageFilter.BLUR),
    'SHARPEN': img.filter(ImageFilter.SHARPEN),
    'EMBOSS': img.filter(ImageFilter.EMBOSS),
    'FIND_EDGES': img.filter(ImageFilter.FIND_EDGES),
    'CONTOUR': img.filter(ImageFilter.CONTOUR),
    'EDGE_ENHANCE': img.filter(ImageFilter.EDGE_ENHANCE),
    'SMOOTH': img.filter(ImageFilter.SMOOTH),
}

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, (name, im) in zip(axes.flat, filters.items()):
    ax.imshow(im)
    ax.set_title(name)
    ax.axis('off')
plt.tight_layout()
plt.show()

print('💡 BLUR: 이미지를 흐리게, SHARPEN: 선명하게')
print('💡 EMBOSS: 양각 효과, FIND_EDGES: 엣지 검출')
print('💡 CONTOUR: 윤곽선 추출, EDGE_ENHANCE: 엣지 강조')

# 가우시안 블러 (반경 지정)
img_gauss = img.filter(ImageFilter.GaussianBlur(radius=5))
img_unsharp = img.filter(ImageFilter.UnsharpMask(radius=3, percent=150))
print('\n💡 GaussianBlur(radius=5): 반경 지정 가우시안 블러')
print('💡 UnsharpMask(radius=3, percent=150): 언샵 마스크 필터')
```

## 2. 이미지 향상 (Enhance)

`ImageEnhance` 모듈로 밝기, 대비, 색상, 선명도를 조정합니다.
각 값은 1.0이 원본, 0.0은 최소, 2.0 이상은 강조입니다.

```python
# ┌─────────────────────────────────────────┐
# │  이미지 향상 — 밝기, 대비, 색상, 선명도  │
# │  ImageEnhance 모듈 사용                  │
# └─────────────────────────────────────────┘

from PIL import ImageEnhance

# 테스트 이미지
img = Image.new('RGB', (300, 200), (120, 160, 200))
draw = ImageDraw.Draw(img)
draw.ellipse([50, 30, 150, 170], fill=(255, 200, 50))
draw.rectangle([180, 50, 270, 150], fill=(200, 50, 100))

# 밝기 조정
brightness_low = ImageEnhance.Brightness(img).enhance(0.5)   # 어둡게
brightness_high = ImageEnhance.Brightness(img).enhance(1.5)  # 밝게

# 대비 조정
contrast_low = ImageEnhance.Contrast(img).enhance(0.5)
contrast_high = ImageEnhance.Contrast(img).enhance(2.0)

# 색상 조정
color_low = ImageEnhance.Color(img).enhance(0.3)   # 희미하게
color_high = ImageEnhance.Color(img).enhance(2.0)  # 선명하게
grayscale = ImageEnhance.Color(img).enhance(0.0)   # 흑백

# 선명도 조정
sharp_low = ImageEnhance.Sharpness(img).enhance(0.5)
sharp_high = ImageEnhance.Sharpness(img).enhance(3.0)

fig, axes = plt.subplots(4, 3, figsize=(12, 16))
rows = [
    ('밝기', [brightness_low, img, brightness_high], ['0.5 (어둡게)', '1.0 (원본)', '1.5 (밝게)']),
    ('대비', [contrast_low, img, contrast_high], ['0.5', '1.0 (원본)', '2.0']),
    ('색상', [color_low, grayscale, color_high], ['0.3', '0.0 (흑백)', '2.0']),
    ('선명도', [sharp_low, img, sharp_high], ['0.5', '1.0 (원본)', '3.0']),
]
for row_idx, (title, images, labels) in enumerate(rows):
    for col_idx, (im, label) in enumerate(zip(images, labels)):
        axes[row_idx, col_idx].imshow(im)
        axes[row_idx, col_idx].set_title(f'{title}: {label}')
        axes[row_idx, col_idx].axis('off')
plt.tight_layout()
plt.show()

print('💡 enhance(1.0) = 원본, <1.0 = 감소, >1.0 = 증가')
print('💡 Color(0.0) = 흑백, Brightness(0.0) = 검은 화면')
```

## 3. 이미지 합성과 마스킹

여러 이미지를 합성하거나 마스크를 사용해 일부만 보이게 할 수 있습니다.
`Image.alpha_composite()`, `Image.paste()`, `Image.blend()` 등을 사용합니다.

```python
# ┌─────────────────────────────────────────┐
# │  이미지 합성과 마스킹                     │
# │  paste, blend, alpha_composite            │
# └─────────────────────────────────────────┘

# 배경 이미지
bg = Image.new('RGB', (400, 300), (180, 200, 230))
draw = ImageDraw.Draw(bg)
for x in range(0, 400, 50):
    draw.line([(x, 0), (x, 300)], fill=(160, 180, 210), width=1)
for y in range(0, 300, 50):
    draw.line([(0, y), (400, y)], fill=(160, 180, 210), width=1)

# 전경 이미지 (RGBA)
fg = Image.new('RGBA', (400, 300), (0, 0, 0, 0))
draw_fg = ImageDraw.Draw(fg)
draw_fg.ellipse([100, 50, 300, 250], fill=(255, 100, 50, 180))  # 반투명 주황
draw_fg.rectangle([50, 80, 180, 220], fill=(50, 100, 255, 150))   # 반투란 파랑

# 1. paste (마스크 사용)
img_paste = bg.copy()
# RGBA를 RGB로 변환하면서 마스크 적용
fg_rgb = fg.convert('RGB')
fg_alpha = fg.split()[3]  # 알파 채널을 마스크로
img_paste.paste(fg_rgb, (0, 0), mask=fg_alpha)

# 2. alpha_composite (RGBA 이미지)
bg_rgba = bg.convert('RGBA')
img_composite = Image.alpha_composite(bg_rgba, fg).convert('RGB')

# 3. blend (알파 블렌딩)
img_blend = Image.blend(bg, fg_rgb, alpha=0.5)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
titles = ['배경', 'paste (마스크)', 'alpha_composite', 'blend (0.5)']
images = [bg, img_paste, img_composite, img_blend]
for ax, im, t in zip(axes, images, titles):
    ax.imshow(im)
    ax.set_title(t)
    ax.axis('off')
plt.tight_layout()
plt.show()

print('💡 paste(mask=alpha): 알파 채널을 마스크로 사용')
print('💡 alpha_composite: 두 RGBA 이미지 합성')
print('💡 blend(alpha=0.5): 50:50 비율로 혼합')
```

## 4. 한글 텍스트 렌더링

OpenCV는 한글 텍스트 렌더링이 어렵지만, Pillow는 시스템 폰트를 사용해
한글을 자연스럽게 렌더링할 수 있습니다.

```python
# ┌─────────────────────────────────────────┐
# │  한글 텍스트 렌더링                     │
# │  시스템 폰트를 사용한 한글 출력           │
# └─────────────────────────────────────────┘

import os

# 한글 폰트 경로 (Windows)
font_paths = [
    'C:/Windows/Fonts/malgun.ttf',       # 맑은 고딕
    'C:/Windows/Fonts/malgunbd.ttf',     # 맑은 고딕 굵게
    'C:/Windows/Fonts/NanumGothic.ttf',  # 나눔고딕
]

# 사용 가능한 폰트 찾기
font_path = None
for fp in font_paths:
    if os.path.exists(fp):
        font_path = fp
        break

img = Image.new('RGB', (400, 200), 'white')
draw = ImageDraw.Draw(img)

if font_path:
    # 폰트 로드
    font_large = ImageFont.truetype(font_path, 28)
    font_small = ImageFont.truetype(font_path, 16)
    font_title = ImageFont.truetype(font_path, 20)

    draw.text((20, 20), '안녕하세요! Pillow 한글 테스트', fill='black', font=font_large)
    draw.text((20, 60), '파이썬 이미지 처리 라이브러리', fill='blue', font=font_title)
    draw.text((20, 95), 'Pillow로 한글 텍스트를 자연스럽게 렌더링할 수 있습니다.',
              fill='gray', font=font_small)
    draw.text((20, 130), '크기, 색상, 폰트를 자유롭게 지정 가능!',
              fill='green', font=font_small)
    draw.text((20, 160), 'OpenCV는 한글이 깨지지만 Pillow는 OK!',
              fill='red', font=font_small)
    print(f'사용 폰트: {font_path}')
else:
    # 기본 폰트
    draw.text((20, 20), 'Pillow Text (default font)', fill='black')
    draw.text((20, 50), 'Korean requires TTF font', fill='blue')
    print('한글 폰트 없음 - 기본 폰트 사용')

plt.figure(figsize=(10, 5))
plt.imshow(img)
plt.axis('off')
plt.title('Pillow 한글 텍스트 렌더링')
plt.tight_layout()
plt.show()

print('\n💡 ImageFont.truetype(경로, 크기): 시스템 TTF 폰트 로드')
print('💡 Windows: C:/Windows/Fonts/ 디렉토리에 폰트 파일 있음')
print('💡 OpenCV putText()는 한글이 깨지지만 Pillow는 정상 렌더링')
```
