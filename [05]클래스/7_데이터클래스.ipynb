# 📘 데이터클래스 (dataclass)

**데이터클래스**는 주로 데이터를 저장하는 클래스를 간결하게 만드는 기능입니다.
`__init__`, `__repr__`, `__eq__` 등을 자동 생성합니다.

**학습 목표:**
- @dataclass 기본 사용법
- 기본값과 field() 옵션
- frozen 데이터클래스 (불변)
- 데이터클래스 상속과 정렬

## 1. 기본 데이터클래스

`@dataclass` 데코레이터를 사용하면 `__init__`, `__repr__`, `__eq__`가 자동으로 생성됩니다.

```python
from dataclasses import dataclass, field, asdict, astuple
from typing import List, Optional

# ┌─────────────────────────────────────────┐
# │  @dataclass가 자동 생성하는 메서드         │
# │  __init__  → 초기화                         │
# │  __repr__  → 문자열 표현                    │
# │  __eq__    → 동등 비교 (==)                 │
# └─────────────────────────────────────────┘

@dataclass
class Point:
    x: float
    y: float

p1 = Point(3.0, 4.0)
p2 = Point(3.0, 4.0)
p3 = Point(0.0, 0.0)

print(p1)                        # Point(x=3.0, y=4.0)
print(f"동등 비교: {p1 == p2}")  # True
print(f"다름: {p1 == p3}")       # False

# 데이터클래스는 기본적으로 가변
p1.x = 5.0
print(f"변경 후: {p1}")         # Point(x=5.0, y=4.0)

# 언팩킹
print(f"튜플: {astuple(p1)}")    # (5.0, 4.0)
print(f"딕셔너리: {asdict(p1)}")  # {'x': 5.0, 'y': 4.0}
```

## 2. 기본값과 field() 옵션

`field()`로 기본값, 팩토리, 제외 필드 등을 설정합니다.
가변 기본값(리스트 등)은 `default_factory`를 사용해야 합니다.

```python
# ┌─────────────────────────────────────────┐
# │  field() 주요 옵션                        │
# │  default       → 기본값                     │
# │  default_factory → 가변 기본값 팩토리        │
# │  repr=False    → __repr__에서 제외            │
# │  compare=False → 비교에서 제외                │
# └─────────────────────────────────────────┘

@dataclass
class Product:
    name: str
    price: float
    quantity: int = 0
    tags: List[str] = field(default_factory=list)
    discount: Optional[float] = None

    @property
    def total(self):
        subtotal = self.price * self.quantity
        if self.discount:
            subtotal *= (1 - self.discount)
        return subtotal

p1 = Product("노트북", 1_500_000, 5)
p2 = Product("폰", 800_000, 3, tags=["전자제품", "모바일"])
p3 = Product("태블릿", 500_000, 2, discount=0.1)

print(p1)
print(p2)
print(f"p3 총액: {p3.total:,.0f}원")
```

## 3. frozen 데이터클래스와 상속

`frozen=True`로 불변 데이터클래스를 만들 수 있습니다.
데이터클래스는 상속도 지원합니다.

```python
# frozen 데이터클래스 — 불변 (immutable)
@dataclass(frozen=True)
class Coordinate:
    x: float
    y: float

    def distance_to(self, other):
        return ((self.x - other.x) ** 2 + (self.y - other.y) ** 2) ** 0.5

c1 = Coordinate(0, 0)
c2 = Coordinate(3, 4)
print(f"거리: {c1.distance_to(c2)}")  # 5.0

# 불변이므로 속성 변경 불가
try:
    c1.x = 10  # FrozenInstanceError!
except Exception as e:
    print(f"에러: {type(e).__name__}")

# 데이터클래스 상속
@dataclass
class Employee:
    name: str
    age: int

@dataclass
class Manager(Employee):
    department: str = ""
    team_size: int = 0

m = Manager("김파이", 35, "개발팀", 10)
print(m)  # Manager(name='김파이', age=35, department='개발팀', team_size=10)
```
