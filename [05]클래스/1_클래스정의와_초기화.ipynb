# 📘 클래스 정의와 초기화

파이썬의 **클래스**는 객체를 만드는 설계도입니다.
`class` 키워드로 정의하고, `__init__`으로 초기화합니다.

**학습 목표:**
- 클래스 정의와 인스턴스 생성
- `__init__` 생성자와 `self`
- 인스턴스 속성 vs 클래스 속성
- `__repr__`과 `__str__`

## 1. 클래스 기본 — 정의와 인스턴스

가장 간단한 클래스는 `pass`만 있는 빈 클래스입니다.
인스턴스를 만들고 속성을 동적으로 추가할 수 있습니다.

```python
# 가장 간단한 클래스
class Dog:
    """개를 나타내는 간단한 클래스"""
    pass

# 인스턴스 생성
dog1 = Dog()
dog1.name = "바둑이"
dog1.age = 3
print(f"{dog1.name}는(은) {dog1.age}살입니다")

# 각 인스턴스는 독립적
dog2 = Dog()
dog2.name = "멍멍이"
dog2.age = 5
print(f"{dog2.name}는(은) {dog2.age}살입니다")
```

```python
# ┌─────────────────────────────────────────┐
# │  __init__ 생성자와 self                   │
# │  self → 인스턴스 자신을 가리키는 참조       │
# │  __init__ → 인스턴스 생성 시 자동 호출      │
# └─────────────────────────────────────────┘

class Person:
    """사람 클래스"""
    def __init__(self, name, age):
        self.name = name    # 인스턴스 속성
        self.age = age

    def greet(self):
        return f"안녕하세요, 저는 {self.name}이고 {self.age}살입니다"

p1 = Person("김파이", 25)
p2 = Person("이코딩", 30)

print(p1.greet())
print(p2.greet())

# self는 인스턴스를 가리킴
print(f"p1.name: {p1.name}")  # 김파이
print(f"p2.name: {p2.name}")  # 이코딩
```

## 2. 인스턴스 속성 vs 클래스 속성

**클래스 속성**은 모든 인스턴스가 공유합니다.
**인스턴스 속성**은 각 인스턴스마다 독립적입니다.

```python
# ┌─────────────────────────────────────────┐
# │  클래스 속성 vs 인스턴스 속성               │
# │  클래스 속성 → 클래스 내부에 직접 정의       │
# │  인스턴스 속성 → self.속성 = 값             │
# └─────────────────────────────────────────┘

class Car:
    # 클래스 속성 — 모든 인스턴스가 공유
    wheel_count = 4
    category = "차량"

    def __init__(self, brand, model):
        # 인스턴스 속성 — 각 인스턴스마다 고유
        self.brand = brand
        self.model = model

car1 = Car("현대", "아반떼")
car2 = Car("기아", "K5")

# 클래스 속성은 모든 인스턴스가 공유
print(f"car1 바퀴 수: {car1.wheel_count}")   # 4
print(f"car2 바퀴 수: {car2.wheel_count}")   # 4

# 인스턴스 속성은 각각 다름
print(f"car1: {car1.brand} {car1.model}")  # 현대 아반떼
print(f"car2: {car2.brand} {car2.model}")  # 기아 K5

# 클래스 속성 변경 — 모든 인스턴스에 영향
Car.wheel_count = 6
print(f"car1 바퀴 수: {car1.wheel_count}")   # 6 (공유 속성 변경)
```

```python
# __repr__과 __str__
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    # __repr__: 개발자용 공식 문자열 (명확하게 식별 가능)
    def __repr__(self):
        return f"Product('{self.name}', {self.price})"

    # __str__: 사용자용 읽기 쉬운 문자열
    def __str__(self):
        return f"{self.name} — {self.price:,}원"

p = Product("노트북", 1_500_000)
print(f"repr: {repr(p)}")   # Product('노트북', 1500000)
print(f"str: {p}")           # 노트북 — 1,500,000원

# print()는 __str__을 호출
# 대화형 셸에서는 __repr__을 호출
```
