# 📘 메서드와 프로퍼티

파이썬 클래스에는 **세 가지 메서드**가 있으며, `@property`로
속성처럼 접근하는 메서드를 만들 수 있습니다.

**학습 목표:**
- 인스턴스 메서드, 클래스 메서드, 정적 메서드
- @classmethod로 대체 생성자 만들기
- @property로 속성 접근 제어
- 실용 예제: 날짜 클래스

## 1. 세 가지 메서드

| 종류 | 데코레이터 | 첫 번째 인자 | 호출 방식 |
|------|-----------|-------------|----------|
| 인스턴스 메서드 | (없음) | `self` | `obj.method()` |
| 클래스 메서드 | `@classmethod` | `cls` | `Class.method()` |
| 정적 메서드 | `@staticmethod` | (없음) | `Class.method()` |

```python
# ┌─────────────────────────────────────────┐
# │  인스턴스 / 클래스 / 정적 메서드             │
# │  self  → 인스턴스 속성/메서드에 접근         │
# │  cls   → 클래스 속성/메서드에 접근           │
# │  (없음) → 클래스/인스턴스 모두 독립           │
# └─────────────────────────────────────────┘

class MyClass:
    class_variable = 0

    def __init__(self, value):
        self.instance_variable = value

    def instance_method(self):
        return f"인스턴스 메서드: {self.instance_variable}"

    @classmethod
    def class_method(cls):
        return f"클래스 메서드: {cls.class_variable}"

    @staticmethod
    def static_method(x, y):
        return f"정적 메서드: {x} + {y} = {x + y}"

obj = MyClass(42)
print(obj.instance_method())        # 인스턴스 메서드: 42
print(MyClass.class_method())       # 클래스 메서드: 0
print(MyClass.static_method(3, 5))  # 정적 메서드: 3 + 5 = 8
```

## 2. @classmethod — 대체 생성자

`@classmethod`를 사용하면 클래스의 새로운 생성 방법을 정의할 수 있습니다.

```python
import json

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Person(name='{self.name}', age={self.age})"

    # ┌─────────────────────────────────────────┐
    # │  @classmethod 대체 생성자                  │
    # │  cls를 받아 다른 형식에서 객체 생성         │
    # └─────────────────────────────────────────┘

    @classmethod
    def from_dict(cls, data):
        """딕셔너리에서 Person 생성"""
        return cls(name=data["name"], age=data["age"])

    @classmethod
    def from_string(cls, text):
        """콤마 구분 문자열에서 Person 생성"""
        name, age = text.split(",")
        return cls(name=name.strip(), age=int(age.strip()))

# 일반 생성
p1 = Person("김파이", 25)
print(p1)

# 딕셔너리에서 생성
p2 = Person.from_dict({"name": "이코딩", "age": 30})
print(p2)

# 문자열에서 생성
p3 = Person.from_string("박개발, 28")
print(p3)
```

## 3. @staticmethod — 유틸리티 메서드

`@staticmethod`는 `self`나 `cls`를 받지 않는 메서드입니다.
클래스 네임스페이스 안에 있지만 인스턴스나 클래스와 무관합니다.

```python
class MathUtils:
    # 정적 메서드 — 클래스/인스턴스와 무관한 유틸리티
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def is_even(n):
        return n % 2 == 0

    @staticmethod
    def clamp(value, min_val, max_val):
        return max(min_val, min(value, max_val))

print(f"add(3, 5): {MathUtils.add(3, 5)}")
print(f"is_even(4): {MathUtils.is_even(4)}")
print(f"clamp(15, 0, 10): {MathUtils.clamp(15, 0, 10)}")
```

## 4. @property — 속성 접근 제어

`@property`로 메서드를 속성처럼 사용할 수 있습니다.
getter/setter 로직을 캡슐화합니다.

```python
# ┌─────────────────────────────────────────┐
# │  @property 사용법                          │
# │  @property         → getter (읽기)          │
# │  @속성명.setter     → setter (쓰기)          │
# │  읽기 전용 → setter 없이 @property만 정의     │
# └─────────────────────────────────────────┘

class Circle:
    def __init__(self, radius):
        self.radius = radius  # setter를 통해 검증

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value < 0:
            raise ValueError("반지름은 음수가 될 수 없습니다")
        self._radius = value

    @property
    def area(self):
        """넓이 (읽기 전용 계산 속성)"""
        import math
        return round(math.pi * self._radius ** 2, 2)

    @property
    def circumference(self):
        """둘레 (읽기 전용 계산 속성)"""
        import math
        return round(2 * math.pi * self._radius, 2)

c = Circle(5)
print(f"반지름: {c.radius}")
print(f"넓이: {c.area}")
print(f"둘레: {c.circumference}")

c.radius = 10
print(f"변경 후 넓이: {c.area}")

try:
    c.radius = -1
except ValueError as e:
    print(f"에러: {e}")
```
