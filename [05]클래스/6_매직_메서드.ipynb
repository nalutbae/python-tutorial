# 📘 매직 메서드 (Magic Methods)

**매직 메서드**(dunder methods)는 `__init__`, `__str__`처럼
이중 밑줄로 둘러싸인 특수 메서드입니다. 파이썬의 연산자와
내장 함수가 객체와 상호작용하는 방식을 커스터마이징합니다.

**학습 목표:**
- 객체 생성/소멸: `__init__`, `__repr__`, `__str__`
- 비교 연산자: `__eq__`, `__lt__`, `__le__`
- 산술 연산자: `__add__`, `__mul__`
- 컨테이너: `__len__`, `__getitem__`

## 1. 문자열 표현

`__str__`은 사용자용, `__repr__`은 개발자용 문자열을 반환합니다.
`__format__`은 f-string의 포맷 지정을 커스터마이징합니다.

```python
# ┌─────────────────────────────────────────┐
# │  매직 메서드 주요 종류                      │
# │  __init__   → 객체 초기화                   │
# │  __repr__   → 공식 문자열 (개발자용)        │
# │  __str__    → 읽기 쉬운 문자열 (사용자용)    │
# │  __format__ → f-string 포매팅               │
# └─────────────────────────────────────────┘

class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def __str__(self):
        return f"{self.name} — {self.price:,}원 x {self.quantity}개"

    def __repr__(self):
        return f"Product('{self.name}', {self.price}, {self.quantity})"

    def __format__(self, format_spec):
        if format_spec == "short":
            return f"{self.name}: {self.price:,}원"
        elif format_spec == "detail":
            return f"{self.name} | 가격: {self.price:,}원 | 수량: {self.quantity}개"
        return str(self)

p = Product("노트북", 1_500_000, 5)
print(f"str: {str(p)}")
print(f"repr: {repr(p)}")
print(f"short: {p:short}")
print(f"detail: {p:detail}")
```

## 2. 비교 연산자

`__eq__`(==), `__lt__`(<), `__le__`(<=) 등을 구현하면
객체 간 비교가 가능합니다.

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def __eq__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.score == other.score

    def __lt__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.score < other.score

    def __le__(self, other):
        return self == other or self < other

    def __repr__(self):
        return f"Student('{self.name}', {self.score})"

s1 = Student("김파이", 85)
s2 = Student("이코딩", 92)
s3 = Student("박개발", 85)

print(f"s1 == s3: {s1 == s3}")  # True (같은 점수)
print(f"s1 < s2: {s1 < s2}")    # True (85 < 92)
print(f"s1 <= s2: {s1 <= s2}")  # True

# 정렬 가능!
students = [s2, s1, s3]
students.sort()
print(f"정렬 후: {students}")
```

## 3. 컨테이너 프로토콜

`__len__`, `__getitem__`, `__contains__` 등을 구현하면
커스텀 객체에 `len()`, 인덱싱, `in` 연산자를 사용할 수 있습니다.

```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add(self, song):
        self.songs.append(song)

    def __len__(self):
        return len(self.songs)

    def __getitem__(self, index):
        return self.songs[index]

    def __contains__(self, song):
        return song in self.songs

    def __repr__(self):
        return f"Playlist('{self.name}', {len(self)}곡)"

playlist = Playlist("내 플레이리스트")
playlist.add("Dynamite")
playlist.add("Butter")
playlist.add("Permission to Dance")

print(playlist)                      # Playlist('내 플레이리스트', 3곡)
print(f"곡 수: {len(playlist)}")      # 3
print(f"첫 곡: {playlist[0]}")        # Dynamite
print(f"마지막 곡: {playlist[-1]}")    # Permission to Dance

# in 연산자
print(f"'Butter' 포함: {'Butter' in playlist}")  # True
print(f"'DNA' 포함: {'DNA' in playlist}")         # False

# 반복 가능!
for song in playlist:
    print(f"  🎵 {song}")
```
