# 정답 확인

아래 셀을 실행하여 각 문제의 정답을 확인하세요.


```python
# ===== 문제 1 정답 =====
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def greet(self):
        return f'안녕하세요, 저는 {self.name}입니다.'

    def is_passed(self):
        return self.score >= 60

s = Student('김민수', 85)
assert s.greet() == '안녕하세요, 저는 김민수입니다.'
assert s.is_passed() is True
assert Student('이영', 30).is_passed() is False
print('문제 1 ✅')

# ===== 문제 2 정답 =====
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def speak(self):
        return f'{self.name}가 멍멍합니다!'

d = Dog('바둑이', '진돗개')
assert d.speak() == '바둑이가 멍멍합니다!'
assert d.breed == '진돗개'
print('문제 2 ✅')

# ===== 문제 3 정답 =====
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value < 0:
            raise ValueError('반지름은 음수일 수 없습니다.')
        self._radius = value

    @property
    def area(self):
        return math.pi * self._radius ** 2

c = Circle(5)
assert c.radius == 5
assert abs(c.area - math.pi * 25) < 1e-6
c.radius = 3
assert abs(c.area - math.pi * 9) < 1e-6
try:
    c.radius = -1
except ValueError:
    pass
else:
    raise AssertionError('음수 반지름이 허용됨')
print('문제 3 ✅')

# ===== 문제 4 정답 =====
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f'{self.title} — {self.author}'

    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', {self.pages})"

b = Book('데미안', '헤르만 헤세', 256)
assert str(b) == '데미안 — 헤르만 헤세'
assert repr(b) == "Book('데미안', '헤르만 헤세', 256)"
print('문제 4 ✅')

# ===== 문제 5 정답 =====
@dataclass
class Point:
    x: float
    y: float

    def __post_init__(self):
        if not isinstance(self.x, (int, float)) or not isinstance(self.y, (int, float)):
            raise TypeError('x와 y는 숫자여야 합니다.')

    def distance_to(self, other: 'Point') -> float:
        return math.sqrt((self.x - other.x) ** 2 + (self.y - other.y) ** 2)

p1 = Point(0, 0)
p2 = Point(3, 4)
assert p1.distance_to(p2) == 5.0
try:
    Point('a', 0)
except TypeError:
    pass
else:
    raise AssertionError('문자 허용됨')
print('문제 5 ✅')

# ===== 문제 6 정답 =====
class Temperature:
    def __init__(self, celsius: float):
        self.celsius = celsius

    @classmethod
    def from_fahrenheit(cls, f):
        c = (f - 32) * 5 / 9
        return cls(c)

    @staticmethod
    def is_freezing(c):
        return c < 0

t1 = Temperature(0)
t2 = Temperature.from_fahrenheit(32)
assert abs(t2.celsius) < 1e-9
assert Temperature.is_freezing(-5) is True
assert Temperature.is_freezing(10) is False
print('문제 6 ✅')

# ===== 문제 7 정답 =====
class Money:
    def __init__(self, amount: int, currency: str):
        self.amount = amount
        self.currency = currency

    def __eq__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        if self.currency != other.currency:
            raise ValueError('다른 통화는 비교할 수 없습니다.')
        return self.amount == other.amount

    def __repr__(self):
        return f"Money({self.amount}, '{self.currency}')"

a = Money(1000, 'KRW')
b = Money(1000, 'KRW')
c = Money(1000, 'USD')
assert a == b
assert not (a != b)
try:
    a == c
except ValueError:
    pass
else:
    raise AssertionError('다른 통화 비교가 허용됨')
print('문제 7 ✅')

# ===== 문제 8 정답 =====
class BankAccount:
    def __init__(self, owner, balance=0):
        self._owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError('입금액은 양수여야 합니다.')
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError('출금액은 양수여야 합니다.')
        if amount > self.__balance:
            raise ValueError('잔액이 부족합니다.')
        self.__balance -= amount

    def get_balance(self):
        return self.__balance

acc = BankAccount('홍길동', 1000)
acc.deposit(500)
assert acc.get_balance() == 1500
acc.withdraw(200)
assert acc.get_balance() == 1300
try:
    acc.withdraw(9999)
except ValueError:
    pass
else:
    raise AssertionError('잔액 초과 출금 허용됨')
print('문제 8 ✅')

# ===== 문제 9 정답 =====
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

class Triangle(Shape):
    def __init__(self, base, height_tri, side1, side2):
        self.base = base
        self.height_tri = height_tri
        self.side1 = side1
        self.side2 = side2

    def area(self):
        return self.base * self.height_tri / 2

    def perimeter(self):
        return self.base + self.side1 + self.side2

r = Rectangle(4, 5)
assert r.area() == 20
assert r.perimeter() == 18
t = Triangle(6, 4, 5, 5)
assert t.area() == 12.0
assert t.perimeter() == 16
try:
    Shape()
except TypeError:
    pass
else:
    raise AssertionError('ABC 인스턴스화 허용됨')
print('문제 9 ✅')

# ===== 문제 10 정답 =====
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f'{self.horsepower}마력 엔진 시동!'

class Car:
    def __init__(self, brand, engine):
        self.brand = brand
        self.engine = engine

    def drive(self):
        return f'{self.brand} 출발: {self.engine.start()}'

e = Engine(150)
c = Car('현대', e)
assert c.drive() == '현대 출발: 150마력 엔진 시동!'
assert not isinstance(c, Engine)
print('문제 10 ✅')

print('\n🎉 모든 문제 정답 확인 완료!')
```
