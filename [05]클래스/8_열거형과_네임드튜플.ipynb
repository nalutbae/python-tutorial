# 📘 열거형(Enum)과 네임드튜플(NamedTuple)

**Enum**은 이름 붙은 상수 집합을 만들고, **NamedTuple**은
이름 붙은 튜플로 읽기 쉬운 데이터 구조를 만듭니다.

**학습 목표:**
- Enum 기본 사용법 (이름, 값, 반복)
- IntEnum, auto()
- NamedTuple 정의와 사용
- Enum vs dataclass vs NamedTuple 비교

## 1. Enum 기본

`Enum`은 관련된 상수들을 이름으로 그룹화합니다.
매직 넘버를 피하고 코드를 읽기 쉽게 만듭니다.

```python
from enum import Enum, IntEnum, auto

# ┌─────────────────────────────────────────┐
# │  Enum 기본 사용법                         │
# │  class Color(Enum):                       │
# │      RED = "빨강"    ← 이름=값             │
# │  Color.RED           ← 멤버 접근          │
# │  Color.RED.name      ← "RED" (이름)      │
# │  Color.RED.value     ← "빨강" (값)        │
# └─────────────────────────────────────────┘

class Color(Enum):
    RED = "빨강"
    GREEN = "초록"
    BLUE = "파랑"

# 멤버 접근
print(Color.RED)         # Color.RED
print(Color.RED.name)    # RED
print(Color.RED.value)   # 빨강

# 이름으로 접근
print(Color["RED"])      # Color.RED

# 값으로 접근
print(Color("빨강"))     # Color.RED

# 반복
for color in Color:
    print(f"  {color.name} = {color.value}")

# 비교
print(f"Color.RED == Color.RED: {Color.RED == Color.RED}")   # True
print(f"Color.RED == Color.BLUE: {Color.RED == Color.BLUE}")  # False
```

## 2. IntEnum과 auto()

`IntEnum`은 정수 값을 가지는 Enum입니다. 산술 연산과 비교가 가능합니다.
`auto()`를 사용하면 값을 자동으로 할당할 수 있습니다.

```python
# IntEnum — 정수 값 Enum (산술 연산 가능)
class Priority(IntEnum):
    LOW = 1
    MEDIUM = 2
    HIGH = 3
    CRITICAL = 4

print(f"LOW < HIGH: {Priority.LOW < Priority.HIGH}")  # True
print(f"LOW + 1: {Priority.LOW + 1}")                  # Priority.MEDIUM

# auto() — 값 자동 할당
class Status(Enum):
    PENDING = auto()
    ACTIVE = auto()
    COMPLETED = auto()
    FAILED = auto()

for status in Status:
    print(f"  {status.name} = {status.value}")

# Enum을 조건문에 사용
def describe_priority(priority):
    if priority >= Priority.HIGH:
        return "⚠️ 높은 우선순위"
    elif priority >= Priority.MEDIUM:
        return "📋 보통 우선순위"
    else:
        return "✅ 낮은 우선순위"

print(describe_priority(Priority.CRITICAL))
print(describe_priority(Priority.MEDIUM))
print(describe_priority(Priority.LOW))
```

## 3. NamedTuple — 이름 붙은 튜플

`NamedTuple`은 튜플에 이름을 붙여 읽기 쉽게 만듭니다.
인덱스 접근과 이름 접근이 모두 가능합니다.

```python
from typing import NamedTuple

# ┌─────────────────────────────────────────┐
# │  NamedTuple 정의                           │
# │  class Point(NamedTuple):                  │
# │      x: float                              │
# │      y: float                              │
# │  p = Point(3.0, 4.0)                       │
# │  p.x  → 3.0  (이름 접근)                    │
# │  p[0] → 3.0  (인덱스 접근)                  │
# └─────────────────────────────────────────┘

class Point(NamedTuple):
    x: float
    y: float

class Student(NamedTuple):
    name: str
    age: int
    grade: float

p = Point(3.0, 4.0)
print(f"이름 접근: p.x={p.x}, p.y={p.y}")
print(f"인덱스 접근: p[0]={p[0]}, p[1]={p[1]}")

s = Student("김파이", 25, 4.3)
print(f"학생: {s.name}, {s.age}세, 학점: {s.grade}")

# NamedTuple은 불변
# s.age = 26  # AttributeError!

# 딕셔너리로 변환
print(f"딕셔너리: {s._asdict()}")
```

## 4. Enum vs dataclass vs NamedTuple 비교

| 특징 | Enum | dataclass | NamedTuple |
|------|------|-----------|------------|
| 용도 | 이름 붙은 상수 | 데이터 저장 | 경량 데이터 |
| 가변성 | 불변 | 가변 (frozen=True 시 불변) | 불변 |
| 타입 힌트 | X | O | O |
| 메서드 | O | O | 제한적 |
| 상속 | Enum 전용 | 자유 | NamedTuple 전용 |

```python
from dataclasses import dataclass
from enum import Enum
from typing import NamedTuple

# 세 가지 방식 비교 — 색상 표현

# Enum: 고정된 상수 집합
class Color(Enum):
    RED = "#FF0000"
    GREEN = "#00FF00"
    BLUE = "#0000FF"

# dataclass: 가변 데이터
@dataclass
class ColorInfo:
    name: str
    hex_code: str
    opacity: float = 1.0

# NamedTuple: 불변 경량 데이터
class ColorTuple(NamedTuple):
    name: str
    hex_code: str

# 사용 비교
c = Color.RED
print(f"Enum: {c.name} = {c.value}")

ci = ColorInfo("빨강", "#FF0000", 0.8)
print(f"dataclass: {ci}")

ci.opacity = 0.5  # 가변
print(f"변경 후: {ci}")

ct = ColorTuple("초록", "#00FF00")
print(f"NamedTuple: {ct}")
# ct.name = "파랑"  # AttributeError! 불변
```
