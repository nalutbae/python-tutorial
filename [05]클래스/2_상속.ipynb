# 📘 상속 (Inheritance)

**상속**은 부모 클래스의 속성과 메서드를 자식 클래스가 물려받는 기능입니다.
코드 재사용과 계층 구조를 만드는 핵심 개념입니다.

**학습 목표:**
- 기본 상속과 super()
- 메서드 오버라이딩
- 다중 상속과 MRO
- isinstance()와 issubclass()

## 1. 기본 상속과 super()

자식 클래스는 부모 클래스의 모든 속성과 메서드를 상속받습니다.
`super()`로 부모의 메서드를 호출할 수 있습니다.

```python
# ┌─────────────────────────────────────────┐
# │  상속 기본 구조                             │
# │  class 자식(부모):                          │
# │      def __init__(self, ...):               │
# │          super().__init__(...)  ← 부모 초기화│
# └─────────────────────────────────────────┘

class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def speak(self):
        return f"{self.name}이(가) {self.sound}!"

    def info(self):
        return f"{self.name} (소리: {self.sound})"

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, "멍멍")  # 부모의 __init__ 호출
        self.breed = breed

class Cat(Animal):
    def __init__(self, name, indoor=True):
        super().__init__(name, "야옹")
        self.indoor = indoor

dog = Dog("바둑이", "골든 리트리버")
cat = Cat("나비")

print(dog.speak())   # 바둑이이(가) 멍멍!
print(cat.speak())   # 나비이(가) 야옹!
print(f"개 품종: {dog.breed}")
print(f"실내 고양이: {cat.indoor}")
```

```python
# super()로 부모 메서드 호출
class Vehicle:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year

    def info(self):
        return f"{self.brand} ({self.year})"

class ElectricCar(Vehicle):
    def __init__(self, brand, year, battery_kwh):
        super().__init__(brand, year)  # 부모 __init__ 호출
        self.battery_kwh = battery_kwh

    def info(self):
        base_info = super().info()  # 부모 메서드 호출
        return f"{base_info}, 배터리: {self.battery_kwh}kWh"

car = ElectricCar("테슬라", 2024, 75)
print(car.info())  # 테슬라 (2024), 배터리: 75kWh
```

## 2. 메서드 오버라이딩

자식 클래스에서 부모의 메서드를 재정의(오버라이딩)할 수 있습니다.
`super().메서드()`로 부모의 원래 메서드를 호출할 수도 있습니다.

```python
# 메서드 오버라이딩 예시
class Shape:
    def __init__(self, name):
        self.name = name

    def area(self):
        return 0

    def __str__(self):
        return f"{self.name}: 넓이 = {self.area()}"

class Circle(Shape):
    def __init__(self, radius):
        super().__init__("원")
        self.radius = radius

    def area(self):
        import math
        return round(math.pi * self.radius ** 2, 2)

class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__("사각형")
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

shapes = [Circle(5), Rectangle(4, 6)]
for shape in shapes:
    print(shape)
# 원: 넓이 = 78.54
# 사각형: 넓이 = 24
```

## 3. 다중 상속과 MRO

파이썬은 다중 상속을 지원합니다.
MRO(Method Resolution Order)는 메서드 검색 순서를 결정합니다.

```python
# ┌─────────────────────────────────────────┐
# │  다중 상속과 MRO (Method Resolution Order) │
# │  메서드 검색 순서: 자식 → 왼쪽 부모 → 오른쪽  │
# │  확인: ClassName.__mro__ 또는 ClassName.mro()│
# └─────────────────────────────────────────┘

class A:
    def greet(self):
        return "A에서 인사"

class B(A):
    def greet(self):
        return "B에서 인사"

class C(A):
    def greet(self):
        return "C에서 인사"

class D(B, C):
    pass

d = D()
print(d.greet())          # B에서 인사 (MRO에 따라 B가 먼저)
print(D.__mro__)
# (D, B, C, A, object)

# isinstance()와 issubclass()
print(f"isinstance(d, D): {isinstance(d, D)}")     # True
print(f"isinstance(d, A): {isinstance(d, A)}")     # True
print(f"issubclass(D, B): {issubclass(D, B)}")     # True
print(f"issubclass(D, A): {issubclass(D, A)}")")  # True
```
