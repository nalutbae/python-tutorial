# 📘 다형성 (Polymorphism)

**다형성**은 같은 메서드 이름이 다른 동작을 수행하는 개념입니다.
파이썬에서는 상속 기반 다형성과 **덕 타이핑**을 모두 지원합니다.

**학습 목표:**
- 상속 기반 다형성
- 덕 타이핑 (Duck Typing)
- 연산자 오버로딩
- 추상 기반 클래스 (ABC)

## 1. 상속 기반 다형성

부모 클래스의 메서드를 자식 클래스에서 각자 다르게 구현합니다.
같은 메서드 호출이 객체에 따라 다르게 동작합니다.

```python
# ┌─────────────────────────────────────────┐
# │  다형성 — 같은 메서드, 다른 동작            │
# │  부모 클래스의 메서드를 자식이 재정의         │
# │  동일한 인터페이스로 다양한 객체 처리          │
# └─────────────────────────────────────────┘

class Animal:
    def speak(self):
        raise NotImplementedError("서브클래스에서 구현해야 합니다")

class Dog(Animal):
    def speak(self):
        return "멍멍!"

class Cat(Animal):
    def speak(self):
        return "야옹!"

class Duck(Animal):
    def speak(self):
        return "꽥!"

# 다형적 함수 — 어떤 Animal 서브클래스든 동작
def make_sound(animal):
    print(f"{animal.__class__.__name__}: {animal.speak()}")

animals = [Dog(), Cat(), Duck()]
for animal in animals:
    make_sound(animal)
```

## 2. 덕 타이핑

"오리처럼 걷고 오리처럼 꽥꽥거리면 오리다."
파이썬은 상속 없이도 같은 메서드를 가진 객체를 동일하게 처리합니다.

```python
# 덕 타이핑 — 공통 기반 클래스 없이 다형성 구현

class PDFDocument:
    def read(self):
        return "PDF 문서를 읽습니다"

class TextDocument:
    def read(self):
        return "텍스트 문서를 읽습니다"

class SpreadsheetDocument:
    def read(self):
        return "스프레드시트를 읽습니다"

# 공통 기반 클래스 없이 — read() 메서드만 있으면 됨
def open_document(doc):
    return doc.read()

pdf = PDFDocument()
txt = TextDocument()
xls = SpreadsheetDocument()

print(open_document(pdf))
print(open_document(txt))
print(open_document(xls))
```

## 3. 연산자 오버로딩

매직 메서드를 구현하면 사용자 정의 객체에 `+`, `<` 등의 연산자를 사용할 수 있습니다.

```python
# 연산자 오버로딩 예시 — Vector 클래스
import math

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __abs__(self):
        return math.sqrt(self.x ** 2 + self.y ** 2)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(f"v1 + v2 = {v1 + v2}")
print(f"v1 * 2 = {v1 * 2}")
print(f"|v1| = {abs(v1):.2f}")
print(f"v1 == Vector(3,4): {v1 == Vector(3, 4)}")
```

## 4. 추상 기반 클래스 (ABC)

`ABC`(Abstract Base Class)는 서브클래스가 반드시 구현해야 하는 메서드를 정의합니다.

```python
from abc import ABC, abstractmethod

# ┌─────────────────────────────────────────┐
# │  추상 기반 클래스 (ABC)                     │
# │  @abstractmethod → 서브클래스에서 반드시 구현  │
# │  추상 클래스는 인스턴스 생성 불가             │
# └─────────────────────────────────────────┘

class Shape(ABC):
    @abstractmethod
    def area(self):
        """도형의 넓이를 반환"""
        pass

    @abstractmethod
    def perimeter(self):
        """도형의 둘레를 반환"""
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        import math
        return round(math.pi * self.radius ** 2, 2)

    def perimeter(self):
        import math
        return round(2 * math.pi * self.radius, 2)

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

shapes = [Circle(5), Rectangle(4, 6)]
for shape in shapes:
    name = shape.__class__.__name__
    print(f"{name}: 넓이={shape.area()}, 둘레={shape.perimeter()}")

# 추상 클래스는 인스턴스 생성 불가
# shape = Shape()  # TypeError!
```
