## 2. Tasks API 구조 이해

MediaPipe 1.0의 핵심은 **Tasks API**입니다.

```
mp.tasks.BaseOptions (모델 파일 경로 지정)
    ↓
mp.tasks.vision.{TaskName}Options (태스크별 옵션)
    ↓
mp.tasks.vision.{TaskName}.create_from_options() (태스크 객체 생성)
    ↓
task.detect(image) / task.detect_for_video(image, timestamp) (추론 실행)
```

## 4. 테스트 이미지 생성

MediaPipe는 웹캠뿐 아니라 정적 이미지에도 사용할 수 있습니다.
이 튜토리얼에서는 NumPy로 테스트 이미지를 생성하여 사용합니다.
(웹캠 코드는 주석 처리하여 실행 환경에 맞게 사용하세요)

````python
# 테스트 이미지 생성 함수 (이후 노트북에서 재사용)
import numpy as np

def create_test_image(width=640, height=480):
    """단색 테스트 이미지 생성 (OpenCV BGR 형식)"""
    return np.zeros((height, width, 3), dtype=np.uint8)

def create_gradient_image(width=640, height=480):
    """그라디언트 테스트 이미지 생성"""
    img = np.zeros((height, width, 3), dtype=np.uint8)
    for i in range(height):
        img[i, :] = [i % 256, (i * 2) % 256, (i * 3) % 256]
    return img

# MediaPipe Image 객체로 변환하는 헬퍼
import mediapipe as mp

def numpy_to_mp_image(np_image):
    """NumPy 배열을 MediaPipe Image로 변환"""
    return mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=np_image
    )

# 테스트
test_img = create_gradient_image()
mp_img = numpy_to_mp_image(test_img)
print(f'NumPy 이미지: shape={test_img.shape}, dtype={test_img.dtype}')
print(f'MediaPipe Image: {type(mp_img).__name__}')
print(f'MP Image 크기: {mp_img.width}x{mp_img.height}')
````

## 🎯 정리

| 개념 | 설명 |

| MediaPipe | Google의 실시간 ML 비전 라이브러리 |
